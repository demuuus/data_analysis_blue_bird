# <span style="color:#0061AF">Blue Bird — Krisis Ritase &amp; Kesenjangan Armada</span>

# <span style="color:#0061AF">Business Understanding</span>

## <span style="color:#0061AF">Latar Belakang</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Blue Bird adalah pionir taksi konvensional dengan armada fisik dan pengemudi tetap, bukan model <em>ride-hailing</em> murni. Mobil dan pengemudi adalah <strong>biaya tetap</strong>, sehingga setiap jam pengemudi bertugas tanpa penumpang adalah biaya tanpa pendapatan. Karena itu utilisasi armada menjadi metrik acuan manajemen (brief mencontohkan target sekitar 70% pada 2018, sebagai konteks sejarah dan bukan pembanding dalam analisis ini). Ekspansi kompetitor berbasis aplikasi membuat fokus bergeser dari <em>menambah</em> armada menjadi <em>memaksimalkan produktivitas</em> armada yang ada, termasuk ritase dan idle time pengemudi per garasi/pool.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Blue Bird juga melayani segmen B2B melalui <em>Corporate Solution</em>, dengan pembayaran non-tunai <strong>eVoucher</strong> untuk perjalanan dinas karyawan klien. Tim Finance beberapa klien mempertanyakan transparansi pemakaiannya, sehingga kewajaran pemakaian perlu dikaji. Dari sisi data, tim Data Engineering mencatat inkonsistensi akibat rekonsiliasi antara sistem aplikasi dan sistem dispatch lama (log argometer dan kategori layanan), sehingga kualitas data harus dipastikan lebih dulu (Part 1).
</p>

## <span style="color:#0061AF">Pernyataan Masalah &amp; Tujuan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Brief merumuskan masalah utama: produktivitas armada (ritase) <em>diduga</em> menurun terkait idle time pengemudi, dan ada indikasi masalah kontrol pada eVoucher korporat. Brief menyebut tiga indikasi yang besarnya <strong>masih harus dikonfirmasi</strong> lewat data:
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>1. Ritase dan idle time.</strong> Ada indikasi variasi ritase (trip per pengemudi per hari) antar pool, yang diduga terkait idle time. Besar dan polanya diukur di 2.5 dan 2.7.<br><strong>2. Permintaan tidak terpenuhi.</strong> Ada indikasi request pelanggan tidak selalu terpenuhi pada waktu atau lokasi tertentu (dibatalkan karena tidak ada pengemudi atau <em>timeout</em>). Dipetakan di 2.4.<br><strong>3. Pemakaian eVoucher.</strong> Ada indikasi trip eVoucher korporat terjadi pada waktu yang tidak lazim untuk perjalanan dinas. Diperiksa di 2.6.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Tujuan analisis</strong> ada dua: (1) <strong>Fleet Optimization</strong>, yaitu memahami produktivitas dan idle time antar pool untuk mendukung strategi penempatan/dispatch pengemudi; (2) <strong>B2B Fraud Mitigation</strong>, yaitu memahami pola eVoucher untuk memperkuat kontrol dan transparansi pelaporan ke klien B2B. Keandalan data argometer dan standardisasi kategori layanan adalah prasyarat kedua tujuan itu, sehingga dikerjakan lebih dulu. Brief meminta baseline, besaran gap, dan target terukur dirumuskan sendiri dari data (lihat Rekomendasi). Data mencakup Oktober–Desember 2023 dari enam tabel: <code>bb_trips</code>, <code>bb_service_requests</code>, <code>bb_driver_shifts</code>, <code>bb_drivers</code>, <code>bb_pools</code>, dan <code>bb_corporate_clients</code>.
</p>

### <span style="color:#0061AF">Ringkasan Eksekutif</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Jawaban singkat atas masalah utama brief (rincian dan bukti ada di bagian yang disebut):
</p>

1. **Ritase tidak menurun, tetapi rendah karena kelebihan kapasitas.** Rata-rata 1,26 trip per shift dan utilisasi 9,4%, tanpa tren dalam 12 minggu. Permintaan yang tercatat hanya membutuhkan sekitar 18,9% jam-pengemudi yang ada jika pengemudi sibuk 50%, dan shift yang lebih panjang tidak menghasilkan trip tambahan (2.5.10, 2.5.11, 2.5.13).
2. **Perbedaan antar pool hanya mencerminkan kuota zona kerja yang ditetapkan per pool (2.5.14).** Setelah zona diperhitungkan, pool tidak berbeda (selisih paling besar 0,03 trip; F-test p = 0,62). Pengemudi Airport paling idle (2.5.5, 2.7.3).
3. **Request gagal terkonsentrasi di CBD pada 07–08 dan 17–18** (36,6% vs 5,5% pada jam lain), terkonfirmasi di data Desember. Puncak sore adalah celah pergantian shift; puncak pagi bukan kekurangan pengemudi dan perlu audit dispatch, terutama kanal App yang gagal lebih sering di setiap zona (2.4.4, 2.4.8, 2.5.1, 2.7.3).
4. **eVoucher: satu pola nyata, belum ada bukti kecurangan.** 51,4% trip eVoucher Airport terjadi di akhir pekan (vs 17,6%), seragam di 200 perusahaan, sehingga yang dibutuhkan adalah aturan tertulis; 1.290 trip diprioritaskan untuk tinjauan sampel (2.6.5, 2.6.7).
5. **Data cukup andal untuk perbandingan relatif.** 7.091 trip (2,2%) dikeluarkan tanpa membuat data condong, tetapi skala dan keteraturan data menunjukkan kemungkinan sampel atau simulasi, sehingga angka absolut perlu dikonfirmasi (1.8.5, 2.2.3).


# <span style="color:#0061AF">PART 1 — Data Cleaning</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Setiap tabel dibersihkan satu per satu sebelum dianalisis. Prinsipnya: data asli tidak diubah, masalah ditandai dengan kolom <code>flag_*</code>, dan versi aman dari sebuah nilai disimpan di kolom <code>*_clean</code>. Baris trip hanya dikeluarkan jika melanggar aturan yang sudah dikonfirmasi (jarak negatif atau delay booking-ke-meter negatif); selain itu tidak ada baris yang dihapus.
</p>

## <span style="color:#0061AF">1.1 Ruang Lingkup Proyek dan Sumber Data</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Sebelum dianalisis, data harus dipastikan andal. Enam tabel sumber dibersihkan satu per satu sebelum digabung, karena setiap tabel punya aturan dan key sendiri, dan penggabungan yang tergesa-gesa dapat menggandakan baris. Hubungan antar tabel: <code>bb_pools.pool_id</code> → <code>bb_drivers.pool_id</code>; <code>bb_drivers.driver_id</code> → <code>bb_driver_shifts.driver_id</code> dan <code>bb_trips.driver_id</code>; <code>bb_driver_shifts.shift_id</code> → <code>bb_trips.shift_id</code>; <code>bb_corporate_clients.company_id</code> → <code>bb_trips.company_id</code>; <code>bb_trips.trip_id</code> → <code>bb_service_requests.matched_trip_id</code>.

| Tabel | Isi | Primary key |
|---|---|---|
| <code>bb_trips</code> | Trip selesai: waktu, jarak, tarif, layanan, pembayaran | <code>trip_id</code> |
| <code>bb_service_requests</code> | Request pelanggan dan statusnya (terpenuhi / dibatalkan) | <code>request_id</code> |
| <code>bb_driver_shifts</code> | Shift pengemudi: clock-in/out, jumlah trip, idle time | <code>shift_id</code> |
| <code>bb_drivers</code> | Atribut pengemudi | <code>driver_id</code> |
| <code>bb_pools</code> | Pool (garasi): wilayah dan afinitas zona | <code>pool_id</code> |
| <code>bb_corporate_clients</code> | Klien korporat dan industri | <code>company_id</code> |
</p>

### <span style="color:#0061AF">1.1.1 Import Library dan Folder Output</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Pada tahap ini library dan folder output disiapkan. Dataset disimpan di folder <code>data/</code> (subfolder <code>raw/</code> dan <code>cleaned/</code>), sedangkan hasil analisis disimpan di folder <code>outputs/</code> (subfolder <code>exceptions/</code>, <code>reports/</code>, dan <code>charts/</code>). Data sumber di database tidak pernah ditimpa.
</p>

```python
import os
from pathlib import Path

import numpy as np
import pandas as pd
import pymysql
from dotenv import load_dotenv
from IPython.display import display

load_dotenv()

# Datasets go to data/, analysis results go to outputs/. The source data (MySQL tables) is never written to.
data_dir = Path("data")
raw_dir = data_dir / "raw"                  # RAW: tables exactly as loaded, before cleaning (PART 4)
cleaned_dir = data_dir / "cleaned"          # CLEANED: final tables used in the analysis (PART 4)
output_dir = Path("outputs")
exceptions_dir = output_dir / "exceptions"  # EXCEPTION: flagged/excluded records and review lists
reports_dir = output_dir / "reports"        # REPORT: quality reports and the statistical test summary
for folder in [data_dir, raw_dir, cleaned_dir, output_dir, exceptions_dir, reports_dir]:
    folder.mkdir(parents=True, exist_ok=True)

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:,.2f}".format)

print(f"Datasets (RAW, CLEANED): {data_dir.as_posix()}/ | Results (exceptions, reports, charts): {output_dir.as_posix()}/")
```

## <span style="color:#0061AF">1.2 Load dan Inspeksi Data</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Pertanyaan pertama: apakah keenam tabel termuat lengkap dan bagaimana bentuknya? Tabel dimuat dari database, disalin sebagai data asli (<code>*_raw</code>), lalu jumlah baris, kolom, dan tipe datanya diperiksa.
</p>

### <span style="color:#0061AF">1.2.1 Load Enam Tabel Sumber</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Keenam tabel dimuat dari database. Kredensial dibaca dari environment variable, bukan ditulis di notebook.
</p>

```python
# Credentials should be stored in environment variables or a local .env file.
# Do not hard-code database passwords in the notebook.
required_env = ["USER", "PASS", "DATABASE"]
missing_env = [name for name in required_env if not os.getenv(name)]
if missing_env:
    raise EnvironmentError(
        "Missing environment variables: " + ", ".join(missing_env)
    )

conn = pymysql.connect(
    host="vmlp.pwdlearning.cloud",
    user=os.getenv("USER"),
    password=os.getenv("PASS"),
    database=os.getenv("DATABASE")
)

try:
    df_clients_raw = pd.read_sql_query(
        "SELECT * FROM bb_corporate_clients", conn
    )
    df_trips_raw = pd.read_sql_query(
        "SELECT * FROM bb_trips", conn
    )
    df_shifts_raw = pd.read_sql_query(
        "SELECT * FROM bb_driver_shifts", conn
    )
    df_drivers_raw = pd.read_sql_query(
        "SELECT * FROM bb_drivers", conn
    )
    df_pools_raw = pd.read_sql_query(
        "SELECT * FROM bb_pools", conn
    )
    df_requests_raw = pd.read_sql_query(
        "SELECT * FROM bb_service_requests", conn
    )
finally:
    conn.close()

print("Loaded all six source tables.")
```

### <span style="color:#0061AF">1.2.2 Salinan Kerja dan Profil Tabel</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Tabel asli (<code>*_raw</code>) disimpan tanpa perubahan. Salinan kerja dibuat untuk dibersihkan, lalu dilihat jumlah baris, kolom, dan tipe datanya.
</p>

```python
# Raw DataFrames are retained unchanged as a source backup in memory.
df_clients = df_clients_raw.copy()
df_trips = df_trips_raw.copy()
df_shifts = df_shifts_raw.copy()
df_drivers = df_drivers_raw.copy()
df_pools = df_pools_raw.copy()
df_requests = df_requests_raw.copy()

# Fingerprint of every raw table right after loading (a number that changes if ANY value changes).
# PART 4 recomputes it to prove the raw exports are the untouched source data.
raw_fingerprints = {
    name: int(pd.util.hash_pandas_object(df, index=True).sum())
    for name, df in {
        "bb_corporate_clients": df_clients_raw, "bb_trips": df_trips_raw, "bb_driver_shifts": df_shifts_raw,
        "bb_drivers": df_drivers_raw, "bb_pools": df_pools_raw, "bb_service_requests": df_requests_raw,
    }.items()
}

def make_tables():
    return {
        "clients": df_clients,
        "trips": df_trips,
        "shifts": df_shifts,
        "drivers": df_drivers,
        "pools": df_pools,
        "requests": df_requests,
    }

tables = make_tables()

profile_rows = []
for table_name, df in tables.items():
    profile_rows.append({
        "table": table_name,
        "rows": len(df),
        "columns": len(df.columns),
        "exact_duplicate_rows": int(df.duplicated().sum()),
    })

display(pd.DataFrame(profile_rows))
```

```python
# Review each table's column names and data types.
for table_name, df in tables.items():
    print(f"\n{table_name.upper()} — first 3 rows")
    display(df.head(3))
    display(df.dtypes.rename("data_type").to_frame())
```

## <span style="color:#0061AF">1.3 Penilaian Awal Kualitas Data</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Setelah tabel termuat, perlu diketahui seberapa banyak nilai kosong sebelum ada yang diubah. Laporan ini menjadi garis dasar untuk dibandingkan dengan hasil akhir.
</p>

### <span style="color:#0061AF">1.3.1 Laporan Missing Value Awal</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Jumlah nilai kosong dihitung per kolom sebagai titik awal sebelum pembersihan.
</p>

```python
def missing_summary(df):
    return pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_percentage": (df.isna().mean() * 100).round(2),
    }).sort_values("missing_percentage", ascending=False)

for table_name, df in tables.items():
    print(f"\nMISSING VALUES: {table_name.upper()}")
    display(missing_summary(df))
```

## <span style="color:#0061AF">1.4 Standarisasi Kategori dan Konversi Tipe Data</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Label dan tipe data yang tidak konsisten membuat validasi berikutnya menyesatkan (misalnya urutan waktu tidak bisa dicek jika kolom masih berupa teks). Karena itu teks dirapikan, layanan dipetakan, dan tipe data diubah lebih dulu. Label asli selalu dipertahankan.
</p>

### <span style="color:#0061AF">1.4.1 Merapikan Format Teks</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Spasi berlebih dan perbedaan huruf besar-kecil dirapikan agar label yang sama tidak terhitung sebagai kategori berbeda.
</p>

```python
def clean_text_columns(df):
    df = df.copy()
    for col in df.select_dtypes(include=["object", "string"]).columns:
        df[col] = df[col].map(
            lambda value: value.strip() if isinstance(value, str) else value
        )
        df[col] = df[col].replace("", pd.NA)
    return df

df_clients = clean_text_columns(df_clients)
df_trips = clean_text_columns(df_trips)
df_shifts = clean_text_columns(df_shifts)
df_drivers = clean_text_columns(df_drivers)
df_pools = clean_text_columns(df_pools)
df_requests = clean_text_columns(df_requests)

tables = make_tables()
print("Trimmed surrounding whitespace and converted empty strings to missing values.")
```

### <span style="color:#0061AF">1.4.2 Standarisasi Kategori Layanan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Label <code>service_type</code> yang tidak konsisten dipetakan ke <code>Reguler</code>, <code>Silver</code>, <code>Golden</code>, dan <code>E-Cabbie</code> (nama kelas layanan, mis. label <code>Bluebird</code> dan <code>BB Reguler</code> menjadi <code>Reguler</code>). Catatan: <code>Reguler</code> di <code>service_type_clean</code> adalah kelas layanan, berbeda dari <code>Reguler</code> di kolom <code>trip_type</code> (Reguler vs Charter). Label asli disimpan di <code>service_type_original</code>.
</p>

```python
# Keep the original label for traceability.
df_trips["service_type_original"] = df_trips["service_type"]

service_normalized = (
    df_trips["service_type"]
    .astype("string")
    .str.strip()
    .str.casefold()
    .str.replace(r"\s+", " ", regex=True)
)

# Confirmed mappings from the project rules and known variants in the data.
service_mapping = {
    "bluebird": "Reguler",
    "blue bird": "Reguler",
    "bb reguler": "Reguler",
    "reguler": "Reguler",
    "silverbird": "Silver",
    "silver bird": "Silver",
    "silver": "Silver",
    "goldenbird": "Golden",
    "golden bird": "Golden",
    "golden": "Golden",
    "e-cabbie": "E-Cabbie",
    "e cabbie": "E-Cabbie",
    "ecabbie": "E-Cabbie",
    "listrik": "E-Cabbie",
    "electric": "E-Cabbie",
}

df_trips["service_type_clean"] = service_normalized.map(service_mapping)

unmapped_service_types = (
    df_trips.loc[
        service_normalized.notna() & df_trips["service_type_clean"].isna(),
        "service_type_original",
    ]
    .value_counts(dropna=False)
)

print("Unmapped service categories (review these before finalizing):")
display(unmapped_service_types)
print("Standardized service categories:")
display(df_trips["service_type_clean"].value_counts(dropna=False))
```

### <span style="color:#0061AF">1.4.3 Validasi Pemetaan Layanan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Pemetaan dicek terhadap <code>vehicle_type_assigned</code> pada tabel driver. Jika jumlah pasangan yang salah adalah 0, pemetaan terkonfirmasi.
</p>

```python
trip_vehicle_check = df_trips[["driver_id", "service_type_clean"]].merge(
    df_drivers[["driver_id", "vehicle_type_assigned"]], on="driver_id", how="left"
)
assert len(trip_vehicle_check) == len(df_trips), "merge changed the number of rows"

display(pd.crosstab(trip_vehicle_check["service_type_clean"], trip_vehicle_check["vehicle_type_assigned"]))

# Expected pairs: each service class must be driven by the matching vehicle type.
expected_pairs = {
    "Reguler": "Reguler",
    "Silver": "Silver",
    "Golden": "Golden",
    "E-Cabbie": "Electric",
}
# Every service label must have an expected vehicle type; otherwise .map() gives NaN and every trip looks wrong.
missing_labels = set(trip_vehicle_check["service_type_clean"].dropna().unique()) - set(expected_pairs)
assert not missing_labels, f"service labels without an expected vehicle type: {missing_labels}"
wrong_pairs = (
    trip_vehicle_check["service_type_clean"].map(expected_pairs)
    != trip_vehicle_check["vehicle_type_assigned"]
).sum()
print(f"Trips where service_type_clean does not match the driver's vehicle type: {wrong_pairs}")
assert wrong_pairs == 0, "service_type_clean does not match vehicle_type_assigned"
```

### <span style="color:#0061AF">1.4.4 Cek Kolom Kategori Lain</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Nilai unik kolom kategori lain dicetak untuk memastikan tidak ada variasi ejaan yang perlu dipetakan.
</p>

```python
category_columns = {
    "clients": (df_clients, ["industry"]),
    "trips": (df_trips, ["trip_type", "pickup_zone_type", "booking_channel", "payment_method", "service_type_clean"]),
    "shifts": (df_shifts, ["dominant_zone_worked"]),
    "drivers": (df_drivers, ["pool_id", "employment_status", "vehicle_type_assigned"]),
    "pools": (df_pools, ["region_area", "primary_zone_affinity"]),
    "requests": (df_requests, ["pickup_zone_type", "request_source", "status"]),
}

for table_name, (df, columns) in category_columns.items():
    for col in columns:
        print(f"{table_name}.{col}: {sorted(df[col].dropna().unique())}")
```

### <span style="color:#0061AF">1.4.5 Konversi Tipe Data</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Kolom waktu diubah menjadi <em>datetime</em> dan kolom angka menjadi numerik, karena validasi urutan waktu dan nilai hanya bisa dilakukan setelah tipe data benar.
</p>

```python
def convert_datetime_columns(df, columns, table_name):
    df = df.copy()
    for col in columns:
        original_not_null = df[col].notna()
        df[col] = pd.to_datetime(df[col], errors="coerce")
        conversion_errors = int((original_not_null & df[col].isna()).sum())
        print(f"{table_name}.{col}: {conversion_errors} conversion errors")
    return df

df_trips = convert_datetime_columns(
    df_trips,
    ["booking_time", "meter_start_time", "dropoff_time"],
    "trips",
)
df_requests = convert_datetime_columns(
    df_requests, ["requested_time"], "requests"
)
df_drivers = convert_datetime_columns(
    df_drivers, ["join_date"], "drivers"
)
df_shifts = convert_datetime_columns(
    df_shifts, ["shift_date", "clock_in_time", "clock_out_time"], "shifts"
)

numeric_columns = {
    "trips": ["distance_km", "total_fare"],
    "shifts": ["trip_count", "total_idle_minutes", "ritase_target_met"],
    "drivers": ["driver_rating"],
    "pools": ["distance_to_soetta_km", "distance_to_cbd_km"],
    "requests": ["wait_time_minutes"],
}

tables = make_tables()
numeric_conversion_errors = []
for table_name, columns in numeric_columns.items():
    df = tables[table_name].copy()
    for col in columns:
        if col not in df.columns:
            continue
        original_not_null = df[col].notna()
        df[col] = pd.to_numeric(df[col], errors="coerce")
        errors = int((original_not_null & df[col].isna()).sum())
        numeric_conversion_errors.append({
            "table": table_name, "column": col, "conversion_errors": errors
        })
    tables[table_name] = df

df_clients = tables["clients"]
df_trips = tables["trips"]
df_shifts = tables["shifts"]
df_drivers = tables["drivers"]
df_pools = tables["pools"]
df_requests = tables["requests"]

display(pd.DataFrame(numeric_conversion_errors))
```

## <span style="color:#0061AF">1.5 Menerapkan Aturan Pembersihan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Pada tahap ini masalah kualitas ditandai dengan kolom <code>flag_*</code> (jarak negatif, delay booking-ke-meter negatif, idle time melebihi durasi shift, dan lainnya). Penandaan dilakukan sebelum ada baris yang dikeluarkan. Aturan eksklusi diterapkan pada bagian 1.8 setelah semua relasi divalidasi, agar dampaknya pada tabel lain bisa diukur.
</p>

### <span style="color:#0061AF">1.5.1 Flag Kualitas Data</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Kolom <code>flag_*</code> dibuat untuk jarak negatif, delay booking-ke-meter negatif, dan idle time yang melebihi durasi shift. Nilai asli tidak diubah.
</p>

```python
# Trip timing and value checks
df_trips["booking_to_meter_minutes"] = (
    df_trips["meter_start_time"] - df_trips["booking_time"]
).dt.total_seconds() / 60

df_trips["trip_duration_minutes"] = (
    df_trips["dropoff_time"] - df_trips["meter_start_time"]
).dt.total_seconds() / 60

df_trips["flag_negative_booking_delay"] = (
    df_trips["booking_to_meter_minutes"] < 0
)
df_trips["flag_nonpositive_trip_duration"] = (
    df_trips["trip_duration_minutes"] <= 0
)
df_trips["flag_missing_trip_timestamp"] = (
    df_trips[["booking_time", "meter_start_time", "dropoff_time"]]
    .isna().any(axis=1)
)
df_trips["flag_negative_distance"] = df_trips["distance_km"] < 0
df_trips["flag_negative_fare"] = df_trips["total_fare"] < 0

# Shift duration uses full timestamps, so overnight shifts are handled.
df_shifts["shift_duration_minutes"] = (
    df_shifts["clock_out_time"] - df_shifts["clock_in_time"]
).dt.total_seconds() / 60

df_shifts["flag_invalid_shift_duration"] = (
    df_shifts["shift_duration_minutes"] <= 0
)
df_shifts["flag_negative_idle_time"] = df_shifts["total_idle_minutes"] < 0

# Small tolerance accommodates one-decimal rounding in idle time.
rounding_tolerance = 0.1
df_shifts["flag_idle_exceeds_duration"] = (
    df_shifts["total_idle_minutes"]
    > df_shifts["shift_duration_minutes"] + rounding_tolerance
)

trip_flag_summary = pd.Series({
    "Negative distance": int(df_trips["flag_negative_distance"].sum()),
    "Negative booking-to-meter delay": int(df_trips["flag_negative_booking_delay"].sum()),
    "Nonpositive trip duration": int(df_trips["flag_nonpositive_trip_duration"].sum()),
    "Missing trip timestamp": int(df_trips["flag_missing_trip_timestamp"].sum()),
    "Negative fare": int(df_trips["flag_negative_fare"].sum()),
})
shift_flag_summary = pd.Series({
    "Invalid shift duration": int(df_shifts["flag_invalid_shift_duration"].sum()),
    "Negative idle time": int(df_shifts["flag_negative_idle_time"].sum()),
    "Idle time exceeds duration (with 0.1-minute tolerance)": int(
        df_shifts["flag_idle_exceeds_duration"].sum()
    ),
})

print("TRIP QUALITY FLAGS")
display(trip_flag_summary.to_frame("records"))
print("SHIFT QUALITY FLAGS")
display(shift_flag_summary.to_frame("records"))

display(
    df_shifts.loc[
        df_shifts["flag_invalid_shift_duration"]
        | df_shifts["flag_negative_idle_time"]
        | df_shifts["flag_idle_exceeds_duration"],
        [
            "shift_id", "driver_id", "shift_date", "clock_in_time",
            "clock_out_time", "shift_duration_minutes", "total_idle_minutes",
            "trip_count", "flag_invalid_shift_duration",
            "flag_negative_idle_time", "flag_idle_exceeds_duration",
        ],
    ].head(20)
)
```

### <span style="color:#0061AF">Interpretasi Flag Kualitas Data</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Negative distance:</strong> Menandai perjalanan dengan <code>distance_km</code> kurang dari 0. Jarak negatif tidak sesuai dengan makna normal jarak perjalanan sehingga catatan tersebut perlu ditangani berdasarkan aturan pembersihan data.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Negative booking-to-meter delay:</strong> Menandai perjalanan ketika <code>meter_start_time</code> lebih awal daripada <code>booking_time</code>, sehingga selisih waktu pemesanan ke meter bernilai negatif. Kondisi ini menunjukkan urutan waktu yang tidak sesuai dengan asumsi bahwa pemesanan terjadi sebelum meter dimulai.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Nonpositive trip duration:</strong> Menandai perjalanan dengan durasi nol atau negatif, berdasarkan selisih <code>dropoff_time</code> dan <code>meter_start_time</code>. Kondisi ini dapat menunjukkan timestamp yang tidak konsisten atau durasi perjalanan yang tidak valid.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Missing trip timestamp:</strong> Menandai perjalanan yang memiliki nilai kosong pada salah satu kolom <code>booking_time</code>, <code>meter_start_time</code>, atau <code>dropoff_time</code>. Timestamp yang tidak lengkap dapat membatasi analisis berbasis waktu.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Negative fare:</strong> Menandai perjalanan dengan <code>total_fare</code> kurang dari 0. Nilai ini perlu diperiksa untuk menentukan apakah merupakan kesalahan pencatatan atau penyesuaian transaksi yang sah.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Invalid shift duration:</strong> Menandai shift dengan durasi nol atau negatif, berdasarkan selisih <code>clock_out_time</code> dan <code>clock_in_time</code>. Shift yang melewati tengah malam harus ditangani dengan benar agar durasinya tidak salah dihitung.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Negative idle time:</strong> Menandai shift dengan <code>total_idle_minutes</code> kurang dari 0. Nilai ini perlu diperiksa karena waktu menganggur secara normal tidak bernilai negatif.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Idle time exceeds duration (with 0.1-minute tolerance):</strong> Menandai shift ketika <code>total_idle_minutes</code> melebihi <code>shift_duration_minutes</code> lebih dari 0,1 menit. Toleransi ini digunakan untuk mengakomodasi perbedaan pembulatan kecil, bukan sebagai bukti bahwa data yang ditandai pasti salah.
</p>

## <span style="color:#0061AF">1.6 Investigasi Exception dan Nilai Ekstrem</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Tidak semua nilai ekstrem salah. Bagian ini membedakan nilai yang <strong>tidak mungkin</strong> (jarak negatif) dari nilai yang <strong>hanya tidak biasa</strong> (trip Charter yang panjang). Outlier IQR dan batas jarak ditandai untuk ditinjau, tidak dihapus otomatis.
</p>

### <span style="color:#0061AF">1.6.1 Flag Outlier</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Outlier ditandai dengan metode IQR untuk ditinjau, tidak dihapus otomatis, karena nilai ekstrem belum tentu salah.
</p>

```python
def add_iqr_outlier_flag(df, column):
    df = df.copy()
    values = df[column].dropna()
    flag_col = f"flag_outlier_{column}"

    if values.empty:
        df[flag_col] = False
        return df

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    df[flag_col] = (
        (df[column] < lower_bound) | (df[column] > upper_bound)
    ).fillna(False)

    print(
        f"{column}: lower={lower_bound:.2f}, upper={upper_bound:.2f}, "
        f"flagged={int(df[flag_col].sum())}"
    )
    return df

df_trips = add_iqr_outlier_flag(df_trips, "distance_km")
df_trips = add_iqr_outlier_flag(df_trips, "total_fare")
```

### <span style="color:#0061AF">1.6.2 Cek Jarak Berbasis Aturan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Flag IQR menandai sekitar 8% trip, sebagian besar normal (trip Charter memang panjang). Jarak yang <strong>tidak mungkin</strong> dicek dengan aturan sederhana berdasarkan celah pada data: jarak negatif, jarak di atas 500 km, dan jarak positif di bawah 1 km. Batas 500 km dan 1 km adalah penilaian dari data ini, bukan aturan bisnis resmi. Trip tetap disimpan dan hanya <code>distance_km_clean</code> yang dikosongkan, karena tarif, waktu, dan pengemudi pada trip tersebut valid.
</p>

```python
# Evidence 1: Charter trips are long on purpose, Reguler trips are not.
print(df_trips.groupby("trip_type")["distance_km"].describe().round(1))

# Evidence 2: look for the gap in the data around 1 km.
print("\nTrips with distance between 0.85 and 1.25 km:")
print(df_trips.loc[df_trips["distance_km"].between(0.85, 1.25), "distance_km"].value_counts().sort_index())

# Evidence 3: how fast would the taxi have to drive on the trips above 500 km?
too_high_check = df_trips.loc[df_trips["distance_km"] > 500]
speed_kmh = too_high_check["distance_km"] / (too_high_check["trip_duration_minutes"] / 60)
print(f"\nTrips above 500 km: {len(too_high_check)} | all Reguler: {(too_high_check['trip_type'] == 'Reguler').all()}")
print(f"Implied speed (km/h): min {speed_kmh.min():,.0f}, max {speed_kmh.max():,.0f}  -> impossible for a taxi")

# Evidence 4: what the IQR flag from section 9 really caught.
iqr_flagged = df_trips["flag_outlier_distance_km"]
is_charter = df_trips["trip_type"].eq("Charter")
is_reguler = df_trips["trip_type"].eq("Reguler")
is_below_1km = df_trips["distance_km"].between(0, 1, inclusive="neither")
print(f"\nIQR distance flags: {int(iqr_flagged.sum()):,}")
print(f"  Charter trips (long on purpose):            {int((iqr_flagged & is_charter).sum()):,}")
print(f"  Reguler trips up to 500 km (long, possible): {int((iqr_flagged & is_reguler & (df_trips['distance_km'] <= 500)).sum()):,}")
print(f"  Reguler trips above 500 km (real errors):    {int((iqr_flagged & is_reguler & (df_trips['distance_km'] > 500)).sum()):,}")
print(f"  Trips below 1 km caught by the IQR flag:     {int((iqr_flagged & is_below_1km).sum())} of {int(is_below_1km.sum())}")
```

```python
MAX_VALID_DISTANCE_KM = 500   # Reguler jumps from about 100 km to 600+ km; Charter never goes above 320 km
MIN_VALID_DISTANCE_KM = 1     # no trip exists between 0.9 and 1.2 km

df_trips["flag_distance_too_high"] = df_trips["distance_km"] > MAX_VALID_DISTANCE_KM
df_trips["flag_distance_too_low"] = (
    (df_trips["distance_km"] > 0) & (df_trips["distance_km"] < MIN_VALID_DISTANCE_KM)
)
df_trips["flag_distance_unreliable"] = (
    df_trips["flag_negative_distance"]
    | df_trips["flag_distance_too_high"]
    | df_trips["flag_distance_too_low"]
)

# Bad distances become NaN in a NEW column. The original distance_km stays as it is.
df_trips["distance_km_clean"] = df_trips["distance_km"].where(~df_trips["flag_distance_unreliable"])

distance_flag_summary = pd.Series({
    "Negative distance": int(df_trips["flag_negative_distance"].sum()),
    "Distance above 500 km": int(df_trips["flag_distance_too_high"].sum()),
    "Distance below 1 km": int(df_trips["flag_distance_too_low"].sum()),
    "Unreliable distance (any of the three)": int(df_trips["flag_distance_unreliable"].sum()),
    "distance_km_clean is NaN": int(df_trips["distance_km_clean"].isna().sum()),
})
display(distance_flag_summary.to_frame("trips"))
```

## <span style="color:#0061AF">1.7 Validasi Key dan Relasi Antar Tabel</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Primary key, foreign key, dan hubungan antar tabel divalidasi sebelum eksklusi dilakukan. Tujuannya mengetahui ID yang tidak punya pasangan, baris yang terdampak, serta mana sel kosong yang struktural (bukan error).
</p>

### <span style="color:#0061AF">1.7.1 Primary Key dan Duplikat</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Keunikan primary key dan baris duplikat dicek untuk setiap tabel. ID yang berulang pada tabel lain bisa sah, sehingga tidak dihapus otomatis.
</p>

```python
tables = make_tables()
primary_keys = {
    "clients": "company_id",
    "trips": "trip_id",
    "shifts": "shift_id",
    "drivers": "driver_id",
    "pools": "pool_id",
    "requests": "request_id",
}

key_check_rows = []
for table_name, df in tables.items():
    key = primary_keys[table_name]
    key_check_rows.append({
        "table": table_name,
        "exact_duplicate_rows": int(df.duplicated().sum()),
        "missing_primary_keys": int(df[key].isna().sum()),
        "duplicate_primary_keys": int(df[key].dropna().duplicated().sum()),
    })

key_check_report = pd.DataFrame(key_check_rows)
display(key_check_report)
```

### <span style="color:#0061AF">1.7.2 Validasi Foreign Key</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Hubungan antar tabel dicek sebelum ada baris yang dikeluarkan: berapa ID yang tidak punya pasangan dan berapa baris yang terdampak.
</p>

```python
def find_unmatched_keys(child_df, child_key, parent_df, parent_key):
    child_values = child_df[child_key].dropna()
    parent_values = parent_df[parent_key].dropna()
    return child_values[~child_values.isin(parent_values)].drop_duplicates()

tables = make_tables()
relationships = [
    ("Trips -> Drivers", df_trips, "driver_id", df_drivers, "driver_id"),
    ("Trips -> Shifts", df_trips, "shift_id", df_shifts, "shift_id"),
    ("Trips -> Clients", df_trips, "company_id", df_clients, "company_id"),
    ("Shifts -> Drivers", df_shifts, "driver_id", df_drivers, "driver_id"),
    ("Drivers -> Pools", df_drivers, "pool_id", df_pools, "pool_id"),
    ("Requests -> Trips", df_requests, "matched_trip_id", df_trips, "trip_id"),
]

relationship_results = []
for label, child, child_key, parent, parent_key in relationships:
    unmatched = find_unmatched_keys(child, child_key, parent, parent_key)
    relationship_results.append({
        "relationship": label,
        "unmatched_unique_ids": int(len(unmatched)),
    })
    print(f"{label}: {len(unmatched)} unmatched unique IDs")
    if not unmatched.empty:
        display(unmatched.head(20))

relationship_report = pd.DataFrame(relationship_results)
```

### <span style="color:#0061AF">1.7.3 Null Struktural vs Missing Value Sebenarnya</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Sel kosong belum tentu error. <code>company_id</code> kosong pada trip non-korporat, serta <code>matched_trip_id</code> dan <code>wait_time_minutes</code> kosong pada request yang dibatalkan. Sel kosong ini sengaja dibiarkan.
</p>

```python
# 1) company_id only exists for eVoucher trips.
display(pd.crosstab(df_trips["payment_method"], df_trips["company_id"].notna().rename("has_company_id")))

is_evoucher = df_trips["payment_method"].eq("eVoucher Corporate")
print("eVoucher trips WITHOUT company_id:", int((is_evoucher & df_trips["company_id"].isna()).sum()), "(should be 0)")
print("Non-eVoucher trips WITH company_id:", int((~is_evoucher & df_trips["company_id"].notna()).sum()), "(should be 0)")

# 2) matched_trip_id and wait_time_minutes only exist for Fulfilled requests.
display(pd.crosstab(df_requests["status"], df_requests["matched_trip_id"].notna().rename("has_matched_trip_id")))

is_fulfilled = df_requests["status"].eq("Fulfilled")
print("Fulfilled requests without matched_trip_id:", int((is_fulfilled & df_requests["matched_trip_id"].isna()).sum()), "(should be 0)")
print("Cancelled requests WITH matched_trip_id:", int((~is_fulfilled & df_requests["matched_trip_id"].notna()).sum()), "(should be 0)")
print("matched_trip_id and wait_time_minutes empty in different rows:",
      int((df_requests["matched_trip_id"].isna() != df_requests["wait_time_minutes"].isna()).sum()), "(should be 0)")

# 3) Trips that have no request.
trip_has_request = df_trips["trip_id"].isin(df_requests["matched_trip_id"].dropna())
print("\nTrips with no request, by booking_channel:")
print(df_trips.loc[~trip_has_request, "booking_channel"].value_counts())
```

### <span style="color:#0061AF">1.7.4 Cek Trip terhadap Shift dan Trip Lain</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Pengemudi trip harus sama dengan pengemudi shift, meter harus mulai di dalam shift, dan trip tidak boleh dimulai sebelum trip sebelumnya selesai. Hanya trip yang <strong>mulai</strong> di luar shift yang ditandai. Delay booking-ke-meter negatif ditandai dan hanya nilai delay-nya yang dikosongkan di <code>booking_to_meter_minutes_clean</code>.
</p>

```python
# Join each trip to its shift. shift_id is unique in shifts, so the number of rows must NOT change.
trip_shift = df_trips[["trip_id", "shift_id", "driver_id", "meter_start_time", "dropoff_time"]].merge(
    df_shifts[["shift_id", "driver_id", "clock_in_time", "clock_out_time"]],
    on="shift_id", how="left", suffixes=("", "_shift"),
)
assert len(trip_shift) == len(df_trips), "merge changed the number of rows"

# Check 1: the trip's driver must be the shift's driver.
print("Trips whose driver differs from their shift's driver:",
      int((trip_shift["driver_id"] != trip_shift["driver_id_shift"]).sum()))

# Check 2: the meter must start inside the shift.
starts_outside_shift = (
    (trip_shift["meter_start_time"] < trip_shift["clock_in_time"])
    | (trip_shift["meter_start_time"] > trip_shift["clock_out_time"])
)
df_trips["flag_meter_outside_shift"] = starts_outside_shift.to_numpy()
print("Trips that START outside the shift window:", int(df_trips["flag_meter_outside_shift"].sum()))

# Trips that END after clock-out: only a few minutes late is normal.
minutes_after_clock_out = (trip_shift["dropoff_time"] - trip_shift["clock_out_time"]).dt.total_seconds() / 60
late_trips = minutes_after_clock_out[minutes_after_clock_out > 0]
print(f"Trips ending after clock-out: {len(late_trips):,} | median {late_trips.median():.1f} min | 99th percentile {late_trips.quantile(0.99):.1f} min -> normal, leave")

# Check 3: one driver cannot start a trip before the previous trip has ended.
trips_sorted = df_trips.sort_values(["driver_id", "meter_start_time"])
previous_dropoff = trips_sorted.groupby("driver_id")["dropoff_time"].shift()
df_trips["flag_overlap_same_driver"] = trips_sorted["meter_start_time"] < previous_dropoff
print("Trips that start before the same driver's previous trip ended:", int(df_trips["flag_overlap_same_driver"].sum()))

# Negative booking-to-meter delay: keep the trip, blank only the delay in a NEW column.
df_trips["booking_to_meter_minutes_clean"] = df_trips["booking_to_meter_minutes"].where(
    ~df_trips["flag_negative_booking_delay"]
)
print("booking_to_meter_minutes_clean is NaN:", int(df_trips["booking_to_meter_minutes_clean"].isna().sum()))
```

### <span style="color:#0061AF">1.7.5 Cek Shift dan Driver</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Cek urutan shift, <code>join_date</code>, rating 1–5, shift tanpa trip, dan aturan <code>ritase_target_met</code>. Temuan hanya ditandai karena belum bisa dipastikan mana yang salah.
</p>

```python
# Check 1: a driver cannot start a new shift before the previous shift ended.
shifts_sorted = df_shifts.sort_values(["driver_id", "clock_in_time"])
previous_clock_out = shifts_sorted.groupby("driver_id")["clock_out_time"].shift()
df_shifts["flag_overlap_prev_shift"] = shifts_sorted["clock_in_time"] < previous_clock_out
print("Shifts that start before the same driver's previous shift ended:", int(df_shifts["flag_overlap_prev_shift"].sum()))

# Check 2: shifts dated before the driver joined.
DATA_END_DATE = pd.Timestamp("2023-12-31")   # the brief says the data covers October-December 2023
join_date_by_driver = df_drivers.set_index("driver_id")["join_date"]
first_shift_by_driver = df_shifts.groupby("driver_id")["shift_date"].min()

df_shifts["flag_shift_before_join_date"] = df_shifts["shift_date"] < df_shifts["driver_id"].map(join_date_by_driver)
df_drivers["flag_join_after_data_end"] = df_drivers["join_date"] > DATA_END_DATE
df_drivers["flag_first_shift_before_join"] = df_drivers["driver_id"].map(first_shift_by_driver) < df_drivers["join_date"]

print("Shifts dated before the driver's join_date:", int(df_shifts["flag_shift_before_join_date"].sum()),
      f"of {len(df_shifts):,} ({df_shifts['flag_shift_before_join_date'].mean():.1%})")
print("Drivers with join_date after 2023-12-31 (after the data ends):", int(df_drivers["flag_join_after_data_end"].sum()))
print("Drivers whose first shift is before their join_date:", int(df_drivers["flag_first_shift_before_join"].sum()))

# Check 3: ratings.
print("Drivers with a rating outside 1-5:", int((~df_drivers["driver_rating"].between(1, 5)).sum()), "(should be 0)")

# Check 4: shifts with 0 trips. Are they fully idle, and does the trips table agree?
zero_trip_shift = df_shifts["trip_count"].eq(0)
fully_idle = (df_shifts["total_idle_minutes"] - df_shifts["shift_duration_minutes"]).abs() < 0.1
print(f"\nShifts with 0 trips: {int(zero_trip_shift.sum()):,} ({zero_trip_shift.mean():.1%}) | all fully idle: {bool(fully_idle[zero_trip_shift].all())}")
print("Those shifts have no trips in the trips table either:", bool((~df_shifts.loc[zero_trip_shift, "shift_id"].isin(df_trips["shift_id"])).all()))

# Check 5: how is ritase_target_met defined? (the brief does not say)
print("ritase_target_met equals (trip_count >= 6) in every shift:",
      bool(((df_shifts["trip_count"] >= 6).astype(int) == df_shifts["ritase_target_met"]).all()))

# Check 6: do the pool labels agree with the work the drivers really do?
pool_of_shift = df_shifts["driver_id"].map(df_drivers.set_index("driver_id")["pool_id"]).rename("pool_id")
print("\nShare of shifts by dominant zone, per pool:")
display(pd.crosstab(pool_of_shift, df_shifts["dominant_zone_worked"], normalize="index").round(2))
display(df_pools[["pool_id", "primary_zone_affinity"]])
```

```python
# Evidence for Check 5: where does ritase_target_met switch from 0 to 1?
# The brief only says the column marks whether the daily ritase target was met; it does not give the number.
ritase_flag_table = pd.crosstab(df_shifts["trip_count"], df_shifts["ritase_target_met"])
print("Shifts per trip_count (rows) and ritase_target_met value (columns):")
display(ritase_flag_table)

lowest_trip_count_with_flag = int(df_shifts.loc[df_shifts["ritase_target_met"] == 1, "trip_count"].min())
highest_trip_count_without_flag = int(df_shifts.loc[df_shifts["ritase_target_met"] == 0, "trip_count"].max())
print(f"Highest trip_count with flag 0: {highest_trip_count_without_flag} | lowest trip_count with flag 1: {lowest_trip_count_with_flag}")
print(f"Shifts with flag 1: {int(df_shifts['ritase_target_met'].sum()):,} of {len(df_shifts):,} "
      f"({df_shifts['ritase_target_met'].mean() * 100:.2f}%)")
assert highest_trip_count_without_flag + 1 == lowest_trip_count_with_flag, "the flag does not switch at a single trip count"
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi <code>ritase_target_met</code>:</strong> brief hanya mendefinisikan kolom ini sebagai "penanda apakah target ritase harian tercapai" tanpa menyebut angkanya. Angka targetnya dibaca dari data: tabel di atas menunjukkan semua shift dengan 0–5 trip bernilai 0 dan semua shift dengan 6–10 trip bernilai 1, tanpa pengecualian (trip_count tertinggi dengan flag 0 = 5, terendah dengan flag 1 = 6). Jadi flag ini aktif tepat pada <strong>≥ 6 trip per shift</strong>. Angka 6 adalah aturan yang tersirat di data, bukan target yang kami tetapkan dan bukan target resmi yang disebut brief. Hanya 894 dari 252.524 shift (0,35%) yang bernilai 1, sehingga flag ini hampir tidak pernah tercapai dan tidak dipakai sebagai KPI dalam analisis ini (2.5.9, Rekomendasi).
</p>

### <span style="color:#0061AF">1.7.6 Cek Request terhadap Trip dan Wait Time 0,5 Menit</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Request dan trip yang cocok harus sama pada waktu, zona, dan channel. <code>wait_time_minutes</code> punya batas bawah 0,5 menit, dan semua request yang terkait trip dengan delay negatif bernilai tepat 0,5. Request tetap disimpan dan <code>wait_time_minutes_clean</code> dikosongkan untuk kasus tersebut.
</p>

```python
# Join fulfilled requests to their trip. trip_id is unique, so the number of rows must NOT change.
fulfilled_requests = df_requests.dropna(subset=["matched_trip_id"])
request_trip = fulfilled_requests.merge(
    df_trips[["trip_id", "booking_time", "pickup_zone_type", "booking_channel"]],
    left_on="matched_trip_id", right_on="trip_id", how="left", suffixes=("", "_trip"),
)
assert len(request_trip) == len(fulfilled_requests), "merge changed the number of rows"

print("Requests whose trip is missing:", int(request_trip["trip_id"].isna().sum()), "(should be 0)")
print("requested_time differs from the trip's booking_time:", int((request_trip["requested_time"] != request_trip["booking_time"]).sum()), "(should be 0)")
print("pickup zone differs from the trip's pickup zone:", int((request_trip["pickup_zone_type"] != request_trip["pickup_zone_type_trip"]).sum()), "(should be 0)")
print("request_source differs from the trip's booking_channel:", int((request_trip["request_source"] != request_trip["booking_channel"]).sum()), "(should be 0)")

# The wait-time floor.
print("\nMost common small wait times (minutes):")
print(df_requests["wait_time_minutes"].round(1).value_counts().sort_index().head(6))

negative_delay_trip_ids = set(df_trips.loc[df_trips["flag_negative_booking_delay"], "trip_id"])
df_requests["flag_matched_trip_negative_delay"] = df_requests["matched_trip_id"].isin(negative_delay_trip_ids)
df_requests["flag_wait_at_floor"] = df_requests["wait_time_minutes"] == 0.5
df_requests["wait_time_minutes_clean"] = df_requests["wait_time_minutes"].where(
    ~df_requests["flag_matched_trip_negative_delay"]
)

linked = df_requests.loc[df_requests["flag_matched_trip_negative_delay"], "wait_time_minutes"]
print(f"\nRequests tied to negative-delay trips: {len(linked):,} | their wait_time values: {sorted(linked.unique())}")
print("Requests with wait_time exactly 0.5:", int(df_requests["flag_wait_at_floor"].sum()))
```

Kelengkapan setiap hari dicek karena hari yang separuh kosong menghasilkan angka harian yang palsu rendah. Hari terakhir (2023-12-30) tidak lengkap sehingga tidak dipakai pada perbandingan harian.

## <span style="color:#0061AF">1.8 Tabel Final yang Sudah Dibersihkan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Berdasarkan seluruh validasi, tabel final dibentuk menurut aturan yang terdokumentasi. Trip dengan jarak negatif atau delay booking-ke-meter negatif <strong>dikeluarkan</strong> dari tabel trip final, sedangkan baris aslinya tetap tersimpan di file exception. Aturan lain bersifat flag untuk ditinjau. Dampak eksklusi pada metrik utama dibandingkan pada bagian terakhir.
</p>

### <span style="color:#0061AF">1.8.1 Simpan Data Exception dan Buat Tabel Final</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Semua trip dengan flag masalah disimpan ke file exception beserta alasan dan perlakuannya (kolom <code>treatment</code>). Trip dengan jarak negatif atau delay booking-ke-meter negatif <strong>dikeluarkan</strong> dari tabel trip final, sedangkan flag lainnya hanya untuk ditinjau. Tabel lain tidak dihapus. Dampak eksklusi diukur pada bagian 1.8.4.
</p>

```python
# Save all flagged trips before excluding any rows.
# The exception file lists EVERY trip with a problem flag, and its treatment column says what happens to it.
trip_issue_flags = {
    "flag_negative_distance": "Negative distance",
    "flag_distance_too_high": "Distance above 500 km",
    "flag_distance_too_low": "Distance below 1 km",
    "flag_negative_booking_delay": "Negative booking-to-meter delay",
    "flag_meter_outside_shift": "Meter started outside the shift",
    "flag_overlap_same_driver": "Overlaps the previous trip of the same driver",
}
trip_flag_columns = list(trip_issue_flags)
df_trips["flag_any_trip_issue"] = df_trips[trip_flag_columns].any(axis=1)

# Confirmed exclusion rule: a negative distance OR a negative booking-to-meter delay.
# All other flags (distance too high/low, outside shift, overlap) are review flags only.
df_trips["exclude_trip"] = df_trips["flag_negative_distance"] | df_trips["flag_negative_booking_delay"]

trip_exceptions = df_trips.loc[df_trips["flag_any_trip_issue"]].copy()
trip_exceptions["treatment"] = np.where(
    trip_exceptions["exclude_trip"],
    "Excluded from final trips table",
    "Kept in final trips table (review flag only)",
)

# One text column that says WHY the trip is in the file (a trip can have more than one reason).
trip_exceptions["issue_reason"] = trip_exceptions[trip_flag_columns].apply(
    lambda row: "; ".join(trip_issue_flags[col] for col in trip_flag_columns if row[col]),
    axis=1,
)

trip_exceptions.to_csv(
    exceptions_dir / "bb_trips_exceptions.csv",
    index=False,
    encoding="utf-8-sig",
)

# Apply the exclusion rule. The original rows stay in df_trips_raw and in the exception file.
df_trips_clean = df_trips.loc[~df_trips["exclude_trip"]].copy()

# Do not delete shifts because of idle-time flags; retain them with flags.
df_shifts_clean = df_shifts.copy()

print(f"Trips before exclusions: {len(df_trips):,}")
print(f"Trip exception records (any problem flag): {len(trip_exceptions):,}")
print(f"Trips excluded by the confirmed rule (unique): {int(df_trips['exclude_trip'].sum()):,}")
print(f"Trips retained: {len(df_trips_clean):,}")
assert len(df_trips_clean) + int(df_trips["exclude_trip"].sum()) == len(df_trips)
print(f"Shift records retained: {len(df_shifts_clean):,}")
print(f"Trip exceptions saved to: {exceptions_dir / 'bb_trips_exceptions.csv'}")

# Shift exception file supports investigation without removing shifts.
# It lists every shift with at least one review flag from 1.7.5, with the reason(s).
shift_issue_flags = {
    "flag_invalid_shift_duration": "Invalid shift duration",
    "flag_negative_idle_time": "Negative idle time",
    "flag_idle_exceeds_duration": "Idle time longer than the shift",
    "flag_overlap_prev_shift": "Starts before the driver's previous shift ended",
    "flag_shift_before_join_date": "Shift date before the driver's join_date",
}
shift_flag_columns = list(shift_issue_flags)
assert all(column in df_shifts.columns for column in shift_flag_columns)
shift_exceptions = df_shifts.loc[df_shifts[shift_flag_columns].any(axis=1)].copy()
shift_exceptions["treatment"] = "Kept in final shifts table (review flag only)"
shift_exceptions["issue_reason"] = shift_exceptions[shift_flag_columns].apply(
    lambda row: "; ".join(shift_issue_flags[col] for col in shift_flag_columns if row[col]),
    axis=1,
)
print(f"Shift exception records (any review flag, none removed): {len(shift_exceptions):,}")

shift_exceptions.to_csv(
    exceptions_dir / "bb_driver_shifts_exceptions.csv",
    index=False,
    encoding="utf-8-sig",
)
```

### <span style="color:#0061AF">1.8.2 Tabel Output Final dan Validasi Relasi</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Tabel final disusun dan relasi antar tabel divalidasi pada tabel final tersebut.
</p>

```python
# Use the cleaned trip table for export; other tables remain intact.
tables_final = {
    "clients": df_clients,
    "trips": df_trips_clean,
    "shifts": df_shifts_clean,
    "drivers": df_drivers,
    "pools": df_pools,
    "requests": df_requests,
}

# Removing trips can cause request references to point to excluded trip IDs.
# Report that explicitly rather than silently modifying requests.
unmatched_requests_after_exclusions = find_unmatched_keys(
    df_requests, "matched_trip_id", df_trips_clean, "trip_id"
)
print(
    "Unique request matched_trip_id values not found in cleaned trips:",
    len(unmatched_requests_after_exclusions),
)
if not unmatched_requests_after_exclusions.empty:
    display(unmatched_requests_after_exclusions.head(20))
```

### <span style="color:#0061AF">1.8.3 Sinkronisasi <code>shifts.trip_count</code></span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<code>trip_count</code> dihitung ulang dari tabel trip final dan shift yang tidak cocok ditandai. Shift yang punya trip yang dikeluarkan akan ditandai tidak cocok (kondisi yang diharapkan); shift itu tidak dihapus.
</p>

```python
trips_in_final_table = df_trips_clean.groupby("shift_id").size()

df_shifts_clean["trips_in_table"] = df_shifts_clean["shift_id"].map(trips_in_final_table).fillna(0).astype(int)
df_shifts_clean["flag_trip_count_mismatch"] = df_shifts_clean["trip_count"] != df_shifts_clean["trips_in_table"]

print(f"sum of shifts.trip_count: {df_shifts_clean['trip_count'].sum():,}")
print(f"trips in the final trips table: {len(df_trips_clean):,}")
print(f"Shifts where trip_count does not match the trips table: {int(df_shifts_clean['flag_trip_count_mismatch'].sum()):,}")

# These shifts are kept (not deleted). They are exactly the shifts that had an excluded trip.
shifts_with_excluded_trips = df_trips.loc[df_trips["exclude_trip"], "shift_id"].nunique()
print(f"Shifts that had at least one excluded trip: {shifts_with_excluded_trips:,}")
assert int(df_shifts_clean["flag_trip_count_mismatch"].sum()) == shifts_with_excluded_trips
```

### <span style="color:#0061AF">1.8.4 Dampak Eksklusi terhadap Metrik Utama</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Eksklusi 7.091 trip unik (505 jarak negatif + 6.595 delay negatif − 9 trip yang terkena keduanya) mengubah jumlah baris dan beberapa metrik. Tabel berikut mengukur dampaknya dibanding sebelum eksklusi (<em>sensitivity comparison</em>). Tidak ada request atau shift yang dihapus: 6.052 request tetap menunjuk ke trip yang dikeluarkan dan 5.780 shift ditandai karena jumlah trip-nya tidak lagi sama dengan tabel trip.
</p>

```python
# Sensitivity: how do the confirmed exclusions change the key numbers?
neg_dist = df_trips["flag_negative_distance"]
neg_delay = df_trips["flag_negative_booking_delay"]

exclusion_summary = pd.DataFrame({
    "trips": [
        neg_dist.sum(),
        neg_delay.sum(),
        (neg_dist & neg_delay).sum(),
        (neg_dist | neg_delay).sum(),
    ]
}, index=["negative distance", "negative booking-to-meter delay", "BOTH flags (overlap)", "UNIQUE trips excluded"])
display(exclusion_summary)
# Unique = distance + delay - overlap (the two flags must not simply be added)
assert exclusion_summary.iloc[3, 0] == exclusion_summary.iloc[0, 0] + exclusion_summary.iloc[1, 0] - exclusion_summary.iloc[2, 0]
assert exclusion_summary.iloc[3, 0] == len(df_trips) - len(df_trips_clean)

# Consequences for the other tables (nothing is deleted there)
excluded_trip_ids = set(df_trips.loc[df_trips["exclude_trip"], "trip_id"])
requests_pointing_to_excluded = int(df_requests["matched_trip_id"].isin(excluded_trip_ids).sum())
print(f"Requests whose matched trip was excluded (kept, reference now unmatched): {requests_pointing_to_excluded:,}")
print(f"Shifts that had an excluded trip (kept, flagged in flag_trip_count_mismatch): {shifts_with_excluded_trips:,}")


def trip_metrics(trips):
    """Key trip-level numbers for one version of the trips table."""
    return pd.Series({
        "trips": len(trips),
        "total fare (Rp)": trips["total_fare"].sum(),
        "average fare (Rp)": trips["total_fare"].mean(),
        "average distance (km, clean)": trips["distance_km_clean"].mean(),
        "average booking-to-meter delay (min, clean)": trips["booking_to_meter_minutes_clean"].mean(),
        "eVoucher share of trips (%)": trips["company_id"].notna().mean() * 100,
        "Airport share of trips (%)": (trips["pickup_zone_type"] == "Airport").mean() * 100,
    })


sensitivity = pd.DataFrame({
    "before exclusion": trip_metrics(df_trips),
    "after exclusion (final)": trip_metrics(df_trips_clean),
})
sensitivity["change (%)"] = (sensitivity["after exclusion (final)"] / sensitivity["before exclusion"] - 1) * 100
display(sensitivity.round(2))

# Shift-level ritase: source trip_count (includes the excluded trips) vs trips in the final table
shift_sensitivity = pd.DataFrame({
    "source trip_count": [df_shifts_clean["trip_count"].mean(), (df_shifts_clean["trip_count"] == 0).mean() * 100],
    "trips in final table": [df_shifts_clean["trips_in_table"].mean(), (df_shifts_clean["trips_in_table"] == 0).mean() * 100],
}, index=["average trips per shift", "shifts with 0 trips (%)"])
display(shift_sensitivity.round(2))
print("Request fulfilment rates are unchanged: requests are not deleted.")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> eksklusi mengurangi 2,23% trip (318.098 menjadi 311.007) dan 2,14% total tarif (Rp 26,48 miliar menjadi Rp 25,91 miliar), tetapi hampir tidak mengubah rata-rata: tarif +0,09%, jarak +0,08%, porsi eVoucher tetap 18,03%, dan porsi Airport +0,17%. Rata-rata delay booking-ke-meter bersih tetap 5,01 menit, dan tingkat fulfilment request tidak berubah karena request tidak dihapus. Ritase per shift 1,26 jika memakai <code>trip_count</code> asli shift (termasuk trip yang dikeluarkan) dan 1,23 jika hanya menghitung trip pada tabel final; shift tanpa trip 30,65% vs 31,57%. Analisis ritase memakai <code>trip_count</code> asli, karena trip yang dikeluarkan tetap perjalanan nyata yang hanya bermasalah pada jarak atau delay-nya. Kesimpulan bisnis tidak berubah oleh eksklusi. <strong>Batasan:</strong> waktu sibuk pengemudi (utilisasi) dihitung dari trip pada tabel final sehingga sedikit lebih rendah (9,4% dibanding sekitar 9,6% jika trip yang dikeluarkan ikut dihitung).
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Hubungan trip yang dikeluarkan dengan tabel request dan shift.</strong> Angka 6.052 request dan 5.780 shift di atas dirinci di sini dengan penyebutnya: berapa trip yang dikeluarkan punya request, apa status request tersebut, dan berapa persen tabel request dan shift yang terdampak.
</p>

```python
# How do the 7,091 excluded trips connect to the request and shift tables? (nothing is deleted there)
excluded_trip_table = df_trips.loc[df_trips["exclude_trip"], ["trip_id", "shift_id", "booking_channel"]]
excluded_has_request = excluded_trip_table["trip_id"].isin(df_requests["matched_trip_id"].dropna())

link_summary = pd.DataFrame({
    "trips": [len(excluded_trip_table), int(excluded_has_request.sum()), int((~excluded_has_request).sum())],
}, index=["excluded trips (total)", "excluded trips WITH a request", "excluded trips WITHOUT a request"])
display(link_summary)
assert link_summary.iloc[1, 0] + link_summary.iloc[2, 0] == link_summary.iloc[0, 0]

print("Booking channel of the excluded trips, with vs without a request:")
display(pd.crosstab(excluded_trip_table["booking_channel"], excluded_has_request.rename("has_request")))

requests_to_excluded = df_requests[df_requests["matched_trip_id"].isin(excluded_trip_table["trip_id"])]
print("Status of the requests that point to an excluded trip:")
display(requests_to_excluded["status"].value_counts().to_frame("requests"))

impact_on_tables = pd.DataFrame({
    "affected": [len(requests_to_excluded), excluded_trip_table["shift_id"].nunique()],
    "table total": [len(df_requests), len(df_shifts_clean)],
}, index=["requests pointing to an excluded trip", "shifts with at least one excluded trip"])
impact_on_tables["affected (%)"] = (impact_on_tables["affected"] / impact_on_tables["table total"] * 100).round(1)
display(impact_on_tables)
# Same numbers as in 1.8.3 and the cell above.
assert impact_on_tables.loc["requests pointing to an excluded trip", "affected"] == requests_pointing_to_excluded
assert impact_on_tables.loc["shifts with at least one excluded trip", "affected"] == shifts_with_excluded_trips
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> dari 7.091 trip yang dikeluarkan, 6.052 punya request dan 1.039 tidak punya request. Ke-1.039 trip tanpa request semuanya Street Hail, yang memang tidak melewati request; ke-6.052 trip dengan request terdiri dari App 4.306 dan Call Center 1.746. Ke-6.052 request itu semuanya berstatus Fulfilled dan tetap disimpan, sehingga tingkat kegagalan request tidak berubah. Tabel yang terdampak: 6.052 dari 289.622 request (2,1%) dan 5.780 dari 252.524 shift (2,3%). Karena 7.091 trip tersebar di 5.780 shift, sebagian shift punya lebih dari satu trip yang dikeluarkan.
</p>

### <span style="color:#0061AF">1.8.5 Apakah Trip yang Dikeluarkan Berbeda dari Trip yang Disimpan?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Eksklusi hanya aman jika trip yang dikeluarkan tidak berbeda secara sistematis dari trip yang disimpan. Komposisi channel, zona, metode bayar, dan tipe trip dibandingkan, dan besar delay negatif diperiksa.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Cara membaca tabel:</strong> kedua kolom persentase memakai penyebut yang berbeda. Kolom <code>excluded (%)</code> adalah porsi di antara <strong>7.091 trip yang dikeluarkan</strong> (jarak negatif atau delay booking-ke-meter negatif), sedangkan kolom <code>kept (%)</code> adalah porsi di antara <strong>311.007 trip yang disimpan</strong> di tabel final. Masing-masing kolom berjumlah 100%. Contoh: App 60,7% = 4.306 dari 7.091 trip yang dikeluarkan, sedangkan App 60,0% = 186.535 dari 311.007 trip yang disimpan (jumlah trip di balik setiap persentase ditampilkan di tabel kedua). Kolom <code>difference (points)</code> = <code>excluded (%)</code> − <code>kept (%)</code>, dalam poin persentase. Kedua kolom tidak diharapkan sama persis: kelompok yang dikeluarkan jauh lebih kecil, sehingga persentasenya wajar bergeser karena kebetulan. Besar pergeseran yang masih wajar dihitung di tabel kedua (kolom <code>chance range (± points)</code> = 1,96 × standard error porsi untuk 7.091 trip), dan kolom <code>beyond chance range</code> menandai kategori yang selisihnya lebih besar dari itu. Yang diuji adalah apakah selisihnya cukup besar untuk membuat tabel final condong ke kelompok tertentu.
</p>

```python
# Are the excluded trips different from the trips we keep? If their mix is the same,
# the exclusion does not bias the composition of the final table.
excluded_trips = df_trips[df_trips["exclude_trip"]]
for column in ["booking_channel", "pickup_zone_type", "payment_method", "trip_type"]:
    mix = pd.DataFrame({
        "excluded (%)": excluded_trips[column].value_counts(normalize=True) * 100,
        "kept (%)": df_trips_clean[column].value_counts(normalize=True) * 100,
    })
    mix["difference (points)"] = mix["excluded (%)"] - mix["kept (%)"]
    print(f"Mix by {column}: largest difference = {mix['difference (points)'].abs().max():.1f} points")
    display(mix.round(1))

negative_delay_values = df_trips.loc[df_trips["flag_negative_booking_delay"], "booking_to_meter_minutes"]
print("Negative booking-to-meter delays (minutes):")
display(negative_delay_values.describe().round(2).to_frame("booking_to_meter_minutes"))
```

```python
# The same comparison, now with the number of trips behind every percentage, and the size of a
# difference that can appear by chance alone in a group as small as the excluded trips.
n_excluded = len(excluded_trips)
n_kept = len(df_trips_clean)
print(f"Denominators: excluded = {n_excluded:,} trips, kept = {n_kept:,} trips")

for column in ["booking_channel", "pickup_zone_type", "payment_method", "trip_type"]:
    table = pd.DataFrame({
        "excluded (trips)": excluded_trips[column].value_counts(),
        "kept (trips)": df_trips_clean[column].value_counts(),
    }).fillna(0).astype(int)
    assert table["excluded (trips)"].sum() == n_excluded and table["kept (trips)"].sum() == n_kept
    table["excluded (%)"] = table["excluded (trips)"] / n_excluded * 100
    table["kept (%)"] = table["kept (trips)"] / n_kept * 100
    table["difference (points)"] = table["excluded (%)"] - table["kept (%)"]
    # If the excluded trips were a random draw of n_excluded trips, 95% of such draws would differ from the
    # kept share by less than 1.96 x standard error (normal approximation for a proportion).
    share = table["kept (%)"] / 100
    table["chance range (± points)"] = 1.96 * np.sqrt(share * (1 - share) / n_excluded) * 100
    table["beyond chance range"] = table["difference (points)"].abs() > table["chance range (± points)"]
    print(f"\n{column}:")
    display(table.round(2))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> trip yang dikeluarkan tidak berbeda dari trip yang disimpan: selisih komposisi paling besar hanya 1,0 poin (zona), 0,8 poin (metode bayar), 0,7 poin (channel), dan 0,2 poin (tipe trip). Hanya Airport yang selisihnya melewati batas kebetulan (<code>beyond chance range</code> = True): 12,28% vs 13,28%, selisih −1,00 poin, batas ±0,79 poin. Jadi trip Airport sedikit lebih jarang terkena eksklusi, tetapi selisih ini terlalu kecil untuk mengubah hasil: porsi Airport di tabel final hanya berubah +0,17% (perubahan relatif, 1.8.4). Delay negatif berkisar −6,00 sampai −1,00 menit (median −3,48 menit). Jadi eksklusi tidak membuat tabel final condong ke kelompok tertentu, dan dampaknya ke rata-rata di bawah 0,2% (1.8.4). <strong>Catatan keputusan:</strong> selisih beberapa menit bisa berasal dari jam yang tidak sinkron antara aplikasi dan sistem lama (brief menyebut masalah rekonsiliasi), sehingga jarak dan tarif trip tersebut mungkin tetap benar. Alternatif yang lebih lunak adalah hanya mengosongkan nilai delay. Eksklusi dipertahankan karena aturannya sederhana, dapat diaudit, tidak bias, dan dampaknya terukur kecil. Untuk ritase, <code>trip_count</code> asli tetap dipakai sehingga trip ini tetap terhitung sebagai pekerjaan pengemudi.
</p>


## <span style="color:#0061AF">1.9 Laporan Kualitas dan Export</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Laporan kualitas mendokumentasikan jumlah baris awal dan akhir, duplikat, missing value, flag, dan hasil validasi relasi. Tabel bersih dan laporan lalu diexport dan diverifikasi ulang.
</p>

### <span style="color:#0061AF">1.9.1 Laporan Kualitas Data Final</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Laporan ringkas jumlah baris, duplikat, missing value, dan flag per tabel. <strong>Cara membaca kolom:</strong> <code>rows_before_exclusion</code> = jumlah baris sebelum eksklusi, <code>rows_excluded</code> = baris yang dikeluarkan, dan <code>rows</code> = baris yang disimpan di tabel final (<code>rows_before_exclusion</code> − <code>rows_excluded</code>). Hanya tabel trip yang berkurang: 318.098 − 7.091 = 311.007. <code>missing_cells</code> dirinci di tabel <em>Missing cells: structural vs real</em> pada 1.9.2: pada trip, 256.110 sel kosong = 254.925 null struktural (<code>company_id</code> kosong pada trip non-eVoucher) + 1.185 sel kosong di kolom <code>*_clean</code> baru; pada request, 63.820 = 38.794 null struktural (<code>matched_trip_id</code> dan <code>wait_time_minutes</code> kosong pada request yang gagal) + 25.026 sel kosong di kolom <code>*_clean</code>. Missing value sebenarnya (<code>real_missing</code>) = 0 di semua tabel.
</p>

```python
quality_rows = []
for table_name, df in tables_final.items():
    key = primary_keys[table_name]
    quality_rows.append({
        "table": table_name,
        "rows_before_exclusion": len(tables[table_name]),
        "rows_excluded": len(tables[table_name]) - len(df),
        "rows": len(df),
        "columns": len(df.columns),
        "exact_duplicate_rows": int(df.duplicated().sum()),
        "missing_cells": int(df.isna().sum().sum()),
        "missing_primary_keys": int(df[key].isna().sum()),
        "duplicate_primary_keys": int(df[key].dropna().duplicated().sum()),
    })

quality_report = pd.DataFrame(quality_rows)
display(quality_report)
print("Relationship checks on pre-exclusion source-cleaned tables:")
display(relationship_report)
```

### <span style="color:#0061AF">1.9.2 Pengecekan Akhir Tabel Final</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Relasi dicek ulang pada tabel final, missing value sebenarnya dipisahkan dari null struktural, dan semua flag masalah dihitung dalam satu tabel.
</p>

```python
# 1) Relationship checks on the FINAL tables.
final_relationships = [
    ("Trips -> Drivers", tables_final["trips"], "driver_id", tables_final["drivers"], "driver_id"),
    ("Trips -> Shifts", tables_final["trips"], "shift_id", tables_final["shifts"], "shift_id"),
    ("Trips -> Clients", tables_final["trips"], "company_id", tables_final["clients"], "company_id"),
    ("Shifts -> Drivers", tables_final["shifts"], "driver_id", tables_final["drivers"], "driver_id"),
    ("Drivers -> Pools", tables_final["drivers"], "pool_id", tables_final["pools"], "pool_id"),
    ("Requests -> Trips", tables_final["requests"], "matched_trip_id", tables_final["trips"], "trip_id"),
]

final_relationship_rows = []
for label, child, child_key, parent, parent_key in final_relationships:
    unmatched = find_unmatched_keys(child, child_key, parent, parent_key)
    final_relationship_rows.append({
        "relationship": label,
        "unmatched_unique_ids": int(len(unmatched)),
    })

final_relationship_report = pd.DataFrame(final_relationship_rows)
print("Relationship checks on the FINAL tables:")
display(final_relationship_report)

# 2) Real missing values vs structural nulls vs empty cells of the new *_clean columns.
source_columns = {
    "clients": list(df_clients_raw.columns),
    "trips": list(df_trips_raw.columns),
    "shifts": list(df_shifts_raw.columns),
    "drivers": list(df_drivers_raw.columns),
    "pools": list(df_pools_raw.columns),
    "requests": list(df_requests_raw.columns),
}

trips_final = tables_final["trips"]
requests_final = tables_final["requests"]
structural_nulls = {
    # company_id is empty when the trip was not paid by eVoucher
    "trips": int(trips_final.loc[trips_final["payment_method"].ne("eVoucher Corporate"), "company_id"].isna().sum()),
    # matched_trip_id and wait_time_minutes are empty when the request was not fulfilled
    "requests": int(requests_final.loc[requests_final["status"].ne("Fulfilled"), ["matched_trip_id", "wait_time_minutes"]].isna().sum().sum()),
}

missing_rows = []
for table_name, df in tables_final.items():
    total_missing = int(df.isna().sum().sum())
    missing_in_source_columns = int(df[source_columns[table_name]].isna().sum().sum())
    structural = structural_nulls.get(table_name, 0)
    missing_rows.append({
        "table": table_name,
        "missing_cells_total": total_missing,
        "empty_cells_in_new_clean_columns": total_missing - missing_in_source_columns,
        "structural_nulls_expected": structural,
        "real_missing": missing_in_source_columns - structural,
    })

structural_null_report = pd.DataFrame(missing_rows)
print("Missing cells: structural vs real:")
display(structural_null_report)

# 3) How many rows does each problem flag mark?
flag_columns = {
    "trips": [
        "flag_negative_booking_delay", "flag_nonpositive_trip_duration", "flag_missing_trip_timestamp",
        "flag_negative_distance", "flag_negative_fare", "flag_distance_too_high", "flag_distance_too_low",
        "flag_distance_unreliable", "flag_meter_outside_shift", "flag_overlap_same_driver",
        "flag_any_trip_issue", "flag_outlier_distance_km", "flag_outlier_total_fare",
    ],
    "shifts": [
        "flag_invalid_shift_duration", "flag_negative_idle_time", "flag_idle_exceeds_duration",
        "flag_overlap_prev_shift", "flag_shift_before_join_date", "flag_trip_count_mismatch",
    ],
    "drivers": ["flag_join_after_data_end", "flag_first_shift_before_join"],
    "requests": ["flag_matched_trip_negative_delay", "flag_wait_at_floor"],
}

flag_rows = []
for table_name, columns in flag_columns.items():
    df = tables_final[table_name]
    for col in columns:
        flag_rows.append({
            "table": table_name,
            "flag": col,
            "rows_flagged": int(df[col].sum()),
            "percent_of_table": round(float(df[col].mean()) * 100, 2),
        })

cleaning_flag_summary = pd.DataFrame(flag_rows)
display(cleaning_flag_summary)
```

Laporan kualitas disimpan ke <code>outputs/reports/</code> (CSV, <code>utf-8-sig</code>, tanpa index). Tabel mentah dan tabel bersih diexport bersama-sama di <strong>PART 4 — EXPORTING</strong>, agar semua dataset ada di satu tempat dan bisa divalidasi sekaligus.

### <span style="color:#0061AF">2.2.1 Siapkan Tabel Eksplorasi</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Kolom bantu (jam, hari, flag) ditambahkan. Tidak ada baris yang dihapus.
</p>

```python
# Requests: one row = one customer request (unit of analysis for Questions 1 and 2).
req = df_requests.copy()
req["hour"] = req["requested_time"].dt.hour
req["date"] = req["requested_time"].dt.date
req["month"] = req["requested_time"].dt.strftime("%Y-%m")
req["weekday"] = req["requested_time"].dt.day_name()
req["is_failed"] = req["status"] != "Fulfilled"
req["is_no_driver"] = req["status"] == "Cancelled-No Driver"
req["is_timeout"] = req["status"] == "Cancelled-Timeout"

# Sanity checks: nothing was added or lost, and the status labels are what we expect.
assert len(req) == len(df_requests)
assert req["request_id"].is_unique
assert set(req["status"].unique()) == {"Fulfilled", "Cancelled-No Driver", "Cancelled-Timeout"}
assert (req["is_failed"] == (req["is_no_driver"] | req["is_timeout"])).all()

def request_rates(data, by):
    """Counts and failure rates (in %) for any grouping. Denominator = all requests in the group."""
    out = data.groupby(by, observed=True).agg(
        requests=("request_id", "size"),
        failed=("is_failed", "sum"),
        no_driver=("is_no_driver", "sum"),
        timeout=("is_timeout", "sum"),
    )
    out["failed_pct"] = out["failed"] / out["requests"] * 100
    out["no_driver_pct"] = out["no_driver"] / out["requests"] * 100
    out["timeout_pct"] = out["timeout"] / out["requests"] * 100
    return out

print(f"Requests: {len(req):,} | period: {req['requested_time'].min()} to {req['requested_time'].max()}")
```

```python
trips_eda = df_trips_clean.copy()
trips_eda["is_corporate"] = trips_eda["company_id"].notna()
trips_eda["hour"] = trips_eda["booking_time"].dt.hour

for column in ["trip_type", "booking_channel", "service_type_clean", "payment_method", "pickup_zone_type"]:
    print(f"Trips by {column}:")
    table = trips_eda.groupby(column).agg(
        trips=("trip_id", "size"),
        median_fare=("total_fare", "median"),
        total_fare=("total_fare", "sum"),
        median_distance_km=("distance_km_clean", "median"),
    )
    table["trip_share_pct"] = table["trips"] / len(trips_eda) * 100
    display(table)

# Reconcile trips and requests: every fulfilled request has one trip; the rest are Street Hail.
print(f"Fulfilled requests: {(req['status'] == 'Fulfilled').sum():,} | trips with a request: {trips_eda['trip_id'].isin(req['matched_trip_id']).sum():,}")
print(f"Trips without a request (Street Hail): {(~trips_eda['trip_id'].isin(req['matched_trip_id'])).sum():,} "
      f"| Street Hail trips: {(trips_eda['booking_channel'] == 'Street Hail').sum():,}")
```

### <span style="color:#0061AF">2.2.3 Apakah Skala Data Masuk Akal?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Sebelum menafsirkan utilisasi dan ritase, skala data diperiksa: berapa trip per pengemudi per hari, dan apakah ada pola yang terlalu rapi. Jika tabel trip hanya sampel atau data simulasi, angka absolut (mis. utilisasi 9,4%) tidak boleh dibandingkan langsung dengan kondisi nyata Blue Bird.
</p>


```python
# Is the scale of the data plausible for a taxi company? Simple ratios per average day.
days_in_data = N_DAYS_DATA = trips_eda["booking_time"].dt.normalize().nunique()
active_drivers = df_drivers["driver_id"].nunique()
scale = pd.Series({
    "days with trips": days_in_data,
    "drivers in bb_drivers": active_drivers,
    "shifts per day (average)": len(df_shifts_clean) / days_in_data,
    "trips per day (average)": len(trips_eda) / days_in_data,
    "trips per driver per day": len(trips_eda) / days_in_data / active_drivers,
    "requests per day (average)": len(df_requests) / days_in_data,
})
display(scale.round(2).to_frame("value"))

# Very regular patterns can mean the data is a sample or simulated. We check a few of them.
print("Share of shifts by clock-in hour (%):")
print((df_shifts_clean["clock_in_time"].dt.hour.value_counts(normalize=True).sort_index() * 100).round(1).to_dict())
print("Street Hail share of trips in each pickup zone (%):")
print((trips_eda.groupby("pickup_zone_type")["booking_channel"].apply(lambda s: (s == "Street Hail").mean() * 100)).round(1).to_dict())
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">

Beberapa pola juga sangat rapi: jam clock-in tepat 20,6% pada setiap jam 05–08 dan 3,5% pada setiap jam 18–22, dan porsi Street Hail hampir persis 15% di setiap zona. Pola serapi ini jarang muncul pada data operasional nyata, sehingga tabel kemungkinan berupa sampel atau data simulasi. <strong>Konsekuensi:</strong> perbandingan relatif (antar zona, jam, pool, dan metode bayar) tetap dapat dipakai, tetapi angka absolut seperti utilisasi 9,4% tidak boleh dibandingkan langsung dengan kondisi nyata Blue Bird (mis. target 70% pada 2018) sebelum kelengkapan data dikonfirmasi ke tim Data Engineering.

## <span style="color:#0061AF">2.3 Permintaan Layanan dan Karakteristik Trip</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Bagian ini melihat trip selesai: seberapa besar kualitas data meter memengaruhi angka rata-rata jarak dan delay. Ini menjadi dasar kepercayaan terhadap seluruh angka berikutnya.

### <span style="color:#0061AF">2.3.1 Seberapa Besar Data Buruk Mengubah Angka Operasional?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Rata-rata yang sama dihitung dengan dan tanpa nilai yang ditandai untuk menilai pengaruh kualitas data meter.
</p>

```python
quality_impact = pd.DataFrame({
    "raw": [
        trips_eda["distance_km"].mean(),
        trips_eda.loc[trips_eda["trip_type"] == "Reguler", "distance_km"].mean(),
        trips_eda["booking_to_meter_minutes"].mean(),
        df_requests["wait_time_minutes"].mean(),
    ],
    "clean": [
        trips_eda["distance_km_clean"].mean(),
        trips_eda.loc[trips_eda["trip_type"] == "Reguler", "distance_km_clean"].mean(),
        trips_eda["booking_to_meter_minutes_clean"].mean(),
        df_requests["wait_time_minutes_clean"].mean(),
    ],
}, index=["Average distance, all trips (km)", "Average distance, Reguler trips (km)",
          "Average booking-to-meter delay (min)", "Average request wait time (min)"])
quality_impact["difference"] = quality_impact["clean"] - quality_impact["raw"]
quality_impact["difference_pct"] = quality_impact["difference"] / quality_impact["raw"] * 100
display(quality_impact)

distance_by_zone = trips_eda.groupby("pickup_zone_type")[["distance_km", "distance_km_clean"]].mean()
display(distance_by_zone)
print(f"Trips with an unreliable distance: {trips_eda['flag_distance_unreliable'].sum():,} "
      f"({trips_eda['flag_distance_unreliable'].mean()*100:.2f}% of trips)")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> setelah 7.091 trip dengan jarak atau delay negatif dikeluarkan, masih ada 1.185 trip (0,38%) dengan jarak tidak andal (lebih dari 500 km atau kurang dari 1 km), dan trip ini menaikkan rata-rata jarak sekitar 18% (16,08 km mentah vs 13,20 km bersih). Rata-rata delay booking-ke-meter sama (5,01 menit) karena delay negatif sudah dikeluarkan. Karena itu kolom <code>*_clean</code> dipakai untuk jarak, delay, dan waktu tunggu.
</p>

### <span style="color:#0061AF">2.3.2 Apakah Tarif Mengikuti Jarak?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Jika argometer andal, tarif harus naik sejalan dengan jarak, dengan tarif per km yang berbeda menurut jenis layanan. Untuk setiap kombinasi tipe trip dan layanan dihitung garis tarif = bagian tetap + tarif per km tambahan (regresi linear sederhana) dan R-squared-nya. Tarif per km juga dibandingkan antar metode bayar untuk melihat apakah trip eVoucher ditagih lebih mahal. Hanya trip dengan jarak andal yang dipakai.
</p>


```python
# Does the fare follow the distance? Pricing consistency check on trips with a reliable distance.
fare_check = trips_eda[(~trips_eda["flag_distance_unreliable"]) & (trips_eda["distance_km_clean"] > 0)].copy()
fare_check["fare_per_km"] = fare_check["total_fare"] / fare_check["distance_km_clean"]
print(f"Trips used: {len(fare_check):,} of {len(trips_eda):,} (unreliable or zero distances left out)")

pricing_rows = []
for (trip_type, service), group in fare_check.groupby(["trip_type", "service_type_clean"]):
    slope, intercept = np.polyfit(group["distance_km_clean"], group["total_fare"], 1)   # fare = intercept + slope x km
    r = np.corrcoef(group["distance_km_clean"], group["total_fare"])[0, 1]
    pricing_rows.append({"trip_type": trip_type, "service": service, "trips": len(group),
                         "rp_per_extra_km": slope, "fixed_part_rp": intercept, "r_squared": r ** 2,
                         "median_fare_per_km": group["fare_per_km"].median()})
pricing = pd.DataFrame(pricing_rows).set_index(["trip_type", "service"])
display(pricing.round({"rp_per_extra_km": 0, "fixed_part_rp": 0, "r_squared": 3, "median_fare_per_km": 0}))

print("Median fare per km by payment method (Rp):")
fare_per_km_by_payment = fare_check.groupby("payment_method")["fare_per_km"].median()
display(fare_per_km_by_payment.round(0).to_frame("median_fare_per_km"))

# Plausibility of the meter: average speed between meter start and drop-off.
trip_hours = (fare_check["dropoff_time"] - fare_check["meter_start_time"]).dt.total_seconds() / 3600
speed_kmh = fare_check["distance_km_clean"] / trip_hours.where(trip_hours > 0)
display(speed_kmh.describe().round(1).to_frame("speed_km_per_hour"))
speed_by_hour = speed_kmh.groupby(fare_check["meter_start_time"].dt.hour).median()
print(f"Median speed by hour of meter start: from {speed_by_hour.min():.1f} km/h (hour {speed_by_hour.idxmin()}) "
      f"to {speed_by_hour.max():.1f} km/h (hour {speed_by_hour.idxmax()})")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> pada 309.822 trip dengan jarak andal (1.185 trip dengan jarak di atas 500 km atau di bawah 1 km disisihkan dari perhitungan ini saja, tetapi tetap ada di tabel final; trip berjarak 0 km tidak ada),, tarif hampir sepenuhnya ditentukan jarak: R-squared 0,97–0,98 di setiap kombinasi tipe trip dan layanan. Tarif per km tambahan berbeda menurut layanan (trip tipe Reguler: E-Cabbie Rp 4.478, layanan Reguler Rp 4.744, Silver Rp 5.877, Golden Rp 7.304), sesuai urutan kelas layanan. Median tarif per km trip eVoucher (Rp 6.638) <strong>tidak lebih tinggi</strong> dari Cash, Credit Card, dan MyBluebird App (Rp 6.781–6.797), sehingga tidak ada indikasi trip eVoucher ditagih lebih mahal. Kecepatan median 18,2 km/jam (5,8–70,7) wajar untuk lalu lintas kota, dan hampir rata sepanjang hari (median per jam 17,4–18,9 km/jam): tidak ada kemacetan jam sibuk yang terekam, sehingga lonjakan kegagalan 07–08 dan 17–18 tidak dijelaskan oleh lalu lintas. <strong>Makna:</strong> setelah jarak tidak andal disisihkan, argometer konsisten dan tarif tinggi dijelaskan oleh jarak dan kelas layanan, bukan oleh tagihan yang tidak wajar. <strong>Catatan:</strong> hubungan yang serapi ini juga sesuai dengan kemungkinan data simulasi (2.2.3).
</p>

### <span style="color:#0061AF">2.4.1 Seberapa Sering Request Terpenuhi atau Gagal?</span>

### <span style="color:#0061AF">2.4.2 Kegagalan per Jam</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Jumlah request (permintaan) dan tingkat kegagalan dilihat berdampingan, karena jam yang ramai belum tentu jam yang gagal.
</p>

```python
by_hour = request_rates(req, "hour")
overall_failed_pct = req["is_failed"].mean() * 100
display(by_hour[["requests", "failed", "failed_pct", "no_driver_pct", "timeout_pct"]])

# Define the hour groups FROM THE DATA (not from assumption):
#  - failure-peak hours = hours where the failure rate is more than 2x the overall rate
#  - demand-peak hours  = the 5 hours with the most requests
failure_peak_hours = sorted(by_hour.index[by_hour["failed_pct"] > 2 * overall_failed_pct])
demand_peak_hours = sorted(by_hour["requests"].nlargest(5).index)
print(f"Overall failure rate: {overall_failed_pct:.2f}%")
print("Failure-peak hours (rate > 2x overall):", failure_peak_hours)
print("Demand-peak hours (top 5 by request volume):", demand_peak_hours)
print("Overlap between the two groups:", sorted(set(failure_peak_hours) & set(demand_peak_hours)))

req["period"] = np.where(
    req["hour"].isin(failure_peak_hours), "Failure-peak hours",
    np.where(req["hour"].isin(demand_peak_hours), "Demand-peak hours", "Other hours"),
)
display(request_rates(req, "period"))
```

### <span style="color:#0061AF">2.4.3 Kegagalan per Zona, Sumber Request, Bulan, dan Hari</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Tingkat kegagalan dibandingkan antar zona, channel request, bulan, dan hari untuk melihat faktor mana yang berbeda.
</p>

```python
print("By pickup zone:")
display(request_rates(req, "pickup_zone_type").sort_values("failed_pct", ascending=False))
print("By request source:")
display(request_rates(req, "request_source"))
print("By month:")
display(request_rates(req, "month"))
weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
print("By weekday:")
display(request_rates(req, "weekday").reindex(weekday_order))
```

### <span style="color:#0061AF">2.4.4 Kombinasi Zona × Jam</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Zona dan jam dikombinasikan untuk mengetahui apakah masalahnya ada di CBD, jam sibuk, atau keduanya.
</p>

```python
zone_by_period = req.pivot_table(
    index="pickup_zone_type", columns="period", values="is_failed", aggfunc="mean"
) * 100
zone_by_period_n = req.pivot_table(
    index="pickup_zone_type", columns="period", values="is_failed", aggfunc="size"
)
print("Failure rate (%) by zone and hour group:")
display(zone_by_period)
print("Number of requests behind each cell:")
display(zone_by_period_n)

# Same crossing, but for the individual failure-peak hours.
cbd_hours = request_rates(req[req["pickup_zone_type"] == "CBD"], "hour")[["requests", "failed_pct"]]
print("CBD only, failure rate by hour:")
display(cbd_hours.T)

# How much of ALL failures sit in CBD + failure-peak hours?
cbd_peak = (req["pickup_zone_type"] == "CBD") & (req["period"] == "Failure-peak hours")
print(f"CBD requests in failure-peak hours: {cbd_peak.sum():,} ({cbd_peak.mean()*100:.1f}% of all requests)")
print(f"They contain {req.loc[cbd_peak, 'is_failed'].sum():,} of {req['is_failed'].sum():,} failures "
      f"({req.loc[cbd_peak, 'is_failed'].sum() / req['is_failed'].sum() * 100:.1f}% of all failures)")
print(f"Failure rate: CBD + failure-peak = {req.loc[cbd_peak, 'is_failed'].mean()*100:.1f}% "
      f"vs everything else = {req.loc[~cbd_peak, 'is_failed'].mean()*100:.1f}%")

# Statistical check. With ~290k rows almost any difference is 'significant', so we also report the effect size.
try:
    from scipy.stats import chi2_contingency
    table_2x2 = pd.crosstab(cbd_peak, req["is_failed"])
    chi2, p_value, _, _ = chi2_contingency(table_2x2)
    a, b = table_2x2.loc[True, True], table_2x2.loc[True, False]
    c, d = table_2x2.loc[False, True], table_2x2.loc[False, False]
    print(f"Chi-square p-value: {max(p_value, 1e-300):.3g} (effectively 0) | odds ratio: {(a / b) / (c / d):.1f} (failure odds, CBD+peak vs rest)")
except ImportError:
    print("scipy not installed: skip the chi-square test (pip install scipy).")
```

### <span style="color:#0061AF">2.4.5 Jenis Kegagalan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<code>Cancelled-No Driver</code> mengarah ke sisi pasokan. <code>Cancelled-Timeout</code> bisa berarti pengemudi lambat atau pelanggan menyerah.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Cara membaca dua tabel di bawah.</strong> Keduanya memakai penyebut yang berbeda:
</p>

| Tabel | Kolom | Pembilang | Penyebut | Pertanyaan yang dijawab |
|---|---|---|---|---|
| 1 (zona × kelompok jam) | <code>no_driver_pct</code>, <code>timeout_pct</code> | request Cancelled-No Driver / Cancelled-Timeout | <strong>semua request</strong> di zona dan kelompok jam itu (kolom <code>requests</code>) | Seberapa sering request gagal karena jenis ini? |
| 2 (kelompok jam) | <code>no_driver_share_of_failures_pct</code> | request Cancelled-No Driver | <strong>request yang gagal saja</strong> di kelompok jam itu | Dari request yang gagal, berapa persen yang No Driver? (sisanya Timeout) |

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Asal angka tabel 2.</strong> Pembilang dan penyebutnya tercetak di tabel kelompok jam pada <strong>2.4.2</strong> (tabel kedua di sel kode 2.4.2, hasil <code>display(request_rates(req, "period"))</code>): kolom <code>no_driver</code> adalah pembilang, kolom <code>failed</code> adalah penyebut, dan <code>no_driver</code> + <code>timeout</code> = <code>failed</code>.
</p>

| Kelompok jam | <code>no_driver</code> (2.4.2) | <code>failed</code> (2.4.2) | <code>no_driver_share_of_failures_pct</code> (2.4.5) |
|---|---|---|---|
| Demand-peak hours | 4.364 | 6.257 | 69,75% |
| Failure-peak hours | 6.139 | 8.838 | 69,46% |
| Other hours | 2.991 | 4.302 | 69,53% |

```python
split = request_rates(req, ["pickup_zone_type", "period"])[["requests", "no_driver_pct", "timeout_pct"]]
display(split.loc[["CBD", "Airport"]])
print("Share of failures that are 'No Driver', by group (denominator = failed requests in that group):")
failed_only = req[req["is_failed"]]
display((failed_only.groupby("period")["is_no_driver"].mean() * 100).to_frame("no_driver_share_of_failures_pct"))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> di CBD pada jam puncak kegagalan, 25,50% dari 20.436 request gagal karena tidak ada pengemudi dan 11,05% karena timeout, dibanding 4,27% dan 1,80% pada jam lain di CBD (tabel 1). Airport hampir tidak berubah antar kelompok jam (No Driver 1,24–1,43%, Timeout 0,61–1,00%). Namun komposisi jenis kegagalan tetap sama di ketiga kelompok jam: sekitar 70% No Driver (69,46–69,75%) dan sisanya Timeout (tabel 2). Artinya, pada jam puncak yang naik adalah <em>jumlah</em> kegagalan, sedangkan <em>jenisnya</em> tetap sama, sehingga jenis kegagalan tidak bisa membedakan penyebab jam puncak; penyebabnya perlu dicari di faktor lain seperti pasokan pengemudi, kesiapan, atau dispatch (2.5).
</p>

### <span style="color:#0061AF">2.4.6 Waktu Tunggu Request yang Terpenuhi</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Waktu tunggu hanya ada pada request terpenuhi. Dipakai <code>wait_time_minutes_clean</code> dan nilainya bersifat indikatif karena ada batas bawah 0,5 menit.

```python
fulfilled = req[req["status"] == "Fulfilled"]
print("Wait time (minutes), fulfilled requests, using wait_time_minutes_clean:")
display(fulfilled["wait_time_minutes_clean"].describe().to_frame("wait_time_minutes_clean"))
print(f"Fulfilled requests with wait_time_minutes_clean empty: {fulfilled['wait_time_minutes_clean'].isna().sum():,}")
print("Median wait (min) by hour group and zone:")
display(fulfilled.pivot_table(index="pickup_zone_type", columns="period",
                              values="wait_time_minutes_clean", aggfunc="median"))
```

### <span style="color:#0061AF">2.4.7 Besar Masalah Request Gagal (Ilustrasi dengan Asumsi)</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Berapa request dan tarif yang dipertaruhkan jika kegagalan jam sibuk setara jam lain pada zona yang sama? Asumsi: (1) request berlebih akan terpenuhi, (2) tiap request menghasilkan median tarif trip zona yang sama, (3) periode 2023-10-01 sampai 2023-12-30. Ini batas atas ilustratif, bukan prakiraan.
</p>

```python
fare_of_trip = trips_eda.set_index("trip_id")["total_fare"]
req["trip_fare"] = req["matched_trip_id"].map(fare_of_trip)           # fare only exists for fulfilled requests
median_fare_by_zone = req.groupby("pickup_zone_type")["trip_fare"].median()

rows = []
for zone in zones_in_order:
    zone_requests = req[req["pickup_zone_type"] == zone]
    in_peak = zone_requests["period"] == "Failure-peak hours"
    peak_rate = zone_requests.loc[in_peak, "is_failed"].mean()
    reference_rate = zone_requests.loc[~in_peak, "is_failed"].mean()
    excess = in_peak.sum() * (peak_rate - reference_rate)
    rows.append({
        "zone": zone,
        "peak_requests": int(in_peak.sum()),
        "peak_failure_pct": peak_rate * 100,
        "reference_failure_pct": reference_rate * 100,
        "excess_failed_requests": excess,
        "median_fare": median_fare_by_zone[zone],
        "fare_at_stake_rp": excess * median_fare_by_zone[zone],
    })
impact = pd.DataFrame(rows).set_index("zone")
display(impact.round(1))
total_failed = int(req["is_failed"].sum())
print(f"Total failed requests: {total_failed:,}")
print(f"Excess failures in the failure-peak hours (all zones): {impact['excess_failed_requests'].sum():,.0f} "
      f"= {impact['excess_failed_requests'].sum() / total_failed * 100:.1f}% of all failures")
print(f"  of which CBD: {impact.loc['CBD', 'excess_failed_requests']:,.0f} "
      f"= {impact.loc['CBD', 'excess_failed_requests'] / impact['excess_failed_requests'].sum() * 100:.0f}% of the excess")
print(f"Fare at stake (all zones, median fare, 3 months): Rp {impact['fare_at_stake_rp'].sum():,.0f}  "
      f"vs total fare of all trips: Rp {trips_eda['total_fare'].sum():,.0f} "
      f"({impact['fare_at_stake_rp'].sum() / trips_eda['total_fare'].sum() * 100:.1f}%)")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> jika jam puncak kegagalan sama rendahnya dengan jam lain di zona yang sama, ada 6.963 request lebih sedikit yang gagal (35,9% dari 19.397 request gagal), dan <strong>91% di antaranya (6.346 request) ada di CBD</strong>. Dengan median tarif per zona (kolom <code>median_fare</code>), nilainya sekitar Rp 376 juta (Rp 375.898.176) dalam 3 bulan, hanya 1,5% dari total tarif Rp 25,9 miliar (Rp 25.912.927.500). Nilai langsung relatif kecil, sedangkan argumen yang lebih kuat adalah 6.346 request CBD yang tidak terlayani dan dampaknya pada reputasi layanan. Angka ini adalah request, bukan jumlah pelanggan, karena satu pelanggan bisa membuat lebih dari satu request dan data tidak memuat ID pelanggan. Rp 376 juta adalah batas atas, karena mengasumsikan setiap request berlebih akan terpenuhi dengan median tarif zonanya.
</p>

### <span style="color:#0061AF">2.4.8 App vs Call Center: Perbedaan Nyata atau Komposisi?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
2.4.3 menunjukkan request App lebih sering gagal daripada Call Center. Di sini dicek apakah selisih itu tetap ada di dalam setiap zona dan kelompok jam, atau hanya karena App lebih banyak dipakai di CBD dan jam sibuk.
</p>


```python
# Is the App failing more often than the Call Center everywhere, or only because of where/when App requests come from?
source_zone = req.pivot_table(index="pickup_zone_type", columns="request_source", values="is_failed", aggfunc="mean").loc[zones_in_order] * 100
source_zone["App minus Call Center (points)"] = source_zone["App"] - source_zone["Call Center"]
print("Failure rate (%) by zone and request source (denominator = requests of that source in that zone):")
display(source_zone.round(2))
source_period = req.pivot_table(index="period", columns="request_source", values="is_failed", aggfunc="mean") * 100
source_period["App minus Call Center (points)"] = source_period["App"] - source_period["Call Center"]
print("Failure rate (%) by hour group and request source:")
display(source_period.round(2))
print("Share of each zone's requests that come from the App (%):")
display((req.groupby("pickup_zone_type")["request_source"].apply(lambda s: (s == "App").mean() * 100)).loc[zones_in_order].round(1).to_frame("app_share_pct"))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> request App lebih sering gagal daripada Call Center di <strong>setiap</strong> zona: selisihnya +0,36 poin (Airport) sampai +2,04 poin (CBD), misalnya CBD 11,66% vs 9,62% (tabel 1). Selisih ini juga ada di setiap kelompok jam: +0,79 poin pada jam puncak permintaan, +3,79 poin pada jam puncak kegagalan (22,45% vs 18,66%), dan +0,99 poin pada jam lain (tabel 2). Porsi request dari App hampir sama di semua zona, 70,6–71,2% (tabel 3), sehingga selisih ini bukan efek komposisi zona. <strong>Hipotesis (belum diuji):</strong> proses pencocokan otomatis di aplikasi (batas waktu dan penugasan) mungkin berbeda dengan operator call center yang bisa mencari pengemudi secara aktif; menguji ini membutuhkan log dispatch, yang tidak ada di data. Temuan ini relevan karena brief mengaitkan pertumbuhan pendapatan semester I 2024 dengan pertumbuhan pengguna aplikasi MyBluebird, padahal di data ini kanal App justru lebih sering gagal. Temuan ini juga konsisten dengan 2.5.1: pada pagi hari jumlah pengemudi yang bertugas naik, tetapi kegagalan tetap tinggi, sehingga kekurangan jumlah pengemudi saja tidak menjelaskan kegagalan pagi; peran proses dispatch masih hipotesis.
</p>

## <span style="color:#0061AF">2.5 Pengemudi dan Shift</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Setelah mengetahui <em>di mana dan kapan</em> request gagal, pertanyaan berikutnya adalah sisi pasokan: apakah pengemudi cukup, berapa trip yang mereka dapat, dan seberapa idle mereka? Unit analisis: satu shift pengemudi.
</p>

### <span style="color:#0061AF">2.5.1 Apakah Pasokan Pengemudi per Jam Cukup?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Untuk setiap jam dihitung rata-rata jumlah shift yang aktif pada jam itu (shift dihitung jika jam clock-in sampai jam clock-out menyentuh jam tersebut, sehingga shift yang hanya aktif sebagian jam tetap dihitung 1) pada 90 hari penuh 2023-10-01 sampai 2023-12-29. Satu shift = satu pengemudi per hari. Label jam adalah awal jam: 17 = 17:00–17:59, lalu dibandingkan dengan jumlah request. Ini pola yang teramati, bukan bukti sebab.
</p>

```python
FIRST_DAY = pd.Timestamp("2023-10-01")
N_DAYS = 90                      # 2023-10-01 ... 2023-12-29 = 31 + 30 + 29 full days
N_SLOTS = N_DAYS * 24

def drivers_on_shift_by_hour(shifts):
    """Average number of shifts that are 'on duty' in each hour of the day (0-23)."""
    start_slot = ((shifts["clock_in_time"].dt.floor("h") - FIRST_DAY) // pd.Timedelta(hours=1)).to_numpy()
    end_slot = ((shifts["clock_out_time"].dt.floor("h") - FIRST_DAY) // pd.Timedelta(hours=1)).to_numpy() + 1
    size = N_SLOTS + 100
    assert start_slot.min() >= 0 and start_slot.max() < N_SLOTS and end_slot.max() < size
    change = np.bincount(start_slot, minlength=size) - np.bincount(end_slot, minlength=size)
    on_duty = np.cumsum(change)[:N_SLOTS]
    return pd.Series(on_duty.reshape(N_DAYS, 24).mean(axis=0), index=range(24))

full_days_req = req[req["requested_time"] < pd.Timestamp("2023-12-30")]
supply_demand = pd.DataFrame({
    "drivers_on_shift": drivers_on_shift_by_hour(df_shifts_clean),
    "requests_per_hour": full_days_req.groupby("hour").size() / N_DAYS,
    "failed_pct": request_rates(full_days_req, "hour")["failed_pct"],
})
supply_demand["requests_per_driver"] = supply_demand["requests_per_hour"] / supply_demand["drivers_on_shift"]
display(supply_demand)

corr = supply_demand["requests_per_driver"].corr(supply_demand["failed_pct"])
print(f"Correlation across the 24 hours: requests_per_driver vs failure rate = {corr:.2f}")
print("Hours with fewer than 400 drivers on shift:", list(supply_demand.index[supply_demand["drivers_on_shift"] < 400]))
```

```python
# The same view for the CBD only: drivers whose dominant zone is CBD vs requests picked up in the CBD.
cbd_supply_demand = pd.DataFrame({
    "cbd_drivers_on_shift": drivers_on_shift_by_hour(df_shifts_clean[df_shifts_clean["dominant_zone_worked"] == "CBD"]),
    "cbd_requests_per_hour": full_days_req[full_days_req["pickup_zone_type"] == "CBD"].groupby("hour").size() / N_DAYS,
    "cbd_failed_pct": request_rates(full_days_req[full_days_req["pickup_zone_type"] == "CBD"], "hour")["failed_pct"],
})
cbd_supply_demand["cbd_requests_per_driver"] = (
    cbd_supply_demand["cbd_requests_per_hour"] / cbd_supply_demand["cbd_drivers_on_shift"]
)
display(cbd_supply_demand.T)
print("Correlation (24 hours): CBD requests per driver vs CBD failure rate =",
      round(cbd_supply_demand["cbd_requests_per_driver"].corr(cbd_supply_demand["cbd_failed_pct"]), 2))
```

### <span style="color:#0061AF">2.5.2 Utilisasi Armada per Shift</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Unit analisis: satu shift pengemudi. Brief hanya mendefinisikan <code>ritase_target_met</code> sebagai penanda apakah target ritase harian tercapai, tanpa menyebut angka targetnya. Dari data, flag ini bernilai 1 tepat ketika <code>trip_count &gt;= 6</code> di setiap shift (dibuktikan di 1.7.5, Check 5). Jadi batas 6 trip adalah aturan yang tersirat di data, bukan target yang kami tetapkan dan bukan target resmi dari brief. Pada output di bawah, "Shifts that met the target (&gt;= 6 trips)" berarti shift dengan flag <code>ritase_target_met</code> = 1.
</p>

```python
# Add driver and pool attributes with .map() so the number of rows can NOT change.
shifts_eda = df_shifts_clean.copy()
shifts_eda["pool_id"] = shifts_eda["driver_id"].map(df_drivers.set_index("driver_id")["pool_id"])
shifts_eda["employment_status"] = shifts_eda["driver_id"].map(df_drivers.set_index("driver_id")["employment_status"])
shifts_eda["vehicle_type_assigned"] = shifts_eda["driver_id"].map(df_drivers.set_index("driver_id")["vehicle_type_assigned"])
shifts_eda["start_hour"] = shifts_eda["clock_in_time"].dt.hour
shifts_eda["start_group"] = np.where(shifts_eda["start_hour"] < 12, "Morning start", "Evening start")
shifts_eda["idle_share_pct"] = shifts_eda["total_idle_minutes"] / shifts_eda["shift_duration_minutes"] * 100
shifts_eda["zero_trip_shift"] = shifts_eda["trip_count"] == 0

assert len(shifts_eda) == len(df_shifts_clean)
assert shifts_eda[["pool_id", "employment_status", "vehicle_type_assigned"]].notna().all().all()
print("Shift start hours present:", [int(h) for h in sorted(shifts_eda["start_hour"].unique())])

print("Trips per shift (trip_count):")
display(shifts_eda["trip_count"].describe().to_frame("trip_count"))
display(shifts_eda["trip_count"].value_counts().sort_index().to_frame("shifts").assign(
    share_pct=lambda t: t["shifts"] / len(shifts_eda) * 100))
print(f"Shifts that met the target (>= 6 trips): {shifts_eda['ritase_target_met'].sum():,} of {len(shifts_eda):,} "
      f"= {shifts_eda['ritase_target_met'].mean()*100:.2f}%")
print(f"Shifts with 0 trips: {shifts_eda['zero_trip_shift'].sum():,} = {shifts_eda['zero_trip_shift'].mean()*100:.1f}%")
print("Shift duration and idle time:")
display(shifts_eda[["shift_duration_minutes", "total_idle_minutes", "idle_share_pct"]].describe())
```

```python
def shift_summary(data, by):
    """One row per group. Unit = driver shift. Target = trip_count >= 6."""
    out = data.groupby(by, observed=True).agg(
        shifts=("shift_id", "size"),
        drivers=("driver_id", "nunique"),
        avg_trips_per_shift=("trip_count", "mean"),
        zero_trip_pct=("zero_trip_shift", lambda s: s.mean() * 100),
        target_met_pct=("ritase_target_met", lambda s: s.mean() * 100),
        avg_idle_share_pct=("idle_share_pct", "mean"),
        avg_duration_min=("shift_duration_minutes", "mean"),
    )
    return out

for column in ["pool_id", "dominant_zone_worked", "start_group", "employment_status", "vehicle_type_assigned"]:
    print(f"Shift summary by {column}:")
    display(shift_summary(shifts_eda, column))
```

### <span style="color:#0061AF">2.5.3 Pool: Ukuran, Lokasi, dan Beban Kerja</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Atribut pool diambil dari <code>df_pools</code> (8 baris) dengan <code>.map()</code> agar jumlah baris tidak berubah.
</p>

```python
pool_table = shift_summary(shifts_eda, "pool_id").join(
    df_pools.set_index("pool_id")[["region_area", "distance_to_soetta_km", "distance_to_cbd_km", "primary_zone_affinity"]]
)
# Trips completed per driver over the whole period (a simple ritase view per driver).
pool_table["trips_per_driver_total"] = (
    shifts_eda.groupby("pool_id")["trip_count"].sum() / pool_table["drivers"]
)
display(pool_table)

# Does the zone a pool's drivers work match the pool's label? (checks the 'zone affinity' attribute)
print("Share of shifts by dominant zone, per pool (rows add to 1):")
display(pd.crosstab(shifts_eda["pool_id"], shifts_eda["dominant_zone_worked"], normalize="index").round(2))

# Are requests where the drivers are? Share of requests by zone vs share of shifts by dominant zone.
zone_balance = pd.DataFrame({
    "request_share_pct": req["pickup_zone_type"].value_counts(normalize=True) * 100,
    "shift_share_pct": shifts_eda["dominant_zone_worked"].value_counts(normalize=True) * 100,
})
zone_balance["avg_trips_per_shift"] = shifts_eda.groupby("dominant_zone_worked")["trip_count"].mean()
display(zone_balance)
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Cara membaca tabel 1:</strong> satu baris = satu pool. <code>avg_trips_per_shift</code> = rata-rata trip per shift, <code>zero_trip_pct</code> = persen shift tanpa trip, <code>target_met_pct</code> = persen shift dengan flag <code>ritase_target_met</code> = 1 (≥ 6 trip, lihat 2.5.2), dan <code>trips_per_driver_total</code> = total trip per pengemudi selama periode data. Penyebut semua persen adalah jumlah shift pool tersebut (kolom <code>shifts</code>).
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> (1) <strong>Pool berbeda cukup jelas.</strong> PL-CGK paling rendah: 0,97 trip per shift dan 40,47% shift tanpa trip, sedangkan tiga pool berlabel CBD-heavy (PL-KJT, PL-KLG, PL-MPG) paling tinggi: 1,39–1,40 trip per shift dan 26,34–26,82% shift tanpa trip. Pool lain berada di 1,24–1,26 trip per shift. Durasi shift hampir sama di semua pool (509,74–510,13 menit), jadi perbedaan ini bukan karena shift yang lebih pendek. (2) <strong>Komposisi zona tiap pool sangat seragam (tabel 2).</strong> PL-CGK (Airport-heavy) 55% shift-nya di Airport; tiga pool CBD-heavy masing-masing 45% di CBD; empat pool Mixed punya komposisi yang persis sama (Airport 20%, CBD 25%, Mall 20%, Other 15%, Residential 20%). Pola serapi ini menunjukkan komposisi zona seperti kuota per pool, dan pool dengan porsi Airport terbesar adalah pool dengan trip per shift terendah. (3) <strong>Pasokan tidak sebanding dengan permintaan per zona (tabel 3).</strong> Airport menerima 12,63% request tetapi menyerap 21,84% shift, dengan hanya 0,67 trip per shift; sebaliknya CBD menerima 39,30% request tetapi hanya 30,98% shift, dengan 1,70 trip per shift. <strong>Catatan:</strong> zona pada shift (<code>dominant_zone_worked</code>) adalah zona tempat pengemudi paling banyak bekerja, sehingga ini asosiasi, bukan bukti bahwa zona menyebabkan perbedaan; apakah perbedaan pool berasal dari pool itu sendiri atau dari komposisi zonanya diuji di 2.5.5, 2.5.6, dan 2.5.14.
</p>

### <span style="color:#0061AF">2.5.4 Definisi Ritase: Satu Shift = Satu Hari Kerja Pengemudi?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Brief mendefinisikan ritase sebagai jumlah trip per pengemudi per hari. Jika setiap pengemudi punya paling banyak satu shift per hari, shift dan driver-day sama.
</p>

```python
shifts_per_driver_day = shifts_eda.groupby(["driver_id", "shift_date"]).size()
print("Shifts per driver per day (value : number of driver-days):")
print(shifts_per_driver_day.value_counts().sort_index().to_dict())
assert shifts_per_driver_day.max() == 1, "Some drivers have 2 shifts in a day: sum trip_count per driver-day first"
print("OK: one shift = one driver-day, so trip_count per shift IS the daily ritase.")

# Average ritase per DRIVER (each driver counts once, so drivers with many shifts do not dominate).
ritase_per_driver = shifts_eda.groupby("driver_id")["trip_count"].mean()
display(ritase_per_driver.describe().to_frame("avg_trips_per_day_per_driver"))
```

### <span style="color:#0061AF">2.5.5 Mengapa Pool Berbeda: Pool atau Zona Kerja?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Rata-rata nyata setiap pool dibandingkan dengan rata-rata yang <em>diharapkan</em> jika setiap shift menghasilkan rata-rata semua pool pada zonanya (standarisasi langsung).
</p>

```python
print("Average trips per shift: pool x dominant zone (compare the rows inside one column):")
display(shifts_eda.pivot_table(index="pool_id", columns="dominant_zone_worked",
                               values="trip_count", aggfunc="mean").round(2))

zone_average = shifts_eda.groupby("dominant_zone_worked")["trip_count"].mean()
shifts_eda["expected_trips_from_zone"] = shifts_eda["dominant_zone_worked"].map(zone_average)
pool_vs_expected = shifts_eda.groupby("pool_id").agg(
    actual_avg_trips=("trip_count", "mean"),
    expected_from_zone_mix=("expected_trips_from_zone", "mean"),
)
pool_vs_expected["gap"] = pool_vs_expected["actual_avg_trips"] - pool_vs_expected["expected_from_zone_mix"]
display(pool_vs_expected.round(3))
print("Largest absolute gap between actual and zone-mix expectation:",
      round(pool_vs_expected["gap"].abs().max(), 3), "trips per shift")
print("Spread between best and worst pool (actual):",
      round(pool_vs_expected["actual_avg_trips"].max() - pool_vs_expected["actual_avg_trips"].min(), 2), "trips per shift")
```

### <span style="color:#0061AF">2.5.6 Apakah <code>dominant_zone_worked</code> Hasil Pekerjaan atau Penugasan?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Jika zona dominan dihitung dari trip yang dikerjakan pengemudi, maka "shift CBD punya lebih banyak trip" sebagian bersifat sirkular. Diuji apakah zona dominan shift sama dengan zona penjemputan tersering dari trip shift itu.
</p>

```python
modal_zone_of_shift = trips_eda.groupby("shift_id")["pickup_zone_type"].agg(lambda s: s.value_counts().index[0])
shifts_with_trips = shifts_eda[shifts_eda["trip_count"] > 0].copy()
shifts_with_trips["modal_pickup_zone"] = shifts_with_trips["shift_id"].map(modal_zone_of_shift)
match_rate = (shifts_with_trips["modal_pickup_zone"] == shifts_with_trips["dominant_zone_worked"]).mean()
print(f"Shifts with trips: {len(shifts_with_trips):,}")
print(f"dominant_zone_worked equals the most common pickup zone of the shift's own trips: {match_rate * 100:.1f}%")
print(f"Shifts with 0 trips still have a dominant zone: {(shifts_eda['trip_count'] == 0).sum():,} shifts")
print("Dominant zone of shifts with 0 trips (% of those shifts) - the label cannot come from trips here:")
print((shifts_eda.loc[shifts_eda["trip_count"] == 0, "dominant_zone_worked"].value_counts(normalize=True) * 100).round(1).to_dict())
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> pada 85,3% shift yang punya trip, zona dominan sama dengan zona penjemputan tersering, sehingga zona dominan sebagian merupakan <em>hasil</em> pekerjaan. Namun 77.393 shift tanpa trip tetap punya zona dominan (36,4% di antaranya Airport), dan label ini tidak mungkin berasal dari trip. Artinya label zona juga mencerminkan penugasan atau posisi tunggu pengemudi, bukan murni hasil. Karena itu efek zona dibaca sebagai asosiasi kuat, bukan sebab yang terbukti. Yang pasti: perbedaan pool hilang setelah komposisi zona diperhitungkan.
</p>


### <span style="color:#0061AF">2.5.7 Apakah Ritase Benar-Benar Turun? Tren Mingguan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Brief menduga ritase menurun. Dengan data tiga bulan, tren diuji di dalam periode tersebut memakai minggu penuh (Senin 2023-10-02 sampai Minggu 2023-12-24).
</p>

```python
shifts_eda["week_start"] = shifts_eda["shift_date"].dt.to_period("W").dt.start_time
req["week_start"] = req["requested_time"].dt.to_period("W").dt.start_time
full_weeks = pd.date_range("2023-10-02", "2023-12-18", freq="W-MON")

weekly = pd.DataFrame({
    "avg_trips_per_shift": shifts_eda.groupby("week_start")["trip_count"].mean(),
    "zero_trip_shift_pct": shifts_eda.groupby("week_start")["zero_trip_shift"].mean() * 100,
    "request_failure_pct": req.groupby("week_start")["is_failed"].mean() * 100,
    "requests": req.groupby("week_start").size(),
}).loc[full_weeks]
display(weekly)

week_number = np.arange(len(weekly))
for column in ["avg_trips_per_shift", "request_failure_pct"]:
    slope = np.polyfit(week_number, weekly[column], 1)[0]
    print(f"{column}: change per week = {slope:+.4f} | first week {weekly[column].iloc[0]:.3f} -> last week {weekly[column].iloc[-1]:.3f} "
          f"| min {weekly[column].min():.3f} | max {weekly[column].max():.3f}")
```

### <span style="color:#0061AF">2.5.8 Apa yang Diukur <code>total_idle_minutes</code>?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Diuji apakah idle time = durasi shift dikurangi waktu trip (booking sampai drop-off).
</p>

```python
trip_minutes = (trips_eda["dropoff_time"] - trips_eda["booking_time"]).dt.total_seconds() / 60
busy_per_shift = trip_minutes.groupby(trips_eda["shift_id"]).sum()

idle_check = shifts_eda[["shift_id", "trip_count", "shift_duration_minutes", "total_idle_minutes"]].copy()
idle_check["busy_minutes_from_trips"] = idle_check["shift_id"].map(busy_per_shift).fillna(0)
idle_check["gap_min"] = (idle_check["shift_duration_minutes"] - idle_check["total_idle_minutes"]) - idle_check["busy_minutes_from_trips"]
display(idle_check["gap_min"].describe().to_frame("gap_min (duration - idle - trip time)"))
for tolerance in [0.2, 1, 5]:
    print(f"Shifts where idle is consistent with trip time within {tolerance} min: {(idle_check['gap_min'].abs() <= tolerance).mean() * 100:.1f}%")

# Average idle per trip count: how many idle minutes does each extra trip remove?
idle_by_trips = shifts_eda.groupby("trip_count")["total_idle_minutes"].agg(["size", "mean"])
display(idle_by_trips.round(1))
slope_idle = np.polyfit(shifts_eda["trip_count"], shifts_eda["total_idle_minutes"], 1)[0]
print(f"Each extra trip reduces idle time by about {abs(slope_idle):.0f} minutes.")

idle_summary = shifts_eda.groupby(["dominant_zone_worked", "start_group"]).agg(
    shifts=("shift_id", "size"),
    avg_idle_minutes=("total_idle_minutes", "mean"),
    avg_idle_share_pct=("idle_share_pct", "mean"),
    avg_trips=("trip_count", "mean"),
)
display(idle_summary.round(1))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> <code>total_idle_minutes</code> adalah angka turunan: durasi shift dikurangi waktu trip dari booking sampai drop-off. Rumus ini cocok dalam 5 menit untuk 93,6% shift. Setiap tambahan satu trip mengurangi sekitar 39 menit idle, dan shift 6 trip pun masih sekitar 54% idle. Idle tinggi karena trip per shift sedikit, bukan karena trip panjang. Mengurangi idle berarti membawa lebih banyak trip ke setiap pengemudi (<em>demand matching</em>).
</p>

### <span style="color:#0061AF">2.5.9 Variabel yang Dipakai dan Tidak Dipakai untuk Uji Statistik</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Sel di bawah menambahkan atribut pengemudi ke setiap shift dengan <code>.map()</code>, sehingga jumlah baris tidak berubah: <code>driver_rating</code>, <code>joined_on</code> (tanggal bergabung), <code>weekday</code> (hari), dan <code>tenure_days</code> (lama bekerja dalam hari). Kolom-kolom ini dipakai di uji statistik 2.7.1, 2.7.2, dan 3.5.5. <code>tenure_days</code> dikosongkan untuk shift yang bertanggal sebelum <code>join_date</code> pengemudinya, karena lama bekerja yang negatif tidak mungkin. Tabel berikut merangkum variabel mana yang dipakai di pengujian dan alasannya.
</p>

| Variabel / data | Keputusan | Alasan | Bukti |
|---|---|---|---|
| Pool, zona kerja, jam mulai shift, status kerja, tipe kendaraan, hari, <code>driver_rating</code> | Dipakai | Lengkap untuk setiap shift; rating berada di rentang 1–5 | 1.7.5 |
| <code>tenure_days</code> | Dipakai, hanya sebagian shift | 44.353 shift bertanggal sebelum <code>join_date</code> pengemudinya, sehingga tenure-nya tidak valid dan dikosongkan | 1.7.5 |
| Pengemudi dengan <code>join_date</code> setelah data berakhir | Tidak dipakai untuk uji tenure | 630 pengemudi punya <code>join_date</code> setelah 31 Des 2023, sehingga tanggalnya tidak mungkin benar | 1.7.5 |
| <code>total_idle_minutes</code> sebagai penjelas | Tidak dipakai | Dihitung dari waktu trip (2.5.8), sehingga memakainya untuk menjelaskan jumlah trip akan melingkar; hanya diperlakukan sebagai hasil | 2.5.8 |
| <code>ritase_target_met</code> | Tidak dipakai | Turunan langsung dari <code>trip_count</code> (≥ 6 trip) dan hanya 894 dari 252.524 shift bernilai 1, sehingga hampir tidak ada variasi untuk diuji | 1.7.5, 2.5.2 |
| Request yang gagal | Tidak dipakai di uji tingkat pengemudi | 19.397 request gagal tidak punya pengemudi, shift, atau pool, sehingga tidak bisa dikaitkan ke idle time | 1.7.3, 2.4.1 |
| Trip Street Hail dalam ukuran permintaan | Tidak dipakai di "request per pengemudi" | 46.834 trip Street Hail tidak punya data request, sehingga permintaan yang bisa dihitung hanya App + Call Center | 2.2.2 |
| 30 Desember 2023 | Tidak dipakai di ukuran per jam dan per minggu | Data berhenti pukul 08:20 pada hari itu, sehingga harinya tidak lengkap | 1.7.7 |
| Jarak pool ke CBD dan Airport | Deskriptif saja | Hanya 8 pool (8 titik data) dan bercampur dengan komposisi zona, sehingga tidak cukup untuk uji statistik | 2.5.3 |
| <code>wait_time_minutes</code> | Tidak dipakai | Hanya ada untuk request yang terpenuhi dan median-nya sama di semua zona dan kelompok jam | 2.4.6 |
| ID, kolom perusahaan, metode bayar | Tidak dipakai | Tidak berhubungan dengan idle time pengemudi | – |

```python
# Driver attributes added to every shift with .map() (the number of rows cannot change).
driver_lookup = df_drivers.set_index("driver_id")
shifts_eda["driver_rating"] = shifts_eda["driver_id"].map(driver_lookup["driver_rating"])
shifts_eda["joined_on"] = shifts_eda["driver_id"].map(driver_lookup["join_date"])
shifts_eda["weekday"] = shifts_eda["shift_date"].dt.day_name()

# Tenure (days since joining) is only valid when the shift is NOT before the join date (flagged in Part 1).
shifts_eda["tenure_days"] = (shifts_eda["shift_date"] - shifts_eda["joined_on"]).dt.days
shifts_eda.loc[df_shifts_clean["flag_shift_before_join_date"].to_numpy(), "tenure_days"] = np.nan
assert len(shifts_eda) == len(df_shifts_clean)
```

### <span style="color:#0061AF">2.5.10 Seberapa Sibuk Armada? Utilisasi per Jam dan Zona</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Untuk setiap jam, waktu pengemudi bertugas dibandingkan dengan waktu pengemudi di dalam trip (booking sampai drop-off, definisi yang sama dengan 2.5.8). Utilisasi = menit sibuk / menit bertugas; porsi idle = 100% − utilisasi.
</p>

```python
def minutes_per_hour_slot(start_times, end_times):
    """For intervals [start, end): how many minutes of each hour slot (since 2023-10-01 00:00) are covered.
    Works with whole arrays at once (no loops over rows)."""
    a = ((start_times - FIRST_DAY) / pd.Timedelta(minutes=1)).to_numpy()
    b = ((end_times - FIRST_DAY) / pd.Timedelta(minutes=1)).to_numpy()
    a = np.clip(a, 0, N_SLOTS * 60)
    b = np.clip(b, 0, N_SLOTS * 60)
    first_slot = (a // 60).astype(int)
    last_slot = (b // 60).astype(int)
    covered = np.zeros(N_SLOTS + 2)
    same_slot = first_slot == last_slot
    np.add.at(covered, first_slot[same_slot], (b - a)[same_slot])
    cross = ~same_slot
    np.add.at(covered, first_slot[cross], (first_slot[cross] + 1) * 60 - a[cross])      # part of the first slot
    np.add.at(covered, last_slot[cross], b[cross] - last_slot[cross] * 60)             # part of the last slot
    whole = np.zeros(N_SLOTS + 3)                                                        # fully covered slots in between
    np.add.at(whole, first_slot[cross] + 1, 60)
    np.add.at(whole, last_slot[cross], -60)
    covered += np.cumsum(whole)[: N_SLOTS + 2]
    return covered[:N_SLOTS]

def per_hour_of_day(slot_values):
    """Average over the 90 full days for each hour of the day (0-23)."""
    return pd.Series(slot_values.reshape(N_DAYS, 24).sum(axis=0) / N_DAYS, index=range(24))

# Trips need the zone of the driver's shift, so we can compare drivers and trips of the same zone.
trips_eda["driver_zone"] = trips_eda["shift_id"].map(shifts_eda.set_index("shift_id")["dominant_zone_worked"])
assert trips_eda["driver_zone"].notna().all()

duty_minutes = minutes_per_hour_slot(shifts_eda["clock_in_time"], shifts_eda["clock_out_time"])
busy_minutes = minutes_per_hour_slot(trips_eda["booking_time"], trips_eda["dropoff_time"])

utilisation_by_hour = pd.DataFrame({
    "drivers_on_shift": per_hour_of_day(duty_minutes) / 60,
    "drivers_busy": per_hour_of_day(busy_minutes) / 60,
})
utilisation_by_hour["utilisation_pct"] = utilisation_by_hour["drivers_busy"] / utilisation_by_hour["drivers_on_shift"] * 100
utilisation_by_hour["failed_pct"] = supply_demand["failed_pct"]
display(utilisation_by_hour.round(1).T)

overall_utilisation = busy_minutes.sum() / duty_minutes.sum() * 100
print(f"Overall utilisation (busy minutes / on-shift minutes): {overall_utilisation:.1f}%  -> idle share {100 - overall_utilisation:.1f}%")
print(f"Highest hour: {utilisation_by_hour['utilisation_pct'].idxmax():02d}h = {utilisation_by_hour['utilisation_pct'].max():.1f}%")

# Same measure for each zone (drivers grouped by the dominant zone of their shift, trips by that zone as well).
zone_util_rows = {}
for zone in zones_in_order:
    zone_shifts = shifts_eda[shifts_eda["dominant_zone_worked"] == zone]
    zone_trips = trips_eda[trips_eda["driver_zone"] == zone]
    zone_duty = per_hour_of_day(minutes_per_hour_slot(zone_shifts["clock_in_time"], zone_shifts["clock_out_time"]))
    zone_busy = per_hour_of_day(minutes_per_hour_slot(zone_trips["booking_time"], zone_trips["dropoff_time"]))
    zone_util_rows[zone] = zone_busy / zone_duty * 100
zone_hour_utilisation = pd.DataFrame(zone_util_rows).T
zone_total_utilisation = {
    zone: minutes_per_hour_slot(trips_eda.loc[trips_eda["driver_zone"] == zone, "booking_time"],
                                trips_eda.loc[trips_eda["driver_zone"] == zone, "dropoff_time"]).sum()
          / minutes_per_hour_slot(shifts_eda.loc[shifts_eda["dominant_zone_worked"] == zone, "clock_in_time"],
                                  shifts_eda.loc[shifts_eda["dominant_zone_worked"] == zone, "clock_out_time"]).sum() * 100
    for zone in zones_in_order
}
print("Utilisation by dominant zone (whole day):")
display(pd.Series(zone_total_utilisation).round(1).to_frame("utilisation_pct"))

# Scale check: how much shift time would the same trips need at a higher utilisation? (illustration only)
what_if = pd.DataFrame({"assumed_utilisation_pct": [20, 30, 50, 70]})
what_if["shift_time_needed_pct_of_today"] = overall_utilisation / what_if["assumed_utilisation_pct"] * 100
display(what_if.round(1))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> pengemudi hanya memakai <strong>9,4%</strong> waktu bertugas di dalam trip (idle 90,6%). Utilisasi di bawah 20% pada 21 dari 24 jam, dan baru mencapai 33,9% pukul 16 dan 46,3% pukul 17. Per zona: CBD 13,2%, Mall 9,7%, Other 8,4%, Residential 7,7%, Airport 5,9%. Angka 9,4% tidak dibandingkan dengan 70% pada brief karena definisinya tidak diketahui. Jumlah pengemudi di bagian ini dihitung menurut menit tugas (rata-rata per jam), bukan jumlah shift yang menyentuh jam itu seperti di 2.5.1, sehingga angkanya lebih kecil (mis. 183 vs 351 pada 17). Dua masalah berbeda muncul: pada <strong>07–08</strong> pengemudi banyak (1.480–2.030) dan hanya sekitar 5% sibuk, tetapi 22% request gagal, sehingga penyebabnya bukan kekurangan pengemudi. Di CBD sendiri juga tidak kurang: 564–731 shift CBD aktif dan hanya 0,14 request per pengemudi per jam (2.5.1), tetapi 36,6% request CBD gagal. Jadi masalah pagi kemungkinan ada pada proses <strong>dispatch atau ketersediaan pengemudi di sistem</strong> (mis. pengemudi baru clock-in, masih dalam perjalanan dari pool, atau belum aktif di aplikasi). Ini hipotesis karena log dispatch tidak tersedia. Pada <strong>17:00</strong> pasokan jatuh (183 pengemudi) dan 46% sibuk, jadi kekurangan di sana nyata.
</p>

### <span style="color:#0061AF">2.5.11 Berapa Pengemudi yang Sebenarnya Dibutuhkan? (Skenario)</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Utilisasi 9,4% menimbulkan pertanyaan bisnis utama: berapa jam pengemudi yang dibutuhkan untuk melayani permintaan yang tercatat? Untuk setiap jam, jumlah pengemudi sibuk (2.5.10) dibagi tingkat utilisasi skenario 30%, 50%, dan 70% (70% hanya acuan sejarah dari brief, definisinya mungkin berbeda). Hasilnya adalah batas bawah: pengemudi bekerja dalam shift utuh, sehingga kebutuhan per jam tidak bisa dipenuhi persis. Ini skenario, bukan target.
</p>


```python
# Scenario: how many drivers would the recorded demand need in each hour?
# drivers_busy = average number of drivers inside a trip (booking -> drop-off) in that hour (from 2.5.10).
# needed = drivers_busy / target utilisation. The targets are SCENARIOS chosen by us, not Blue Bird targets.
target_levels = [0.30, 0.50, 0.70]
capacity = utilisation_by_hour[["drivers_on_shift", "drivers_busy"]].copy()
for level in target_levels:
    capacity[f"needed_at_{int(level * 100)}pct"] = capacity["drivers_busy"] / level
display(capacity.round(0).astype(int).T)

avg_shift_hours = shifts_eda["shift_duration_minutes"].mean() / 60
actual_driver_hours = capacity["drivers_on_shift"].sum()          # driver-hours on duty in an average day
capacity_rows = []
for level in target_levels:
    needed = capacity[f"needed_at_{int(level * 100)}pct"]
    capacity_rows.append({
        "scenario utilisation": f"{level:.0%}",
        "driver-hours needed per day": needed.sum(),
        "% of actual driver-hours": needed.sum() / actual_driver_hours * 100,
        "shift-equivalents per day": needed.sum() / avg_shift_hours,
        "hours where on-shift < needed": int((capacity["drivers_on_shift"] < needed).sum()),
    })
capacity_summary = pd.DataFrame(capacity_rows).set_index("scenario utilisation")
display(capacity_summary.round(1))
print(f"Actual: {actual_driver_hours:,.0f} driver-hours on duty per day "
      f"(= {actual_driver_hours / avg_shift_hours:,.0f} shift-equivalents of {avg_shift_hours:.1f} h; "
      f"{len(shifts_eda) / N_DAYS:,.0f} shifts per day in the shift table)")

# Recovering the excess failed requests (2.4.7) would add very little work:
extra_trips_per_day = impact["excess_failed_requests"].sum() / N_DAYS
busy_minutes_per_trip = ((trips_eda["dropoff_time"] - trips_eda["booking_time"]).dt.total_seconds() / 60).mean()
print(f"Recovering all excess failures adds about {extra_trips_per_day:.0f} trips per day "
      f"= about {extra_trips_per_day * busy_minutes_per_trip / 60:.0f} driver-hours per day "
      f"({extra_trips_per_day * busy_minutes_per_trip / 60 / capacity['drivers_busy'].sum() * 100:.1f}% of today's busy driver-hours)")
assert (capacity["drivers_busy"] <= capacity["drivers_on_shift"]).all()
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> dalam satu hari rata-rata, armada bertugas sekitar 23.822 jam-pengemudi (setara sekitar 2.800 shift 8,5 jam). Permintaan yang tercatat hanya membutuhkan sekitar 4.502 jam-pengemudi jika pengemudi sibuk 50% (18,9% dari yang ada, setara sekitar 530 shift) atau sekitar 7.503 jam-pengemudi jika sibuk 30% (31,5%). Kekurangan hanya muncul di skenario 30% pada pukul 16 dan 17, tepat saat shift pagi berakhir. Memulihkan seluruh kegagalan berlebih (2.4.7) hanya menambah sekitar 77 trip atau 50 jam-pengemudi per hari (2,2% jam sibuk). <strong>Makna bisnis:</strong> masalah utama armada adalah <em>kelebihan kapasitas</em>, bukan kekurangan, kecuali pada pergantian shift sore. Memindahkan shift antar zona tidak mengubah total permintaan, sehingga keuntungan terbesar ada pada penyesuaian jumlah dan jam shift. <strong>Batasan:</strong> ini batas bawah (pengemudi bekerja dalam shift utuh dan butuh cadangan), tingkat utilisasi adalah skenario, dan jika tabel trip hanya sampel (2.2.3) kebutuhan sebenarnya lebih besar.
</p>


### <span style="color:#0061AF">2.5.12 Ritase vs Pendapatan per Shift, per Zona dan Pool</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Ritase menghitung jumlah trip, padahal tarif trip Airport jauh lebih tinggi. Unit analisis: satu shift. Pembilang: total tarif trip pada shift itu (tabel trip final). Penyebut: jumlah shift. Dihitung juga tarif per jam bertugas.
</p>


```python
# Trips are not all worth the same: an Airport trip is about 2.4x the fare of a CBD trip.
# Revenue view: fare of the trips in each shift (final trips table; 2.2% of fare was excluded in 1.8).
fare_per_shift = trips_eda.groupby("shift_id")["total_fare"].sum()
shift_revenue = shifts_eda[["shift_id", "dominant_zone_worked", "pool_id", "trips_in_table", "shift_duration_minutes"]].copy()
shift_revenue["fare_rp"] = shift_revenue["shift_id"].map(fare_per_shift).fillna(0)
assert abs(shift_revenue["fare_rp"].sum() - trips_eda["total_fare"].sum()) < 1      # every trip fare is counted once

def revenue_table(by):
    out = shift_revenue.groupby(by).agg(
        shifts=("shift_id", "size"),
        avg_trips_per_shift=("trips_in_table", "mean"),
        avg_fare_per_shift_rp=("fare_rp", "mean"),
        avg_hours=("shift_duration_minutes", lambda m: m.mean() / 60),
    )
    out["fare_per_on_duty_hour_rp"] = out["avg_fare_per_shift_rp"] / out["avg_hours"]
    out["trips_vs_best_pct"] = out["avg_trips_per_shift"] / out["avg_trips_per_shift"].max() * 100
    out["fare_vs_best_pct"] = out["avg_fare_per_shift_rp"] / out["avg_fare_per_shift_rp"].max() * 100
    return out

revenue_by_zone = revenue_table("dominant_zone_worked").loc[zones_in_order]
revenue_by_pool = revenue_table("pool_id")
display(revenue_by_zone.round({"avg_trips_per_shift": 2, "avg_fare_per_shift_rp": 0, "avg_hours": 2, "fare_per_on_duty_hour_rp": 0, "trips_vs_best_pct": 1, "fare_vs_best_pct": 1}))
display(revenue_by_pool.round({"avg_trips_per_shift": 2, "avg_fare_per_shift_rp": 0, "avg_hours": 2, "fare_per_on_duty_hour_rp": 0, "trips_vs_best_pct": 1, "fare_vs_best_pct": 1}))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> dalam <strong>pendapatan</strong>, jarak antar zona jauh lebih kecil daripada dalam ritase: shift Airport menghasilkan 39,8% trip shift CBD tetapi 73,9% pendapatannya (Rp 90.648 vs Rp 122.709 per shift), karena tarif Airport lebih tinggi. Pendapatan per jam bertugas terendah justru Residential (Rp 10.398) dan Airport (Rp 10.667); CBD tertinggi (Rp 14.435). Antar pool, PL-CGK hanya 69,3% ritase pool terbaik tetapi 86,1% pendapatannya. <strong>Makna:</strong> menilai zona atau pool hanya dengan ritase melebih-lebihkan masalah Airport. Keputusan pengurangan atau pemindahan shift sebaiknya memakai pendapatan per jam, sehingga Residential sama layaknya dengan Airport untuk ditinjau. <strong>Keterbatasan:</strong> pendapatan di sini adalah tarif argometer; biaya (BBM, komisi, perawatan) tidak ada di data.
</p>


### <span style="color:#0061AF">2.5.13 Apakah Pengemudi, Status Kerja, atau Panjang Shift Berpengaruh?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Tiga pertanyaan: (1) apakah ada pengemudi yang konsisten lebih produktif atau kurang produktif dari yang diharapkan zonanya; (2) apakah pengemudi Tetap dan Mitra berbeda dalam kehadiran atau produktivitas; (3) apakah shift yang lebih panjang menghasilkan lebih banyak trip.
</p>


```python
from scipy import stats

# 1) Do some drivers do consistently better or worse than their zone predicts?
#    Each driver: observed trips vs expected trips (zone average for each of his shifts).
#    If differences between drivers are pure chance, the z-scores have a spread (standard deviation) of about 1.
zone_expected = shifts_eda.groupby("dominant_zone_worked")["trip_count"].transform("mean")
driver_check = pd.DataFrame({"driver_id": shifts_eda["driver_id"], "observed": shifts_eda["trip_count"], "expected": zone_expected}).groupby("driver_id").sum()
driver_check["z"] = (driver_check["observed"] - driver_check["expected"]) / np.sqrt(driver_check["expected"])
drivers_beyond_3 = int((driver_check["z"].abs() > 3).sum())
print(f"Spread of driver z-scores: {driver_check['z'].std():.3f} (about 1 = no real difference between drivers)")
print(f"Drivers with |z| > 3: {drivers_beyond_3} (expected by chance alone: {2 * stats.norm.sf(3) * len(driver_check):.0f})")

# 2) Attendance vs productivity by employment status
attendance = shifts_eda.groupby("employment_status").agg(
    drivers=("driver_id", "nunique"), shifts=("shift_id", "size"),
    avg_trips_per_shift=("trip_count", "mean"), avg_shift_hours=("shift_duration_minutes", lambda m: m.mean() / 60),
)
attendance["shifts_per_driver"] = attendance["shifts"] / attendance["drivers"]
display(attendance.round(2))

# 3) Do longer shifts produce more trips?
shift_hours = shifts_eda["shift_duration_minutes"] / 60
length_group = pd.cut(shift_hours, [0, 7, 8, 9, 10, 24], labels=["up to 7 h", "7-8 h", "8-9 h", "9-10 h", "over 10 h"])
by_length = shifts_eda.groupby(length_group, observed=True)["trip_count"].agg(["size", "mean"]).rename(columns={"size": "shifts", "mean": "avg_trips_per_shift"})
display(by_length.round(3))
length_rho, length_p = stats.spearmanr(shift_hours, shifts_eda["trip_count"])
print(f"Spearman rho between shift length and trips: {length_rho:.3f} (p = {length_p:.3g})")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> (1) <strong>Tidak ada pengemudi yang konsisten lebih baik atau lebih buruk</strong>: sebaran z = 0,993 dan hanya 10 pengemudi dengan |z| &gt; 3, sama dengan yang diharapkan dari kebetulan (11). Setelah zona diperhitungkan, perbedaan antar pengemudi adalah kebetulan, sehingga pelatihan atau insentif individual tidak akan menaikkan ritase. (2) Pengemudi Mitra dan Tetap sama produktifnya (1,26 trip per shift), tetapi Mitra bekerja lebih jarang (47,8 vs 66,5 shift per pengemudi dalam tiga bulan). (3) <strong>Panjang shift tidak berpengaruh</strong>: shift hingga 7 jam dan 9–10 jam sama-sama 1,26 trip (rho = −0,001). Jam tambahan tidak menghasilkan trip tambahan, artinya jumlah trip dibatasi permintaan, bukan jam kerja. Ini bukti kedua, terpisah dari 2.5.11, bahwa armada kelebihan kapasitas.
</p>


### <span style="color:#0061AF">2.5.14 Apakah Zona Kerja Ditentukan oleh Pool?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Jika zona dominan hanya hasil dari tempat pengemudi kebetulan mendapat trip, porsinya akan bervariasi bebas antar pool. Jika porsinya sama persis untuk pool dengan afinitas yang sama, zona kerja adalah hasil penugasan yang bisa diubah manajemen.
</p>


```python
# Is the dominant zone of a shift decided by the pool? Share of each pool's shifts in each zone (rows add to 100%).
pool_zone_share = pd.crosstab(shifts_eda["pool_id"], shifts_eda["dominant_zone_worked"], normalize="index")[zones_in_order] * 100
pool_zone_share = pool_zone_share.join(df_pools.set_index("pool_id")["primary_zone_affinity"])
display(pool_zone_share.round(1))
# If the zone were only the result of where drivers happened to find trips, pools with the same affinity would differ.
spread_inside_affinity = pool_zone_share.groupby("primary_zone_affinity")[zones_in_order].agg(lambda s: s.max() - s.min())
print("Largest difference between pools with the SAME affinity (percentage points):")
display(spread_inside_affinity.round(2))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> porsi zona hampir persis sama untuk pool dengan afinitas yang sama (selisih maksimal 0,75 poin): Mixed sekitar 20% Airport, 25% CBD, 20% Mall, 20% Residential, 15% Other; CBD-heavy 10/45/15/15/15; Airport-heavy (PL-CGK) 55/15/10/10/10. Keseragaman ini tidak mungkin terjadi jika zona hanya hasil kebetulan, sehingga zona kerja adalah <strong>kuota penugasan per pool</strong>. Ini memperkuat 2.5.6: label zona terutama mencerminkan penugasan. Akibatnya perbedaan ritase antar pool (2.5.5) adalah akibat kebijakan alokasi zona, dan kebijakan itu adalah tuas yang bisa diubah langsung oleh manajemen.
</p>


### <span style="color:#0061AF">2.5.15 Permintaan dan Pasokan per Hari dalam Seminggu</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Apakah jumlah shift mengikuti naik-turun permintaan antar hari? Hanya hari penuh yang dipakai, karena 2023-12-30 (Sabtu) hanya terekam sebagian.
</p>


```python
# Demand and supply per weekday (full days only: 2023-10-01 to 2023-12-29, so every weekday is complete).
full_day_requests = req[req["requested_time"] < pd.Timestamp("2023-12-30")]
full_day_shifts = shifts_eda[shifts_eda["shift_date"] < pd.Timestamp("2023-12-30")]
weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
request_days = full_day_requests.groupby("weekday")["requested_time"].apply(lambda x: x.dt.normalize().nunique())
shift_weekday = full_day_shifts["shift_date"].dt.day_name()
weekday_balance = pd.DataFrame({
    "requests_per_day": full_day_requests.groupby("weekday").size() / request_days,
    "shifts_per_day": full_day_shifts.groupby(shift_weekday).size() / full_day_shifts.groupby(shift_weekday)["shift_date"].nunique(),
    "failed_pct": full_day_requests.groupby("weekday")["is_failed"].mean() * 100,
}).loc[weekday_order]
weekday_balance["requests_per_shift"] = weekday_balance["requests_per_day"] / weekday_balance["shifts_per_day"]
display(weekday_balance.round(2))
print(f"Requests per day: {weekday_balance['requests_per_day'].min():,.0f} to {weekday_balance['requests_per_day'].max():,.0f}; "
      f"shifts per day: {weekday_balance['shifts_per_day'].min():,.0f} to {weekday_balance['shifts_per_day'].max():,.0f}")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> permintaan per hari sangat stabil (3.170–3.268 request), demikian juga jumlah shift (2.796–2.816) dan tingkat kegagalan (6,6–7,0%). Tidak ada hari yang secara khusus kekurangan atau kelebihan pasokan, sehingga masalah pasokan adalah soal <strong>jam</strong>, bukan hari. Catatan: jika 30 Desember yang hanya terekam sebagian ikut dihitung, Sabtu tampak sekitar 7% lebih sepi; itu artefak data, bukan pola permintaan.
</p>

### <span style="color:#0061AF">2.6.1 Klien Korporat</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Trip korporat adalah trip dengan <code>company_id</code>. Trip tanpa <code>company_id</code> adalah pelanggan biasa (null struktural, bukan error).
</p>

```python
print(f"Corporate trips: {trips_eda['is_corporate'].sum():,} of {len(trips_eda):,} = {trips_eda['is_corporate'].mean()*100:.1f}%")
print("Corporate share of trips by pickup zone / booking channel / trip type:")
for column in ["pickup_zone_type", "booking_channel", "trip_type"]:
    display((trips_eda.groupby(column)["is_corporate"].mean() * 100).round(1).to_frame("corporate_trip_pct"))

corporate_trips = trips_eda[trips_eda["is_corporate"]].merge(df_clients, on="company_id", how="left", validate="many_to_one")
assert len(corporate_trips) == trips_eda["is_corporate"].sum()   # the merge did not multiply rows
assert corporate_trips["industry"].notna().all()

by_industry = corporate_trips.groupby("industry").agg(
    trips=("trip_id", "size"),
    companies_with_trips=("company_id", "nunique"),
    total_fare=("total_fare", "sum"),
    median_fare=("total_fare", "median"),
)
by_industry["trip_share_pct"] = by_industry["trips"] / by_industry["trips"].sum() * 100
by_industry["trips_per_company"] = by_industry["trips"] / by_industry["companies_with_trips"]
display(by_industry.sort_values("trips", ascending=False))

per_company = corporate_trips.groupby("company_id").size().sort_values(ascending=False)
top_10pct = int(np.ceil(len(per_company) * 0.10))
print(f"Companies with at least one trip: {len(per_company)} of {len(df_clients)} in the clients table")
top_trips = per_company.head(top_10pct).sum()
print(f"Top 10% of companies ({top_10pct}) = {top_trips:,} of {per_company.sum():,} corporate trips = {top_trips / per_company.sum() * 100:.1f}%")
print(f"Tie check at the cut-off: company #{top_10pct} has {per_company.iloc[top_10pct - 1]} trips, company #{top_10pct + 1} has {per_company.iloc[top_10pct]} trips")
print(f"If all companies were equal, {top_10pct} companies would give {top_10pct / len(per_company) * 100:.1f}% of trips")
print("Payment method vs corporate flag (should show only eVoucher Corporate = corporate):")
display(pd.crosstab(trips_eda["payment_method"], trips_eda["is_corporate"]))
```

### <span style="color:#0061AF">2.6.2 Trip eVoucher pada Waktu Tidak Lazim (Semua Zona)</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Brief menyebut eVoucher mungkin dipakai pada waktu tidak lazim, tetapi tidak mendefinisikannya. Asumsi kami: tidak lazim = mulai pukul 22:00–04:59 atau hari Sabtu/Minggu. Ini aturan penyaringan, bukan bukti penyalahgunaan.
</p>

```python
ev = trips_eda.copy()
ev["is_evoucher"] = ev["payment_method"] == "eVoucher Corporate"
ev["start_hour"] = ev["meter_start_time"].dt.hour
ev["is_weekend"] = ev["meter_start_time"].dt.dayofweek >= 5
ev["is_late_night"] = (ev["start_hour"] >= 22) | (ev["start_hour"] <= 4)
ev["is_unusual_time"] = ev["is_late_night"] | ev["is_weekend"]

assert ev["is_evoucher"].equals(ev["company_id"].notna())   # same group (checked in Part 1)

unusual = ev.groupby("is_evoucher").agg(
    trips=("trip_id", "size"),
    late_night_pct=("is_late_night", lambda s: s.mean() * 100),
    weekend_pct=("is_weekend", lambda s: s.mean() * 100),
    unusual_time_pct=("is_unusual_time", lambda s: s.mean() * 100),
)
unusual.index = ["Other payment methods", "eVoucher Corporate"]
display(unusual)

# Hour-of-day profile: share of each group's trips that start in each hour.
evoucher_hour_share = pd.DataFrame({
    "eVoucher Corporate": ev[ev["is_evoucher"]]["start_hour"].value_counts(normalize=True).sort_index() * 100,
    "Other payment methods": ev[~ev["is_evoucher"]]["start_hour"].value_counts(normalize=True).sort_index() * 100,
})

# Screening per company (only companies with enough eVoucher trips to be meaningful).
overall_unusual = ev.loc[ev["is_evoucher"], "is_unusual_time"].mean()
company_screen = ev[ev["is_evoucher"]].groupby("company_id").agg(
    evoucher_trips=("trip_id", "size"),
    unusual_time_trips=("is_unusual_time", "sum"),
)
company_screen["unusual_time_pct"] = company_screen["unusual_time_trips"] / company_screen["evoucher_trips"] * 100
# z-score: how many standard errors above the all-company rate? (a screening number, not a verdict)
company_screen["z_score"] = (
    (company_screen["unusual_time_pct"] / 100 - overall_unusual)
    / np.sqrt(overall_unusual * (1 - overall_unusual) / company_screen["evoucher_trips"])
)
MIN_TRIPS_FOR_SCREEN = 50
screened = company_screen[company_screen["evoucher_trips"] >= MIN_TRIPS_FOR_SCREEN]
company_screen = company_screen.join(df_clients.set_index("company_id")[["company_name", "industry"]])
print(f"All-company eVoucher unusual-time rate: {overall_unusual*100:.1f}%")
print(f"Companies with >= {MIN_TRIPS_FOR_SCREEN} eVoucher trips: {len(screened)} of {len(company_screen)}")
print("Companies to review first (z > 3), largest z first:")
display(company_screen.loc[screened.index].query("z_score > 3").sort_values("z_score", ascending=False).round(2))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> pada semua zona digabung, trip eVoucher justru <em>lebih jarang</em> pada malam atau akhir pekan dibanding trip lain (26,2% vs 36,4%), dan tidak ada perusahaan yang menonjol (tidak ada dari 244 perusahaan dengan 50+ trip yang memiliki z &gt; 3). Namun angka gabungan ini menyembunyikan pola sebenarnya, yang terlihat setelah dipisah per zona di 2.6.5.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Kolom baru <code>customer_segment</code> dibuat dari <code>company_id</code> lalu kedua kelompok dibandingkan.
</p>

```python
trips_eda["customer_segment"] = np.where(trips_eda["company_id"].notna(), "B2B", "Non-B2B")

# Sanity checks: the label must match the payment method exactly.
assert trips_eda["customer_segment"].isin(["B2B", "Non-B2B"]).all()
assert (trips_eda["customer_segment"] == "B2B").sum() == (trips_eda["payment_method"] == "eVoucher Corporate").sum()
print("Payment method by customer segment (trips):")
display(pd.crosstab(trips_eda["payment_method"], trips_eda["customer_segment"]))

segment_summary = trips_eda.groupby("customer_segment").agg(
    trips=("trip_id", "size"),
    total_fare=("total_fare", "sum"),
    mean_fare=("total_fare", "mean"),
    median_fare=("total_fare", "median"),
    median_distance_km=("distance_km_clean", "median"),
)
segment_summary["trip_share_pct"] = segment_summary["trips"] / segment_summary["trips"].sum() * 100
segment_summary["fare_share_pct"] = segment_summary["total_fare"] / segment_summary["total_fare"].sum() * 100
display(segment_summary)

print("Pickup zone mix inside each segment (% of the segment's trips):")
segment_zone_mix = pd.crosstab(trips_eda["customer_segment"], trips_eda["pickup_zone_type"], normalize="index") * 100
display(segment_zone_mix.round(1))
print("Booking channel mix inside each segment (% of the segment's trips):")
display((pd.crosstab(trips_eda["customer_segment"], trips_eda["booking_channel"], normalize="index") * 100).round(1))
print("Service type mix inside each segment (% of the segment's trips):")
display((pd.crosstab(trips_eda["customer_segment"], trips_eda["service_type_clean"], normalize="index") * 100).round(1))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">

Trip B2B bertarif lebih tinggi (rata-rata Rp 89.939 vs Rp 81.863) dan lebih sering ke Airport (22,0% vs 11,4%), lebih banyak dipesan lewat Call Center (34,8% vs 22,8%). ### <span style="color:#0061AF">2.6.5 eVoucher pada Waktu Tidak Lazim per Zona Penjemputan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
2.6.2 membandingkan semua zona digabung, padahal trip B2B dua kali lebih sering ke Airport. Perbandingan dilakukan di dalam setiap zona dengan aturan yang sama (22:00–04:59 atau akhir pekan).
</p>

```python
ev["customer_segment"] = np.where(ev["is_evoucher"], "B2B", "Non-B2B")       # same rule as 2.6.3, added to the 2.6.2 table

zone_rule = ev.groupby(["pickup_zone_type", "customer_segment"]).agg(
    trips=("trip_id", "size"),
    late_night_pct=("is_late_night", lambda s: s.mean() * 100),
    weekend_pct=("is_weekend", lambda s: s.mean() * 100),
    unusual_time_pct=("is_unusual_time", lambda s: s.mean() * 100),
)
print("Unusual-time share inside each zone, B2B vs Non-B2B (% of that group's trips in the zone):")
display(zone_rule.round(1))

airport_b2b = ev[(ev["pickup_zone_type"] == "Airport") & (ev["customer_segment"] == "B2B")]
airport_non_b2b = ev[(ev["pickup_zone_type"] == "Airport") & (ev["customer_segment"] == "Non-B2B")]
all_b2b = ev[ev["customer_segment"] == "B2B"]

airport_unusual = airport_b2b[airport_b2b["is_unusual_time"]]
expected_unusual = airport_non_b2b["is_unusual_time"].mean() * len(airport_b2b)
print(f"B2B Airport trips: {len(airport_b2b):,} = {len(airport_b2b) / len(all_b2b) * 100:.1f}% of all B2B trips")
print(f"  weekend: {airport_b2b['is_weekend'].sum():,} | late night: {airport_b2b['is_late_night'].sum():,} | either: {len(airport_unusual):,}")
print(f"  unusual-time share: {airport_b2b['is_unusual_time'].mean() * 100:.1f}% (Non-B2B Airport: {airport_non_b2b['is_unusual_time'].mean() * 100:.1f}%)")
print(f"  unusual-time trips = {len(airport_unusual) / len(all_b2b) * 100:.1f}% of all B2B trips "
      f"and {airport_unusual['total_fare'].sum() / all_b2b['total_fare'].sum() * 100:.1f}% of all B2B fare "
      f"(Rp {airport_unusual['total_fare'].sum():,.0f})")
print(f"  trips above what a Non-B2B Airport pattern would give: about {len(airport_unusual) - expected_unusual:,.0f}")

# Weekday profile of Airport trips (share of the group's Airport trips on each weekday).
weekday_names = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
airport_weekday = pd.DataFrame({
    "B2B": airport_b2b["meter_start_time"].dt.dayofweek.value_counts(normalize=True).sort_index() * 100,
    "Non-B2B": airport_non_b2b["meter_start_time"].dt.dayofweek.value_counts(normalize=True).sort_index() * 100,
})
airport_weekday.index = weekday_names
display(airport_weekday.round(1))

# Is it a few companies, or all of them? (Airport B2B trips only; companies with >= 20 such trips)
airport_company = airport_b2b.groupby("company_id").agg(
    airport_trips=("trip_id", "size"),
    unusual_pct=("is_unusual_time", lambda s: s.mean() * 100),
)
airport_company = airport_company[airport_company["airport_trips"] >= 20]
print(f"Companies with >= 20 B2B Airport trips: {len(airport_company)}")
display(airport_company["unusual_pct"].describe().round(1).to_frame("unusual_pct_per_company"))

airport_by_industry = airport_b2b.merge(df_clients, on="company_id", how="left").groupby("industry")["is_unusual_time"].mean() * 100
display(airport_by_industry.round(1).to_frame("unusual_pct_by_industry"))
# Which part of "unusual time" drives the Airport pattern: weekend or night?
split_rows = []
for label, group in [("B2B Airport", airport_b2b), ("Non-B2B Airport", airport_non_b2b), ("All trips", ev)]:
    split_rows.append({"group": label, "weekend_pct": group["is_weekend"].mean() * 100,
                       "late_night_pct": group["is_late_night"].mean() * 100,
                       "weekday_daytime_pct": (~group["is_unusual_time"]).mean() * 100})
display(pd.DataFrame(split_rows).set_index("group").round(1))
print("If trips were spread evenly over the week, weekend days would hold 2/7 = 28.6% of trips.")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> di semua zona non-Airport, trip B2B lebih jarang pada malam atau akhir pekan (sekitar 15% vs 38%). <strong>Di Airport sebaliknya:</strong> 65,9% trip B2B pada malam atau akhir pekan vs 22,6% trip Non-B2B (Sabtu 25,2% dan Minggu 26,2% dari trip B2B Airport). Ini 8.134 trip, 14,5% trip B2B dan 24,3% tarif B2B (Rp 1,23 miliar). Polanya <strong>bukan milik beberapa perusahaan</strong>: dari 200 perusahaan dengan 20+ trip B2B Airport, porsinya 48–85% (median 67%), dan setiap industri 65–67%. Komponen terbesarnya adalah <strong>akhir pekan</strong>: 51,4% trip B2B Airport terjadi Sabtu–Minggu, dibanding 17,6% trip Non-B2B Airport dan 27,8% seluruh trip (pembagian merata akan memberi 28,6%); porsi malam 20,6% vs 6,9%. Hanya 34,1% trip B2B Airport terjadi pada hari kerja siang. Pola ini jelas tetapi bukan bukti penyalahgunaan: perjalanan dinas ke bandara sering jatuh di akhir pekan atau malam (jadwal penerbangan). Karena polanya seragam di semua perusahaan, persoalannya adalah <strong>aturan dan kontrol di tingkat kebijakan</strong> (apakah akhir pekan diizinkan, apakah alasan trip dicatat), bukan beberapa karyawan tertentu. Untuk memutuskan, dibutuhkan kebijakan perjalanan klien dan tujuan trip yang tidak ada di data.
</p>

### <span style="color:#0061AF">2.6.6 Skenario: Jika Kebijakan eVoucher Hanya Berlaku di CBD</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Pertanyaan:</strong> bagaimana jika kebijakan eVoucher hanya mengizinkan trip di CBD? <strong>Penting:</strong> brief <em>tidak</em> menyebut aturan zona; brief hanya menyebut eVoucher untuk perjalanan dinas. Bagian ini adalah <strong>uji skenario</strong>, bukan pemeriksaan pelanggaran, dan hasilnya tidak boleh dilaporkan sebagai jumlah pelanggaran sebelum aturan dikonfirmasi ke Finance. <strong>Keterbatasan data:</strong> tabel <code>bb_trips</code> hanya punya zona penjemputan (<code>pickup_zone_type</code>) dan tidak punya zona tujuan. Karena itu hanya kasus <em>pickup di luar CBD</em> yang bisa diukur. Kasus hanya-tujuan-di-luar-CBD dan keduanya-di-luar-CBD tidak bisa dibedakan, sehingga jumlah pelanggaran sebenarnya (jika kebijakan berlaku untuk kedua titik) minimal sebesar angka pickup ini. Unit analisis: satu trip eVoucher (trip dengan <code>company_id</code>); penyebut: seluruh trip eVoucher; pembanding: trip dengan metode bayar lain (Non-B2B).
</p>

```python
# Screening: eVoucher trips whose PICKUP zone is not the CBD.
# SCENARIO ONLY: the brief does NOT state a CBD rule. This tests "what if the policy were CBD-only?".
# LIMITATION: bb_trips has only pickup_zone_type. There is no destination zone, so only the pickup endpoint can be checked.
POLICY_ZONE = "CBD"
MIN_COMPANY_TRIPS = 20          # same cut-off as the company screening above

screen = ev.copy()
screen["flag_pickup_outside_cbd"] = screen["pickup_zone_type"].notna() & (screen["pickup_zone_type"] != POLICY_ZONE)
ev_trips = screen[screen["is_evoucher"]]
other_trips = screen[~screen["is_evoucher"]]      # reference group: all other payment methods (Non-B2B)

print(f"eVoucher trips: {len(ev_trips):,} | reference (Non-B2B) trips: {len(other_trips):,}")
print(f"Trips with a missing pickup zone: {int(screen['pickup_zone_type'].isna().sum())}")

flagged_trips = ev_trips[ev_trips["flag_pickup_outside_cbd"]]
total_ev_fare = ev_trips["total_fare"].sum()

screening_summary = pd.DataFrame({
    "trips": [len(ev_trips), len(flagged_trips), len(ev_trips) - len(flagged_trips)],
    "pct_of_evoucher_trips": [100, len(flagged_trips) / len(ev_trips) * 100, (1 - len(flagged_trips) / len(ev_trips)) * 100],
    "fare_rp": [total_ev_fare, flagged_trips["total_fare"].sum(), total_ev_fare - flagged_trips["total_fare"].sum()],
}, index=["All eVoucher trips", "Pickup outside CBD (scenario flag)", "Pickup in CBD"])
screening_summary["pct_of_evoucher_fare"] = screening_summary["fare_rp"] / total_ev_fare * 100
display(screening_summary.round(1))
print("Pickup-only / destination-only / both outside CBD: only 'pickup outside CBD' can be measured.")
print("Destination-only and both-outside trips cannot be identified (no destination column).")

# Which zones make up the flagged trips, compared with the reference group
by_zone = ev_trips.groupby("pickup_zone_type").agg(
    trips=("trip_id", "size"),
    fare_rp=("total_fare", "sum"),
    median_fare=("total_fare", "median"),
).loc[zones_in_order]
by_zone["pct_of_evoucher_trips"] = by_zone["trips"] / by_zone["trips"].sum() * 100
by_zone["pct_of_evoucher_fare"] = by_zone["fare_rp"] / by_zone["fare_rp"].sum() * 100
by_zone["reference_pct_of_trips"] = other_trips["pickup_zone_type"].value_counts(normalize=True).loc[zones_in_order] * 100
display(by_zone.round(1))
```

```python
# Is eVoucher different from the reference group? (two-proportion z-test, confidence interval, chi-square on the zone mix)
from scipy import stats

n_ev, n_ref = len(ev_trips), len(other_trips)
p_ev = ev_trips["flag_pickup_outside_cbd"].mean()
p_ref = other_trips["flag_pickup_outside_cbd"].mean()

pooled = (p_ev * n_ev + p_ref * n_ref) / (n_ev + n_ref)
z_value = (p_ev - p_ref) / np.sqrt(pooled * (1 - pooled) * (1 / n_ev + 1 / n_ref))
p_value = 2 * stats.norm.sf(abs(z_value))
standard_error = np.sqrt(p_ev * (1 - p_ev) / n_ev + p_ref * (1 - p_ref) / n_ref)
diff = p_ev - p_ref
cohens_h = 2 * np.arcsin(np.sqrt(p_ev)) - 2 * np.arcsin(np.sqrt(p_ref))
print(f"Pickup outside CBD: eVoucher {p_ev * 100:.1f}% vs reference {p_ref * 100:.1f}%")
print(f"Difference = {diff * 100:+.1f} percentage points (95% CI {(diff - 1.96 * standard_error) * 100:+.1f} to {(diff + 1.96 * standard_error) * 100:+.1f}), "
      f"z = {z_value:.1f}, p = {p_value:.3g}, Cohen's h = {cohens_h:.2f}")

zone_table = pd.crosstab(screen["is_evoucher"], screen["pickup_zone_type"])
chi2_value, chi2_p, chi2_dof, _ = stats.chi2_contingency(zone_table)
cramers_v = np.sqrt(chi2_value / (zone_table.to_numpy().sum() * (min(zone_table.shape) - 1)))
print(f"Chi-square on the zone mix (eVoucher vs reference): chi2 = {chi2_value:,.0f}, dof = {chi2_dof}, p = {chi2_p:.3g}, Cramer's V = {cramers_v:.3f}")

# Is it a few companies? Share of each company's eVoucher trips picked up outside the CBD (companies with enough trips)
company_flag = ev_trips.groupby("company_id").agg(
    trips=("trip_id", "size"),
    flagged=("flag_pickup_outside_cbd", "sum"),
)
company_flag = company_flag[company_flag["trips"] >= MIN_COMPANY_TRIPS].copy()
company_flag["flagged_pct"] = company_flag["flagged"] / company_flag["trips"] * 100
company_flag["z_score"] = (company_flag["flagged_pct"] / 100 - p_ev) / np.sqrt(p_ev * (1 - p_ev) / company_flag["trips"])
z_limit = stats.norm.isf(0.05 / 2 / len(company_flag))      # Bonferroni: 0.05 shared by all companies
print(f"Companies with >= {MIN_COMPANY_TRIPS} eVoucher trips: {len(company_flag)}")
print(f"Share of trips outside the CBD per company: median {company_flag['flagged_pct'].median():.1f}%, "
      f"range {company_flag['flagged_pct'].min():.1f}% to {company_flag['flagged_pct'].max():.1f}%")
print(f"Companies above the Bonferroni limit (z > {z_limit:.2f}): {int((company_flag['z_score'] > z_limit).sum())}")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Temuan data (terkonfirmasi):</strong> dari 56.082 trip eVoucher, 37.251 trip (66,4%) dijemput di luar CBD dengan total tarif Rp 3,74 miliar, yaitu 74,2% dari tarif eVoucher (Rp 5,04 miliar). Zona penjemputan tidak ada yang kosong. Komposisi trip eVoucher: Airport 22,0% (36,3% tarif), CBD 33,6%, Mall 16,3%, Residential 15,2%, Other 13,0%. Pada kelompok pembanding, 61,7% trip dijemput di luar CBD; selisihnya +4,7 poin persentase (CI 95% +4,3 sampai +5,1; uji z p &lt; 0,001; Cohen's h = 0,10, efek diabaikan). Chi-square pada komposisi zona signifikan tetapi kecil (Cramér's V = 0,121), terutama karena porsi Airport (22,0% vs 11,4%). Per perusahaan (275 perusahaan dengan minimal 20 trip), porsi pickup di luar CBD berkisar 45,0–80,0% (median 66,6%) dan tidak ada perusahaan yang melewati batas Bonferroni (z &gt; 3,74). <strong>Interpretasi:</strong> angka 37.251 trip dan Rp 3,74 miliar <strong>bukan</strong> nilai pelanggaran atau kerugian; ini hanya nilai tarif trip yang dijemput di luar CBD. Jika aturan "hanya CBD" dipakai apa adanya, dua pertiga trip eVoucher akan ditandai, hampir sama dengan pelanggan non-eVoucher (61,7%). Perilaku yang begitu luas dan seragam di semua perusahaan lebih menunjukkan bahwa aturan CBD-only kemungkinan bukan kebijakan yang berlaku atau tidak ditegakkan, bukan bahwa sebagian besar perusahaan melanggar. Satu-satunya perbedaan nyata adalah porsi Airport, yang wajar untuk perjalanan dinas (penerbangan).
</p>

### <span style="color:#0061AF">2.6.7 Distribusi Tarif dan Kasus Tarif Tinggi eVoucher</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Tarif dibandingkan dengan kelompok pembanding (trip Non-B2B pada zona penjemputan dan tipe trip yang sama). Tarif "tinggi" berarti di atas batas atas IQR (Q3 + 1,5 × IQR) kelompok pembanding. Tarif tinggi hanya sinyal untuk ditinjau, bukan bukti kecurangan, karena trip Charter atau Airport yang jauh memang mahal. Tiga sinyal dihitung per trip eVoucher: (1) pickup di luar CBD, (2) mulai akhir pekan atau pukul 22:00–04:59 (asumsi kami), dan (3) tarif di atas batas pembanding.
</p>

```python
# Fare screening. "High" = above the upper fence (Q3 + 1.5 x IQR) of the REFERENCE group (Non-B2B) for the same pickup zone and trip type.
# A high fare is only a review signal: a long Charter or airport trip can legitimately cost a lot.
group_columns = ["pickup_zone_type", "trip_type"]

screen["fare_per_km"] = screen["total_fare"] / screen["distance_km_clean"].where(screen["distance_km_clean"] > 0)
reference_group = screen[~screen["is_evoucher"]]

def upper_fence(column):
    """Q3 + 1.5 x IQR of the reference group, for each pickup zone and trip type."""
    q1 = reference_group.groupby(group_columns)[column].quantile(0.25)
    q3 = reference_group.groupby(group_columns)[column].quantile(0.75)
    return (q3 + 1.5 * (q3 - q1)).rename(column + "_fence").reset_index()

n_before = len(screen)
screen = screen.merge(upper_fence("total_fare"), on=group_columns, how="left")
screen = screen.merge(upper_fence("fare_per_km"), on=group_columns, how="left")
assert len(screen) == n_before          # the merge must not multiply rows

screen["flag_high_fare"] = screen["total_fare"] > screen["total_fare_fence"]
screen["flag_high_fare_per_km"] = screen["fare_per_km"] > screen["fare_per_km_fence"]      # NaN fare_per_km counts as False

ev_trips = screen[screen["is_evoucher"]]
other_trips = screen[~screen["is_evoucher"]]

fare_rows = []
for label, data in [("eVoucher", ev_trips), ("Reference (Non-B2B)", other_trips)]:
    fare_rows.append({
        "group": label,
        "trips": len(data),
        "high fare (% of trips)": data["flag_high_fare"].mean() * 100,
        "high fare per km (% of trips)": data["flag_high_fare_per_km"].mean() * 100,
        "median fare (Rp)": data["total_fare"].median(),
        "maximum fare (Rp)": data["total_fare"].max(),
    })
display(pd.DataFrame(fare_rows).set_index("group").round(1))

x_ev, x_ref = int(ev_trips["flag_high_fare"].sum()), int(other_trips["flag_high_fare"].sum())
p_ev_fare, p_ref_fare = x_ev / len(ev_trips), x_ref / len(other_trips)
pooled = (x_ev + x_ref) / (len(ev_trips) + len(other_trips))
z_fare = (p_ev_fare - p_ref_fare) / np.sqrt(pooled * (1 - pooled) * (1 / len(ev_trips) + 1 / len(other_trips)))
se_fare = np.sqrt(p_ev_fare * (1 - p_ev_fare) / len(ev_trips) + p_ref_fare * (1 - p_ref_fare) / len(other_trips))
print(f"High-fare share: eVoucher {p_ev_fare * 100:.2f}% vs reference {p_ref_fare * 100:.2f}%; difference "
      f"{(p_ev_fare - p_ref_fare) * 100:+.2f} points (95% CI {(p_ev_fare - p_ref_fare - 1.96 * se_fare) * 100:+.2f} to "
      f"{(p_ev_fare - p_ref_fare + 1.96 * se_fare) * 100:+.2f}), z = {z_fare:.1f}, p = {2 * stats.norm.sf(abs(z_fare)):.3g}")

# Priority for review: count only signals that are NOT implied by the pickup zone.
#   1) weekend or 22:00-04:59 start   2) fare above the zone/trip-type fence   3) fare per km above its fence
# "Pickup outside CBD" is NOT counted: every Airport, Mall, Residential and Other trip has it automatically,
# and the CBD rule is not in the brief. It stays in the file as information only.
ev_trips = ev_trips.copy()
signal_columns = ["is_unusual_time", "flag_high_fare", "flag_high_fare_per_km"]
ev_trips["review_signals"] = ev_trips[signal_columns].astype(int).sum(axis=1)
print("Correlation between the flags (phi; 0 = unrelated). Outside-CBD is shown to explain why it is not counted:")
display(ev_trips[signal_columns + ["flag_pickup_outside_cbd"]].astype(int).corr().round(2))
priority = ev_trips.groupby("review_signals").agg(
    trips=("trip_id", "size"),
    fare_rp=("total_fare", "sum"),
)
priority["pct_of_evoucher_trips"] = priority["trips"] / len(ev_trips) * 100
priority["pct_of_evoucher_fare"] = priority["fare_rp"] / ev_trips["total_fare"].sum() * 100
display(priority.round(1))
three_signals = ev_trips[ev_trips["review_signals"] == 3]
print(f"Trips with all three signals: {len(three_signals):,} (Rp {three_signals['total_fare'].sum():,.0f}); by pickup zone: {three_signals['pickup_zone_type'].value_counts().to_dict()}")
print(f"Distance of the trips with all three signals: median {three_signals['distance_km_clean'].median():.1f} km, maximum {three_signals['distance_km_clean'].max():.1f} km")

review_candidates = ev_trips[ev_trips["review_signals"] >= 2].sort_values(["review_signals", "total_fare"], ascending=False)
review_columns = ["trip_id", "company_id", "pickup_zone_type", "trip_type", "meter_start_time", "distance_km_clean",
                  "total_fare", "total_fare_fence", "fare_per_km", "flag_high_fare_per_km",
                  "flag_pickup_outside_cbd", "is_unusual_time", "flag_high_fare", "review_signals"]
review_candidates[review_columns].to_csv(exceptions_dir / "evoucher_review_candidates.csv", index=False, encoding="utf-8-sig")
print(f"Trips with 2 or 3 signals saved for review: {len(review_candidates):,} (evoucher_review_candidates.csv)")
print("Review candidates by pickup zone (%):", (review_candidates["pickup_zone_type"].value_counts(normalize=True) * 100).round(1).to_dict())
print("Highest-fare trips with all three signals:")
display(review_candidates.loc[review_candidates["review_signals"] == 3, review_columns].head(10))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> tarif eVoucher tidak lebih sering tinggi dibanding pembanding (3,43% vs 3,55%; selisih −0,12 poin, CI 95% −0,29 sampai +0,05; p = 0,17), dan tarif per km yang tinggi juga mirip (5,3% vs 5,2%). Median tarif eVoucher Rp 64.500 vs Rp 59.500 (karena porsi Airport lebih besar). <strong>Prioritas tinjauan</strong> hanya menghitung sinyal yang tidak otomatis ikut zona: waktu tidak lazim, tarif di atas batas zona dan tipe trip, dan tarif per km di atas batasnya. "Pickup di luar CBD" tidak dihitung karena setiap trip Airport, Mall, Residential, dan Other otomatis memilikinya dan aturannya tidak ada di brief. Ketiga sinyal yang dihitung hampir tidak berkorelasi (phi −0,01 sampai 0,03), sehingga masing-masing menambah informasi baru. Hasilnya: 0 sinyal 67,5% trip, 1 sinyal 30,2%, 2 sinyal 1.205 trip (2,1%), dan 3 sinyal 85 trip (0,2%; Rp 23,8 juta). Sebanyak 1.290 trip dengan 2–3 sinyal disimpan di <code>outputs/exceptions/evoucher_review_candidates.csv</code> (52,7% Airport). Trip 3 sinyal punya jarak median 35,4 km, tetapi tarif per km-nya tetap di atas batas, sehingga tarifnya tidak dijelaskan jarak saja. Sinyal hanya menentukan urutan tinjauan; tidak ada yang membuktikan kecurangan.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">

### <span style="color:#0061AF">2.6.9 Sebaran Trip eVoucher antar Pengemudi dan Pasangan Pengemudi–Perusahaan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Brief meminta pola pemakaian eVoucher dilihat juga menurut pengemudi, bukan hanya perusahaan. Unit analisis: satu pengemudi. Pembilang: trip eVoucher pengemudi. Penyebut: seluruh trip pengemudi itu. Pengemudi ditandai hanya jika porsi eVoucher-nya jauh di atas porsi seluruh armada (uji z satu sisi, batas Bonferroni). Ini sinyal untuk ditinjau, bukan bukti kecurangan.
</p>


```python
from scipy import stats

# One row per driver: all trips and eVoucher trips (ev comes from 2.6.2, one row per trip).
driver_ev = ev.groupby("driver_id").agg(
    total_trips=("trip_id", "size"),
    evoucher_trips=("is_evoucher", "sum"),
    evoucher_fare=("total_fare", lambda s: s[ev.loc[s.index, "is_evoucher"]].sum()),
)
driver_ev["evoucher_share"] = driver_ev["evoucher_trips"] / driver_ev["total_trips"]

fleet_share = ev["is_evoucher"].mean()   # share of all trips paid by eVoucher
print(f"Fleet-wide eVoucher share of trips: {fleet_share:.2%}")
print(f"Drivers with trips: {len(driver_ev):,} | with at least one eVoucher trip: {(driver_ev['evoucher_trips'] > 0).sum():,}")

# Concentration: how much of all eVoucher trips come from the top 1% / top 10 drivers?
by_volume = driver_ev["evoucher_trips"].sort_values(ascending=False)
top_1pct = by_volume.head(max(1, int(len(by_volume) * 0.01))).sum() / by_volume.sum()
top_10 = by_volume.head(10).sum() / by_volume.sum()
print(f"Share of eVoucher trips from the top 1% of drivers: {top_1pct:.1%} (an even spread would give about 1%)")
print(f"Share of eVoucher trips from the top 10 drivers: {top_10:.1%}")

# One-sided z-test per driver against the fleet share, Bonferroni limit for many drivers.
min_trips = 30
tested = driver_ev[driver_ev["total_trips"] >= min_trips].copy()
standard_error = np.sqrt(fleet_share * (1 - fleet_share) / tested["total_trips"])
tested["z_score"] = (tested["evoucher_share"] - fleet_share) / standard_error
z_limit = stats.norm.isf(0.05 / len(tested))
tested["flag_high_evoucher_share"] = tested["z_score"] > z_limit
print(f"Drivers tested (>= {min_trips} trips): {len(tested):,} | Bonferroni z limit: {z_limit:.2f}")
print(f"Drivers above the limit: {tested['flag_high_evoucher_share'].sum()}")
display(tested.sort_values("z_score", ascending=False).head(10).round(3))

# Repeated driver-company pairs: the same driver serving the same company many times.
pair_counts = ev[ev["is_evoucher"]].groupby(["driver_id", "company_id"]).size()
print("\nTrips per driver-company pair (eVoucher only):")
display(pair_counts.describe().round(2).to_frame("pair_trips"))
print(f"Largest pair count: {pair_counts.max()} trips | pairs with 10+ trips: {(pair_counts >= 10).sum()} of {len(pair_counts):,}")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> trip eVoucher tersebar merata antar pengemudi. Seluruh 4.055 pengemudi pernah melayani trip eVoucher; 1% pengemudi teratas hanya menyumbang 2,0% trip eVoucher dan 10 teratas 0,5%. Pasangan pengemudi–perusahaan tidak berulang (maksimum 4 trip, tidak ada pasangan dengan 10+ trip), sehingga tidak terlihat pola kolusi satu pengemudi dengan satu perusahaan. Satu dari 4.043 pengemudi melewati batas Bonferroni (<code>DRV-HLM-17-0014</code>, 33 dari 89 trip, 37% vs 18,0% armada, z = 4,67). Satu pengemudi dari ribuan yang diuji adalah sinyal lemah dan bisa terjadi karena kebetulan, sehingga cukup dimasukkan ke daftar tinjauan sampel, bukan dianggap pelanggaran. Keterbatasan: hanya tiga bulan, dan pengemudi dengan sedikit trip kurang bertenaga untuk diuji.
</p>


### <span style="color:#0061AF">2.6.10 Transaksi eVoucher yang Berdekatan (Kemungkinan Tagihan Ganda)</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Salah satu pola kecurangan atau kesalahan tagihan adalah dua transaksi perusahaan yang sama hampir bersamaan. Dihitung pasangan trip eVoucher dari perusahaan yang sama dan zona penjemputan yang sama yang dimulai dalam 2 atau 10 menit, lalu dibandingkan dengan jumlah yang terjadi secara kebetulan (uji permutasi). Diperiksa juga apakah satu pengemudi terlalu sering melayani perusahaan yang sama berturut-turut.
</p>


```python
# Possible duplicate or split charges: two eVoucher trips of the SAME company from the SAME pickup zone that start within a few minutes.
# Some of this happens by chance (busy companies, many trips). A permutation test tells how many pairs chance alone produces:
# company labels are shuffled 200 times while the trip times stay the same.
evoucher_trips = trips_eda.loc[trips_eda["company_id"].notna(), ["trip_id", "company_id", "driver_id", "pickup_zone_type", "meter_start_time", "total_fare"]].reset_index(drop=True)

def close_pair_flags(data, company_labels, minutes):
    """True for a trip that starts < minutes after the previous trip of the same company in the same pickup zone."""
    ordered = data.assign(company=company_labels).sort_values(["company", "meter_start_time"])
    gap = ordered.groupby("company")["meter_start_time"].diff().dt.total_seconds() / 60
    same_zone = ordered["pickup_zone_type"].eq(ordered.groupby("company")["pickup_zone_type"].shift())
    return ((gap < minutes) & same_zone).reindex(data.index)

observed = {m: int(close_pair_flags(evoucher_trips, evoucher_trips["company_id"].to_numpy(), m).sum()) for m in (2, 10)}
rng = np.random.default_rng(42)
shuffled = {2: [], 10: []}
for _ in range(200):
    labels = rng.permutation(evoucher_trips["company_id"].to_numpy())
    for m in (2, 10):
        shuffled[m].append(int(close_pair_flags(evoucher_trips, labels, m).sum()))
close_pair_table = pd.DataFrame({
    "observed pairs": observed,
    "expected by chance (mean)": {m: np.mean(v) for m, v in shuffled.items()},
    "excess pairs": {m: observed[m] - np.mean(v) for m, v in shuffled.items()},
    "p-value (share of shuffles >= observed)": {m: (np.sum(np.array(v) >= observed[m]) + 1) / (len(v) + 1) for m, v in shuffled.items()},
})
close_pair_table.index.name = "within minutes"
display(close_pair_table.round(3))

# Same driver serving the same company on consecutive eVoucher trips (possible collusion signal)
by_driver = evoucher_trips.sort_values(["driver_id", "meter_start_time"])
same_company_next = int((by_driver["company_id"] == by_driver.groupby("driver_id")["company_id"].shift()).sum())
shuffled_driver = []
for _ in range(200):
    labels = rng.permutation(evoucher_trips["company_id"].to_numpy())
    shuffled_by_driver = evoucher_trips.assign(company=labels).sort_values(["driver_id", "meter_start_time"])
    shuffled_driver.append(int((shuffled_by_driver["company"] == shuffled_by_driver.groupby("driver_id")["company"].shift()).sum()))
print(f"Consecutive eVoucher trips of one driver for the same company: {same_company_next} "
      f"(expected by chance: {np.mean(shuffled_driver):.0f})")

# Save the < 2 minute pairs (both trips of each pair) for review
flag_2 = close_pair_flags(evoucher_trips, evoucher_trips["company_id"].to_numpy(), 2)
ordered = evoucher_trips.sort_values(["company_id", "meter_start_time"])
previous_of_flagged = ordered.groupby("company_id")["trip_id"].shift()[flag_2.reindex(ordered.index)]
close_pair_trip_ids = set(evoucher_trips.loc[flag_2, "trip_id"]) | set(previous_of_flagged.dropna())
evoucher_close_pairs = evoucher_trips[evoucher_trips["trip_id"].isin(close_pair_trip_ids)].sort_values(["company_id", "meter_start_time"])
evoucher_close_pairs.to_csv(exceptions_dir / "evoucher_close_pairs.csv", index=False, encoding="utf-8-sig")
print(f"Trips in < 2 minute same-company pairs saved for review: {len(evoucher_close_pairs):,} "
      f"(Rp {evoucher_close_pairs['total_fare'].sum():,.0f}) -> outputs/exceptions/evoucher_close_pairs.csv")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> ada sedikit lebih banyak pasangan trip eVoucher yang berdekatan daripada yang dihasilkan kebetulan: 170 pasangan dalam 2 menit (harapan 147; p = 0,03) dan 765 dalam 10 menit (harapan 713; p = 0,04). Kelebihannya kecil (sekitar 23 dan 52 pasangan) dan tidak lolos koreksi Bonferroni untuk dua uji (0,025), sehingga ini <strong>sinyal lemah</strong>. Pasangan seperti ini bisa sah, misalnya dua karyawan memesan dua taksi dari kantor yang sama. Sebanyak 337 trip dalam pasangan &lt; 2 menit (Rp 29,8 juta) disimpan di <code>outputs/exceptions/evoucher_close_pairs.csv</code> untuk dicocokkan dengan nama penumpang. Pengemudi yang sama melayani perusahaan yang sama berturut-turut 271 kali, lebih sedikit dari harapan kebetulan (291), sehingga tidak ada indikasi kolusi pengemudi dengan perusahaan.
</p>


### <span style="color:#0061AF">2.6.11 Pemakaian eVoucher per Zona dan Hari</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Pertanyaan:</strong> ke mana trip eVoucher di akhir pekan, dan bagaimana pemakaian eVoucher per zona dan per hari, termasuk total tarif dan median tarif? <strong>Catatan penting:</strong> data tidak punya zona tujuan, sehingga "ke mana" hanya bisa dijawab dengan <strong>zona penjemputan</strong> (dari mana trip dimulai) dan jarak tempuh. Unit analisis: satu trip eVoucher. Total dihitung untuk seluruh periode; rata-rata per hari hanya memakai hari penuh (2023-12-30 terekam sebagian).
</p>


```python
# eVoucher usage per pickup zone and day of week: number of trips, total fare and median fare.
# Unit: one eVoucher trip (payment_method = "eVoucher Corporate"; all of them have a company_id).
evoucher_usage = trips_eda[trips_eda["payment_method"] == "eVoucher Corporate"].copy()
assert evoucher_usage["company_id"].notna().all()
day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
evoucher_usage["day"] = pd.Categorical(evoucher_usage["meter_start_time"].dt.day_name(), day_order, ordered=True)
evoucher_usage["is_weekend"] = evoucher_usage["meter_start_time"].dt.dayofweek >= 5

# 1) Totals over the whole period (2023-10-01 to 2023-12-30)
ev_day_trips = evoucher_usage.pivot_table(index="pickup_zone_type", columns="day", values="trip_id", aggfunc="size", observed=True).loc[zones_in_order]
ev_day_fare = evoucher_usage.pivot_table(index="pickup_zone_type", columns="day", values="total_fare", aggfunc="sum", observed=True).loc[zones_in_order]
ev_day_median = evoucher_usage.pivot_table(index="pickup_zone_type", columns="day", values="total_fare", aggfunc="median", observed=True).loc[zones_in_order]
assert int(ev_day_trips.to_numpy().sum()) == len(evoucher_usage)
assert abs(ev_day_fare.to_numpy().sum() - evoucher_usage["total_fare"].sum()) < 1
print("eVoucher trips per zone and day (whole period):")
display(ev_day_trips)
print("Total eVoucher fare per zone and day (Rp million, whole period):")
display((ev_day_fare / 1e6).round(1))
print("Median eVoucher fare per trip (Rp):")
display(ev_day_median.round(0))

# 2) Average PER DAY. Only full days are used: 2023-12-30 (a Saturday) is only partly recorded.
full_days = evoucher_usage[evoucher_usage["meter_start_time"] < pd.Timestamp("2023-12-30")]
dates_per_weekday = full_days.groupby("day", observed=True)["meter_start_time"].apply(lambda x: x.dt.normalize().nunique())
ev_per_day = pd.DataFrame({
    "days_in_data": dates_per_weekday,
    "trips_per_day": full_days.groupby("day", observed=True).size() / dates_per_weekday,
    "fare_per_day_rp_million": full_days.groupby("day", observed=True)["total_fare"].sum() / dates_per_weekday / 1e6,
    "median_fare_rp": full_days.groupby("day", observed=True)["total_fare"].median(),
})
ev_trips_per_day_zone = (full_days.pivot_table(index="pickup_zone_type", columns="day", values="trip_id", aggfunc="size", observed=True)
                         .loc[zones_in_order] / dates_per_weekday)
ev_fare_per_day_zone = (full_days.pivot_table(index="pickup_zone_type", columns="day", values="total_fare", aggfunc="sum", observed=True)
                        .loc[zones_in_order] / dates_per_weekday / 1e6)
print("Average per day (full days only):")
display(ev_per_day.round(1))

# 3) Weekend vs weekday by pickup zone (the data has NO destination, so this is where trips START)
ev_weekend_zone = evoucher_usage.groupby(["is_weekend", "pickup_zone_type"]).agg(
    trips=("trip_id", "size"), total_fare_rp=("total_fare", "sum"),
    median_fare_rp=("total_fare", "median"), median_distance_km=("distance_km_clean", "median"),
)
ev_weekend_zone["pct_of_trips_in_period"] = ev_weekend_zone["trips"] / ev_weekend_zone.groupby(level="is_weekend")["trips"].transform("sum") * 100
ev_weekend_zone["pct_of_fare_in_period"] = ev_weekend_zone["total_fare_rp"] / ev_weekend_zone.groupby(level="is_weekend")["total_fare_rp"].transform("sum") * 100
ev_weekend_zone = ev_weekend_zone.rename(index={True: "Weekend", False: "Weekday"}, level="is_weekend")
display(ev_weekend_zone.round(1))

# 4) When do weekend Airport eVoucher trips start?
weekend_airport = evoucher_usage[evoucher_usage["is_weekend"] & (evoucher_usage["pickup_zone_type"] == "Airport")]
weekend_airport_hours = weekend_airport["meter_start_time"].dt.hour.value_counts(normalize=True).sort_index() * 100
print(f"Weekend Airport eVoucher trips starting 12:00-16:59: {weekend_airport_hours.loc[12:16].sum():.1f}%; "
      f"22:00-04:59: {weekend_airport_hours.loc[[22, 23, 0, 1, 2, 3, 4]].sum():.1f}%")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> pada hari kerja ada sekitar 677–698 trip eVoucher per hari, dan pada akhir pekan 453–471 (turun sekitar sepertiga). Total tarif per hari tetap hampir sama setiap hari (Rp 52,3–58,8 juta), karena trip akhir pekan didominasi trip Airport yang mahal. Trip eVoucher Airport naik dari 88–97 per hari pada hari kerja menjadi 249–256 pada Sabtu–Minggu, sedangkan zona lain turun ke sekitar sepertiga. Selama periode data, akhir pekan Airport menyumbang 54,6% trip dan 71,2% tarif eVoucher akhir pekan (Rp 955,9 juta), dibanding 13,5% dan 23,7% pada hari kerja. Median tarif per trip di setiap zona hampir tidak berubah antar hari (Airport Rp 125.500–132.500; CBD Rp 53.000–53.500), dan median jarak Airport sama (23,2 km akhir pekan vs 23,3 km hari kerja): <strong>yang berubah adalah jumlah trip, bukan biaya per trip</strong>. Median gabungan per hari naik ke Rp 94.000–95.500 pada akhir pekan hanya karena porsi Airport lebih besar. Sebanyak 58,7% trip Airport akhir pekan dimulai pukul 12:00–16:59 dan hanya 11,8% pukul 22:00–04:59, sehingga sebagian besar bukan trip larut malam. <strong>Ke mana?</strong> Data hanya menunjukkan dari mana trip dimulai: di akhir pekan eVoucher terutama dipakai untuk trip yang <em>dimulai</em> dari bandara (misalnya karyawan atau tamu yang tiba). Tujuannya tidak tercatat. Blue Bird menyebut Corporate Solution melayani "karyawan atau tamu perusahaan", sehingga pola ini bisa sah. <strong>Implikasi:</strong> pertanyaan untuk tim finance klien adalah apakah kebijakan mereka mengizinkan trip bandara di akhir pekan, bukan apakah tarifnya digelembungkan.
</p>


## <span style="color:#0061AF">2.7 Hubungan Antar Variabel dan Uji Statistik</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Bagian ini menguji apakah pola yang teramati berarti secara statistik. Karena idle time dihitung dari waktu trip, menjelaskan idle sama dengan menjelaskan mengapa sebuah shift punya sedikit trip.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Bagian ini menjawab pertanyaan: <strong>mengapa pengemudi idle di sebuah pool? Apakah karena tempat ramai, atau ada variabel lain?</strong> Karena idle time dihitung dari waktu trip (2.5.8), menjelaskan idle sama dengan menjelaskan mengapa sebuah shift punya sedikit trip. Faktor-faktor diuji terhadap <code>trip_count</code> per shift.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Hipotesis:</strong> H1 armada punya jam shift jauh lebih banyak dari permintaan; H2 tempat (zona) dan jumlah pengemudi yang bersaing (kepadatan); H3 waktu mulai shift; H4 pool itu sendiri; H5 karakteristik pengemudi (status kerja, kendaraan, rating, masa kerja); H6 kalender (hari, tren mingguan). Karena dengan 252.524 shift hampir semua perbedaan signifikan, setiap uji dipasangkan dengan <em>effect size</em>.
</p>

### <span style="color:#0061AF">2.7.1 Uji Hipotesis: Apa yang Menjelaskan Trip per Shift?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<code>trip_count</code> adalah data hitung (0–10) yang tidak berdistribusi normal, sehingga dipakai uji berbasis peringkat (Kruskal-Wallis, Mann-Whitney U, Spearman) dan chi-square. Setiap uji dipasangkan dengan <em>effect size</em>, dan karena banyak uji dijalankan dipakai level signifikansi <strong>Bonferroni</strong>.
</p>

| Effect size | Diabaikan | Kecil | Sedang | Besar |
|---|---|---|---|---|
| epsilon-squared (Kruskal-Wallis) | < 0,01 | 0,01–0,06 | 0,06–0,14 | > 0,14 |
| rank-biserial r, Spearman rho, Cramer's V | < 0,1 | 0,1–0,3 | 0,3–0,5 | > 0,5 |

```python
from scipy import stats

test_rows = []

def effect_label(value, kind):
    """Plain-language size of an effect (thresholds in the table above)."""
    value = abs(value)
    limits = {"epsilon2": (0.01, 0.06, 0.14), "h": (0.2, 0.5, 0.8)}.get(kind, (0.1, 0.3, 0.5))
    for limit, label in zip(limits, ["negligible", "small", "medium"]):
        if value < limit:
            return label
    return "large"

def format_p(p_value):
    return "< 1e-300" if p_value < 1e-300 else f"{p_value:.2e}"

def add_test(name, question, h0, statistic, p_value, effect_name, effect_value, kind):
    test_rows.append({
        "test": name, "question": question, "H0 (no effect)": h0,
        "statistic": statistic, "p_value": p_value,
        "effect size": f"{effect_name} = {effect_value:.3f}", "effect_kind": kind,
        "effect_label": effect_label(effect_value, kind),
    })

trips = shifts_eda["trip_count"]
n_shifts = len(shifts_eda)

# --- Kruskal-Wallis: do 3+ groups have the same distribution of trips per shift?
for column, question in [("dominant_zone_worked", "Do zones differ in trips per shift?"),
                         ("pool_id", "Do pools differ in trips per shift?"),
                         ("vehicle_type_assigned", "Do vehicle types differ?"),
                         ("weekday", "Do weekdays differ?")]:
    groups = [group.to_numpy() for _, group in trips.groupby(shifts_eda[column])]
    h_stat, p_value = stats.kruskal(*groups)
    add_test("Kruskal-Wallis H", question, "all groups have the same distribution", h_stat, p_value,
             "epsilon-squared", h_stat / (n_shifts - 1), "epsilon2")

# --- Mann-Whitney U: do 2 groups differ? (effect = rank-biserial correlation)
for column, group_a, group_b, question in [
        ("start_group", "Morning start", "Evening start", "Do morning-start shifts get more trips than evening-start shifts?"),
        ("employment_status", "Tetap", "Mitra", "Do permanent (Tetap) and partner (Mitra) drivers differ?")]:
    sample_a = trips[shifts_eda[column] == group_a].to_numpy()
    sample_b = trips[shifts_eda[column] == group_b].to_numpy()
    u_stat, p_value = stats.mannwhitneyu(sample_a, sample_b, alternative="two-sided")
    rank_biserial = 2 * u_stat / (len(sample_a) * len(sample_b)) - 1
    add_test("Mann-Whitney U", question, "the two groups have the same distribution", u_stat, p_value,
             f"rank-biserial r ({group_a} higher if > 0)", rank_biserial, "r")

# --- Chi-square test of independence: is a ZERO-trip shift linked to the zone?
zero_by_zone = pd.crosstab(shifts_eda["dominant_zone_worked"], shifts_eda["zero_trip_shift"])
chi2_value, p_value, _, _ = stats.chi2_contingency(zero_by_zone)
cramers_v = np.sqrt(chi2_value / (n_shifts * (min(zero_by_zone.shape) - 1)))
add_test("Chi-square independence", "Is having 0 trips independent of the zone?", "zone and 'zero-trip shift' are independent",
         chi2_value, p_value, "Cramer's V", cramers_v, "r")
display(pd.crosstab(shifts_eda["dominant_zone_worked"], shifts_eda["zero_trip_shift"], normalize="index").round(3)
        .rename(columns={False: "has trips", True: "zero trips"}))
```

### <span style="color:#0061AF">2.7.2 Apakah Karena Tempat "Ramai"? Indeks Kepadatan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Kepadatan diukur sebagai <strong>request per pengemudi bertugas</strong> (request di zona selama jam shift dibagi rata-rata pengemudi bertugas di zona yang sama). Nilai rendah berarti banyak pengemudi untuk sedikit request. Request hanya mencakup App dan Call Center, sehingga permintaan nyata per pengemudi sekitar 15% lebih tinggi.
</p>

```python
def slot_counts(timestamps):
    """Number of events in each hour slot since 2023-10-01 (full 90 days only)."""
    slot = ((timestamps.dt.floor("h") - FIRST_DAY) // pd.Timedelta(hours=1)).to_numpy()
    slot = slot[(slot >= 0) & (slot < N_SLOTS)]
    return np.bincount(slot, minlength=N_SLOTS)

def crowding_for(target_shifts, request_times, duty_shifts):
    """Requests per average on-duty driver during each target shift's hours.
    request_times: request timestamps that count as demand; duty_shifts: shifts that count as supply."""
    demand_slots = slot_counts(request_times)
    first = ((duty_shifts["clock_in_time"].dt.floor("h") - FIRST_DAY) // pd.Timedelta(hours=1)).to_numpy()
    last = np.minimum(((duty_shifts["clock_out_time"].dt.floor("h") - FIRST_DAY) // pd.Timedelta(hours=1)).to_numpy(), N_SLOTS - 1)
    size = N_SLOTS + 100
    on_duty_slots = np.cumsum(np.bincount(first, minlength=size)[:size] - np.bincount(last + 1, minlength=size)[:size])[:N_SLOTS]
    cum_demand = np.concatenate([[0], np.cumsum(demand_slots)])
    cum_duty = np.concatenate([[0], np.cumsum(on_duty_slots)])
    t_first = ((target_shifts["clock_in_time"].dt.floor("h") - FIRST_DAY) // pd.Timedelta(hours=1)).to_numpy()
    t_last = np.minimum(((target_shifts["clock_out_time"].dt.floor("h") - FIRST_DAY) // pd.Timedelta(hours=1)).to_numpy(), N_SLOTS - 1)
    window_demand = cum_demand[t_last + 1] - cum_demand[t_first]
    average_drivers = (cum_duty[t_last + 1] - cum_duty[t_first]) / (t_last - t_first + 1)
    return pd.Series(window_demand / average_drivers, index=target_shifts.index)

# City-wide version (all requests, all drivers) and zone version (same zone only).
shifts_eda["requests_per_driver_city"] = crowding_for(shifts_eda, req["requested_time"], shifts_eda)
shifts_eda["requests_per_driver_zone"] = np.nan
for zone in zones_in_order:
    in_zone = shifts_eda["dominant_zone_worked"] == zone
    shifts_eda.loc[in_zone, "requests_per_driver_zone"] = crowding_for(
        shifts_eda[in_zone], req.loc[req["pickup_zone_type"] == zone, "requested_time"], shifts_eda[in_zone])
assert shifts_eda[["requests_per_driver_city", "requests_per_driver_zone"]].notna().all().all()

print("Requests per on-duty driver during the shift (higher = less crowded with drivers):")
display(shifts_eda.groupby("dominant_zone_worked")[["requests_per_driver_zone", "requests_per_driver_city"]].mean().loc[zones_in_order].round(2))

shifts_eda["crowding_group"] = pd.qcut(shifts_eda["requests_per_driver_city"], 5,
                                       labels=["1 most crowded with drivers", "2", "3", "4", "5 least crowded"])
crowding_table = shifts_eda.groupby("crowding_group", observed=True).agg(
    shifts=("shift_id", "size"),
    avg_requests_per_driver=("requests_per_driver_city", "mean"),
    avg_trips_per_shift=("trip_count", "mean"),
    zero_trip_pct=("zero_trip_shift", lambda s: s.mean() * 100),
    avg_idle_share_pct=("idle_share_pct", "mean"),
)
display(crowding_table.round(2))

# --- Spearman rank correlations (trips vs a continuous variable)
for column, label, subset in [("requests_per_driver_city", "requests per driver, city-wide", None),
                              ("requests_per_driver_zone", "requests per driver, same zone", None),
                              ("driver_rating", "driver rating", None),
                              ("tenure_days", "tenure in days (shifts on/after join date only)", None)]:
    valid = shifts_eda[column].notna()
    rho, p_value = stats.spearmanr(shifts_eda.loc[valid, column], shifts_eda.loc[valid, "trip_count"])
    add_test("Spearman correlation", f"Does {label} move with trips per shift? (n = {valid.sum():,})",
             "rho = 0 (no monotonic relation)", rho, p_value, "Spearman rho", rho, "r")

# Within each zone separately (so the zone itself cannot create the pattern).
within_zone = {zone: stats.spearmanr(shifts_eda.loc[shifts_eda["dominant_zone_worked"] == zone, "requests_per_driver_zone"],
                                      shifts_eda.loc[shifts_eda["dominant_zone_worked"] == zone, "trip_count"])[0]
               for zone in zones_in_order}
print("Spearman rho between requests per driver (same zone) and trips, inside each zone:")
display(pd.Series(within_zone).round(3).to_frame("rho"))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> "ramai pengemudi" terlihat <strong>antar zona, bukan di dalam zona</strong>. Airport hanya 0,61 request per pengemudi vs 1,38 di CBD. Seperlima shift dengan request per pengemudi terendah (0,73) punya trip paling sedikit (1,10 per shift, 35,5% tanpa trip), tetapi di atas sekitar 1 request per pengemudi tambahan permintaan tidak menambah trip (1,31–1,32). Di dalam setiap zona hubungannya diabaikan (Spearman rho 0,05–0,06). Rating (p = 0,18) dan masa kerja (p = 0,05) tidak berhubungan dengan jumlah trip.
</p>

### <span style="color:#0061AF">2.7.3 Model Regresi: Semua Faktor Bersama dan Uji Pool</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>OLS + F-test model bertingkat</strong> menguji apakah <code>pool_id</code> memperbaiki model setelah zona, waktu mulai, status kerja, kendaraan, dan hari dikendalikan (H0: pool tidak menambah apa pun). <strong>Regresi Poisson</strong> (robust standard error) menghasilkan <em>rate ratio</em>. <strong>Regresi logistik</strong> menghasilkan odds shift tanpa trip. Sel ini membutuhkan <code>statsmodels</code>.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Penjelasan sederhana.</strong> Di 2.5.3 terlihat pool berbeda-beda: PL-CGK rata-rata 0,97 trip per shift, sedangkan pool CBD-heavy 1,39–1,40. Tetapi pool juga berbeda komposisi zonanya (PL-CGK 55% shift di Airport). Pertanyaannya: <em>apakah pool itu sendiri yang membuat perbedaan, atau hanya karena zonanya?</em> Regresi menjawabnya dengan membandingkan shift yang zona, jam mulai, status kerja, kendaraan, dan harinya sama, lalu melihat apakah pool masih membuat perbedaan.
</p>

| Uji | Pertanyaan | Cara membaca hasil |
|---|---|---|
| F-test model bertingkat (OLS) | Apakah menambahkan pool membuat model lebih baik? | p &lt; 0,05 = pool berpengaruh; p ≥ 0,05 = tidak ada bukti pool berpengaruh |
| Regresi Poisson (<em>rate ratio</em>) | Berapa kali lipat trip per shift dibanding kelompok pembanding? | 1,00 = sama; 2,49 = 2,49 kali lipat |
| Regresi logistik (<em>odds ratio</em>) | Seberapa besar peluang shift tanpa trip dibanding kelompok pembanding? | di bawah 1 = lebih jarang tanpa trip; 0,22 = sekitar 78% lebih rendah |

```python
try:
    import statsmodels.api as sm
    import statsmodels.formula.api as smf
    have_statsmodels = True
except ImportError:
    have_statsmodels = False
    print("statsmodels is not installed: skip this cell (pip install statsmodels).")

if have_statsmodels:
    model_data = shifts_eda.rename(columns={"dominant_zone_worked": "zone", "pool_id": "pool", "employment_status": "employment",
                                            "vehicle_type_assigned": "vehicle"})
    model_data["zero_trip"] = model_data["zero_trip_shift"].astype(int)
    base_terms = "C(zone, Treatment('Airport')) + C(start_group, Treatment('Evening start')) + C(employment) + C(vehicle) + C(weekday)"

    # 1) Nested F-test for the pool effect after controlling zone etc.
    model_without_pool = smf.ols(f"trip_count ~ {base_terms}", data=model_data).fit()
    model_with_pool = smf.ols(f"trip_count ~ {base_terms} + C(pool)", data=model_data).fit()
    f_table = sm.stats.anova_lm(model_without_pool, model_with_pool)
    f_stat, f_p = f_table.loc[1, "F"], f_table.loc[1, "Pr(>F)"]
    partial_eta2_pool = f_table.loc[1, "ss_diff"] / (f_table.loc[1, "ss_diff"] + model_with_pool.ssr)
    print("Nested F-test: does pool matter once zone, start time, employment, vehicle and weekday are controlled?")
    print(f"  F = {f_stat:.2f}, p = {f_p:.3f}, partial eta-squared = {partial_eta2_pool:.5f}")
    print(f"  Explained variance (R-squared) without pool: {model_without_pool.rsquared:.4f} | with pool: {model_with_pool.rsquared:.4f}")
    add_test("Nested-model F-test (OLS)", "Does pool explain trips per shift AFTER controlling zone, start time, employment, vehicle, weekday?",
             "pool coefficients are all 0", f_stat, f_p, "partial eta-squared", partial_eta2_pool, "epsilon2")

    # 2) Poisson regression (rate ratios) and 3) logistic regression (odds of a zero-trip shift)
    poisson_model = smf.glm(f"trip_count ~ {base_terms} + C(pool)", data=model_data, family=sm.families.Poisson()).fit(cov_type="HC0")
    dispersion = (poisson_model.resid_pearson ** 2).sum() / poisson_model.df_resid
    print(f"Poisson dispersion check (about 1 = Poisson is a good fit): {dispersion:.2f}")
    logit_model = smf.logit(f"zero_trip ~ {base_terms} + C(pool)", data=model_data).fit(disp=0)

    def tidy(model, exponentiate_label):
        table = pd.DataFrame({
            exponentiate_label: np.exp(model.params),
            "ci_low": np.exp(model.conf_int()[0]),
            "ci_high": np.exp(model.conf_int()[1]),
            "p_value": model.pvalues,
        }).drop(index="Intercept")
        table.index = (table.index.str.replace(r"C\((\w+)(, Treatment\('[^']*'\))?\)\[T\.", r"\1: ", regex=True).str.rstrip("]"))
        return table

    rate_ratios = tidy(poisson_model, "rate_ratio")
    odds_ratios = tidy(logit_model, "odds_ratio")
    print("Poisson regression: rate ratios for trips per shift (reference: Airport zone, Evening start, first category of the others)")
    display(rate_ratios.round(3))
    print(f"Logistic regression: odds ratios for a ZERO-trip shift (pseudo R-squared = {logit_model.prsquared:.3f})")
    display(odds_ratios.round(3))
    print(f"OLS R-squared with all factors: {model_with_pool.rsquared:.3f} -> {(1 - model_with_pool.rsquared) * 100:.0f}% of the variation in trips per shift is NOT explained by these factors.")
```

```python
# How much of the variation in trips per shift can ANY model explain?
# Trips in a shift are counts. If they behave like Poisson counts, their pure random variance equals their mean,
# and that part can never be explained by extra variables.
zone_dispersion = shifts_eda.groupby("dominant_zone_worked")["trip_count"].agg(["mean", "var"])
zone_dispersion["variance / mean"] = zone_dispersion["var"] / zone_dispersion["mean"]
display(zone_dispersion.round(3))

total_variance = shifts_eda["trip_count"].var()
random_variance = shifts_eda["trip_count"].mean()          # Poisson: variance = mean
max_explainable_r2 = 1 - random_variance / total_variance
print(f"Total variance {total_variance:.3f}; Poisson random variance about {random_variance:.3f}")
print(f"Largest R-squared any shift-level model could reach: about {max_explainable_r2 * 100:.1f}%")
if have_statsmodels:
    print(f"Our model reaches {model_with_pool.rsquared * 100:.1f}%, i.e. about "
          f"{model_with_pool.rsquared / max_explainable_r2 * 100:.0f}% of what is explainable.")
```

```python
# Robustness: each driver works about 62 shifts, so shifts are not independent.
# Refit the Poisson model with standard errors clustered by driver and check whether the conclusions change.
if have_statsmodels:
    poisson_formula = f"trip_count ~ {base_terms} + C(pool)"
    driver_groups = pd.factorize(model_data["driver_id"])[0]
    poisson_clustered = smf.glm(poisson_formula, data=model_data, family=sm.families.Poisson()).fit(
        cov_type="cluster", cov_kwds={"groups": driver_groups})
    assert len(poisson_clustered.fittedvalues) == len(model_data)          # no rows dropped

    se_ratio = poisson_clustered.bse / poisson_model.bse
    print(f"Shifts per driver: {len(model_data) / model_data['driver_id'].nunique():.1f}")
    print(f"Clustered / simple standard errors: from {se_ratio.min():.2f} to {se_ratio.max():.2f} times")

    pool_terms = [name for name in poisson_clustered.params.index if name.startswith("C(pool)")]
    restriction = np.zeros((len(pool_terms), len(poisson_clustered.params)))
    for row, name in enumerate(pool_terms):
        restriction[row, poisson_clustered.params.index.get_loc(name)] = 1
    pool_wald = poisson_clustered.wald_test(restriction, scalar=True)
    print(f"Joint test 'all pool effects = 0' with clustered errors: chi2 = {float(pool_wald.statistic):.2f}, "
          f"p = {float(pool_wald.pvalue):.3f}")
    zone_ci = np.exp(poisson_clustered.conf_int().loc[[n for n in poisson_clustered.params.index if n.startswith("C(zone")]])
    zone_ci.columns = ["ci_low", "ci_high"]
    zone_ci["rate_ratio"] = np.exp(poisson_clustered.params[zone_ci.index])
    display(zone_ci[["rate_ratio", "ci_low", "ci_high"]].round(2))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> setelah mengendalikan zona, waktu mulai, status kerja, kendaraan, dan hari, <strong>pool tidak menambah apa pun</strong> (F-test p = 0,62; partial eta-squared 0,00002; rate ratio semua pool 1,00–1,01). Pada model Poisson, shift CBD punya <strong>2,49 kali</strong> trip shift Airport (95% CI 2,46–2,52), Mall 1,96, Residential 1,81, Other 1,70, dan shift mulai pagi <strong>1,25 kali</strong> shift mulai sore. Status kerja, kendaraan, dan hari bernilai 1,00. Odds shift tanpa trip 78% lebih rendah di CBD (odds ratio 0,22) dan 31% lebih rendah untuk mulai pagi (0,69). Semua faktor bersama menjelaskan <strong>10,4%</strong> variasi trip per shift (R-squared OLS). Angka ini <strong>bukan</strong> berarti 90% penjelasan hilang: di setiap zona varians trip per shift sama dengan rata-ratanya (rasio 1,00–1,01), ciri data hitung Poisson. Variasi acak ini tidak bisa dijelaskan variabel apa pun, sehingga R-squared maksimum yang mungkin dicapai model tingkat shift hanya sekitar 10,5%. Model sudah menjelaskan sekitar 99% dari bagian yang bisa dijelaskan. <strong>Ketahanan:</strong> setiap pengemudi bekerja sekitar 62 shift, sehingga shift tidak independen. Dengan standard error yang dikelompokkan per pengemudi, standard error hanya berubah 0,96–1,04 kali, pool tetap tidak berarti (uji Wald p = 0,78), dan rate ratio zona tidak berubah. Kesimpulan tidak bergantung pada asumsi independensi.
</p>

```python
# --- Formal two-proportion z-tests for two earlier findings (same logic as 2.4.4 and 2.6.5, now with a confidence interval).
def two_proportion_test(success_a, n_a, success_b, n_b):
    """z-test for the difference of two proportions. Returns (difference, z, p, ci_low, ci_high, Cohen's h)."""
    p_a, p_b = success_a / n_a, success_b / n_b
    pooled = (success_a + success_b) / (n_a + n_b)
    z_value = (p_a - p_b) / np.sqrt(pooled * (1 - pooled) * (1 / n_a + 1 / n_b))
    standard_error = np.sqrt(p_a * (1 - p_a) / n_a + p_b * (1 - p_b) / n_b)
    difference = p_a - p_b
    cohens_h = 2 * np.arcsin(np.sqrt(p_a)) - 2 * np.arcsin(np.sqrt(p_b))
    return difference, z_value, 2 * stats.norm.sf(abs(z_value)), difference - 1.96 * standard_error, difference + 1.96 * standard_error, cohens_h

cbd_peak_rows = (req["pickup_zone_type"] == "CBD") & (req["period"] == "Failure-peak hours")
diff, z_value, p_value, low, high, cohens_h = two_proportion_test(req.loc[cbd_peak_rows, "is_failed"].sum(), cbd_peak_rows.sum(),
                                                         req.loc[~cbd_peak_rows, "is_failed"].sum(), (~cbd_peak_rows).sum())
print(f"CBD in failure-peak hours vs all other requests: failure rate difference = {diff * 100:.1f} percentage points "
      f"(95% CI {low * 100:.1f} to {high * 100:.1f})")
add_test("Two-proportion z-test", "Do CBD requests in the 4 failure-peak hours fail more often than all other requests?",
         "both failure rates are equal", z_value, p_value, "Cohen's h", cohens_h, "h")

airport_b2b_flag = airport_b2b["is_unusual_time"]
airport_other_flag = airport_non_b2b["is_unusual_time"]
diff, z_value, p_value, low, high, cohens_h = two_proportion_test(airport_b2b_flag.sum(), len(airport_b2b_flag),
                                                         airport_other_flag.sum(), len(airport_other_flag))
print(f"Airport trips at unusual times, B2B vs Non-B2B: difference = {diff * 100:.1f} percentage points "
      f"(95% CI {low * 100:.1f} to {high * 100:.1f})")
add_test("Two-proportion z-test", "Are B2B (eVoucher) Airport trips at weekends/night more often than Non-B2B Airport trips?",
         "both rates are equal", z_value, p_value, "Cohen's h", cohens_h, "h")

# --- Trend test for the weekly ritase (2.5.7): is the slope different from 0?
trend = stats.linregress(np.arange(len(weekly)), weekly["avg_trips_per_shift"])
rho_trend, p_trend = stats.spearmanr(np.arange(len(weekly)), weekly["avg_trips_per_shift"])
print(f"Weekly ritase trend: slope = {trend.slope:+.5f} trips per week (p = {trend.pvalue:.2f}); Spearman rho with time = {rho_trend:+.2f} (p = {p_trend:.2f})")
add_test("Linear-regression slope t-test", "Is weekly ritase trending up or down over the 12 full weeks?",
         "slope = 0 (no trend)", trend.slope / trend.stderr, trend.pvalue, "Spearman rho with time", rho_trend, "r")
```

```python
# The failure-peak hours were chosen from the same data that the z-test above uses, so that test is circular.
# Honest check: choose the peak hours with October-November only, then look at December (data not used to choose them).
cutoff = pd.Timestamp("2023-12-01")
train_requests = req[req["requested_time"] < cutoff]
test_requests = req[req["requested_time"] >= cutoff]
train_rates = request_rates(train_requests, "hour")
train_peak_hours = sorted(train_rates.index[train_rates["failed_pct"] > 2 * train_requests["is_failed"].mean() * 100])
print("Peak hours chosen from Oct-Nov only:", train_peak_hours, "| hours used in this notebook:", failure_peak_hours)

holdout_rows = []
for zone_label, data in [("All zones", test_requests), ("CBD", test_requests[test_requests["pickup_zone_type"] == "CBD"])]:
    in_peak = data["hour"].isin(train_peak_hours)
    holdout_rows.append({"December requests": zone_label,
                         "peak-hour requests": int(in_peak.sum()),
                         "failed % in peak hours": data.loc[in_peak, "is_failed"].mean() * 100,
                         "failed % in other hours": data.loc[~in_peak, "is_failed"].mean() * 100})
holdout_table = pd.DataFrame(holdout_rows).set_index("December requests")
display(holdout_table.round(1))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> uji z di atas memakai jam puncak yang dipilih dari data yang sama, sehingga nilai p-nya melingkar dan hanya bersifat deskriptif. Pemeriksaan yang jujur: jam puncak dipilih hanya dari data Oktober–November, dan hasilnya tetap pukul 07, 08, 17, dan 18. Pada data Desember yang tidak dipakai untuk memilih jam, request pada jam tersebut gagal 21,7% vs 4,3% pada jam lain, dan di CBD 36,7% vs 5,5%. Pola jam puncak <strong>tervalidasi di luar sampel</strong>, bukan kebetulan dari cara memilih jam.
</p>


### <span style="color:#0061AF">2.7.4 Semua Uji dalam Satu Tabel</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Kolom <code>decision</code> memakai level Bonferroni (0,05 dibagi jumlah uji), dan <code>effect_label</code> menunjukkan apakah perbedaannya berarti secara praktis.
</p>

```python
tests_table = pd.DataFrame(test_rows)
alpha_adjusted = 0.05 / len(tests_table)
tests_table["p_value_text"] = tests_table["p_value"].map(format_p)
tests_table["decision"] = np.where(tests_table["p_value"] < alpha_adjusted, "reject H0", "cannot reject H0")
tests_table["statistic"] = tests_table["statistic"].map(lambda v: f"{v:,.2f}")
print(f"Number of tests: {len(tests_table)} | Bonferroni-adjusted significance level: {alpha_adjusted:.4f}")
display(tests_table[["test", "question", "statistic", "p_value_text", "effect size", "effect_label", "decision"]])
tests_table.drop(columns=["p_value_text"]).to_csv(reports_dir / "statistical_tests_summary.csv", index=False, encoding="utf-8-sig")
print("Saved:", reports_dir / "statistical_tests_summary.csv")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> dengan level Bonferroni 0,0033 (15 uji), uji yang menolak H0 <strong>dan</strong> berarti secara praktis adalah zona (epsilon-squared 0,10, sedang), waktu mulai pagi vs sore (rank-biserial 0,14, kecil), pool (0,015, kecil, tetapi hilang setelah zona dikendalikan), shift tanpa trip menurut zona (Cramer's V 0,26, kecil), serta dua selisih besar pada tingkat kegagalan dan eVoucher (Cohen's h sekitar 0,9, besar). Catatan: uji kegagalan CBD memakai jam yang dipilih dari data yang sama, sehingga nilai p-nya deskriptif; validasinya adalah pemeriksaan data Desember di 2.7.3. Uji yang tidak menolak H0: tipe kendaraan, hari, status kerja, rating, masa kerja, dan tren mingguan. Uji berlabel "negligible" signifikan hanya karena sampel besar.
</p>

### <span style="color:#0061AF">2.8.1 Temuan Utama, Hipotesis, dan Keterbatasan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Semua angka berikut berasal dari sel di atas. <strong>Teramati</strong> berarti ditunjukkan data, <strong>hipotesis</strong> berarti kemungkinan penyebab yang belum terbukti.
</p>

**Teramati**

1. **Fulfilment:** 93,30% terpenuhi, 4,66% dibatalkan (tidak ada pengemudi), 2,04% timeout. Total gagal 19.397 dari 289.622 request (6,70%).
2. **Kegagalan terkonsentrasi pada 4 jam:** 07–08 dan 17–18 gagal 18–22%, dibanding 3,5–6% pada jam lain. Empat jam ini hanya 14,3% request tetapi 45,6% kegagalan.
3. **Ramai bukan berarti gagal:** jam tersibuk (12–16) hanya gagal 4,08%.
4. **CBD pada jam tersebut adalah titik masalah:** gagal 36,55% dibanding 4,4% untuk yang lain (odds ratio sekitar 12). Request ini 7,1% dari total tetapi 38,5% dari kegagalan.
5. **Airport sebaliknya:** kegagalan terendah (2,03%) tanpa lonjakan jam sibuk.
6. **Tidak musiman:** bulan (6,68–6,71%) dan hari (6,59–6,99%) hampir sama.
7. **Waktu tunggu datar:** median 5,0 menit di semua zona dan kelompok jam, sehingga kegagalan jam sibuk berupa request hilang, bukan tunggu lebih lama.
8. **Pasokan per jam:** shift aktif sekitar 490 pada malam, sekitar 2.300 pada 08–12, lalu turun ke 351 pada 17:00 dan 163 pada 18:00. Rasio request per pengemudi tidak menjelaskan kegagalan (korelasi 0,03); di CBD pagi pengemudi berlimpah (0,14 request per pengemudi per jam) tetapi 36,6% request gagal.
9. **Kelebihan kapasitas:** permintaan tercatat hanya butuh sekitar 18,9% jam-pengemudi yang ada jika pengemudi sibuk 50%, atau 31,5% jika sibuk 30%; celah hanya pada pukul 16–17 (2.5.11).
10. **Utilisasi (Q3):** rata-rata 1,26 trip per shift; 30,6% shift tanpa trip; hanya 0,35% shift mencapai 6 trip. Idle sekitar 90% dari durasi shift (sekitar 510 menit).
11. **Ketimpangan zona dan pool:** shift Airport 21,8% dari semua shift tetapi request Airport hanya 12,6% (0,67 trip per shift). CBD sebaliknya: 39,3% request tetapi 31,0% shift (1,70 trip per shift). Pool PL-CGK terendah (0,97 trip per shift), pool CBD-heavy tertinggi (1,39–1,40).
12. **Shift mulai sore lebih rendah:** 1,01 trip per shift vs 1,31 untuk shift mulai pagi. Status kerja dan tipe kendaraan tidak berbeda.
13. **Korporat (Q5):** 18,0% trip (56.082) punya perusahaan, seluruhnya dibayar eVoucher Corporate. Trip Airport 29,9% korporat vs sekitar 16% di zona lain.
14. **eVoucher:** trip eVoucher Airport 51,4% di akhir pekan vs 17,6% trip Airport lain, seragam di 200 perusahaan. Skenario CBD-only (bukan aturan di brief): 66,4% trip eVoucher dijemput di luar CBD, hampir sama dengan pelanggan lain (61,7%). Tarif eVoucher tidak lebih sering tinggi (3,43% vs 3,55%) dan tidak ada perusahaan yang menonjol. Zona tujuan tidak tersedia, sehingga hanya zona penjemputan yang diukur.

**Hipotesis (perlu validasi)**

- **H1:** lonjakan 17–18 berasal dari celah pergantian shift (didukung 2.5.11); lonjakan 07–08 berasal dari proses dispatch atau ketersediaan pengemudi di sistem, bukan kekurangan pengemudi (perlu log dispatch).
- **H2:** pengemudi dialokasikan ke zona yang kurang tepat (terlalu banyak di Airport, terlalu sedikit di CBD saat jam sibuk).
- **H3:** target 6 trip per shift mungkin bukan target operasional yang sebenarnya.

**Keterbatasan:** request hanya mencakup App dan Call Center (demand Street Hail tidak terlihat), dan request gagal tidak membawa informasi pengemudi atau perusahaan.

### <span style="color:#0061AF">2.8.2 Inventaris Temuan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Semua temuan penting dari Part 1–2 dikumpulkan di sini sebelum memilih problem statement. <strong>Kekuatan bukti:</strong> Kuat = terukur langsung dan tahan uji ketahanan; Sedang = terukur tetapi bergantung pada asumsi; Lemah = indikasi. Temuan "nol" (tidak ada efek) ikut dicatat karena mencegah kesimpulan yang salah.
</p>

|---|---|---|---|---|---|

# <span style="color:#0061AF">PART 3 — Data Visualization</span>

## <span style="color:#0061AF">3.1 Strategi Visualisasi</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Eksplorasi sudah mengidentifikasi pola yang layak dikomunikasikan. Grafik dipilih berdasarkan pertanyaan dan jenis variabel: batang untuk perbandingan kategori, histogram untuk distribusi, garis untuk tren waktu, dan heatmap untuk dua dimensi sekaligus. Warna: <strong>oranye</strong> = request gagal (masalah), <strong>biru</strong> = seri utama, <strong>abu-abu</strong> = pembanding.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Setiap grafik menjawab satu pertanyaan bisnis dan memakai angka dari Part 1–2. Grafik juga disimpan sebagai PNG di <code>outputs/charts/</code>.
</p>

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# One place for colours and style, so all charts look the same.
COLOR_MAIN = "#2a78d6"      # blue  : normal / main series
COLOR_PROBLEM = "#eb6834"   # orange: failures, the issue to look at
COLOR_NEUTRAL = "#b9b8b2"   # grey  : reference / other
COLOR_TEXT = "#333333"

plt.rcParams.update({
    "figure.dpi": 110,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.edgecolor": "#999999",
    "axes.titlesize": 13,
    "axes.titleweight": "bold",
    "axes.labelsize": 11,
    "axes.labelcolor": COLOR_TEXT,
    "xtick.color": COLOR_TEXT,
    "ytick.color": COLOR_TEXT,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": "#e6e6e6",
})

charts_dir = output_dir / "charts"
charts_dir.mkdir(parents=True, exist_ok=True)

def save_chart(fig, name):
    """Save the figure as PNG (for the presentation) and show it in the notebook."""
    fig.savefig(charts_dir / f"{name}.png", dpi=150, bbox_inches="tight")
    plt.show()
    print(f"Saved: {charts_dir / (name + '.png')}")

hour_labels = [f"{h:02d}" for h in range(24)]
```

## <span style="color:#0061AF">3.2 Pola Permintaan dan Layanan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Grafik pertama menunjukkan pola permintaan dan cara pelanggan memesan serta membayar, sebagai latar sebelum membahas kegagalan.
</p>

### <span style="color:#0061AF">3.2.1 Apakah Street Hail Terjadi pada Jam Berbeda?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Garis menunjukkan porsi trip setiap channel pada tiap jam.
</p>

```python
fig, ax = plt.subplots(figsize=(11, 4.5))
line_styles = {"App": (COLOR_MAIN, "-"), "Call Center": ("#52514e", "--"), "Street Hail": ("#1baf7a", ":")}
for channel, (color, style) in line_styles.items():
    ax.plot(channel_hour_share.index, channel_hour_share[channel], color=color, linestyle=style, linewidth=2.2, label=channel)
ax.set_xticks(range(0, 24, 2))
ax.set_xticklabels(hour_labels[::2])
ax.set_xlabel("Hour the meter started")
ax.set_ylabel("Share of the channel's trips (%)")
ax.set_title("Street Hail follows the same hourly pattern as App and Call Center trips")
ax.legend(frameon=False)
fig.tight_layout()
save_chart(fig, "v13_channel_hour_profile")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> ketiga garis hampir berhimpit dan semua channel memuncak pukul 13–15. Street Hail bukan permintaan dengan pola jam yang berbeda.
</p>

### <span style="color:#0061AF">3.2.2 Cara Bayar per Channel</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Setiap batang adalah 100% trip pada channel tersebut.
</p>

```python
payment_order = ["MyBluebird App", "Credit Card", "Cash", "eVoucher Corporate"]       # same order in every bar
payment_colors = {"MyBluebird App": "#2a78d6", "Credit Card": "#1baf7a", "Cash": "#eda100", "eVoucher Corporate": "#6250d6"}
label_colors = {"MyBluebird App": "white", "Credit Card": COLOR_TEXT, "Cash": COLOR_TEXT, "eVoucher Corporate": "white"}
channel_order = ["Street Hail", "Call Center", "App"]

fig, ax = plt.subplots(figsize=(11, 4.2))
left = np.zeros(len(channel_order))
for payment in payment_order:
    values = channel_payment_mix.loc[channel_order, payment].to_numpy()
    ax.barh(channel_order, values, left=left, color=payment_colors[payment], edgecolor="white", linewidth=1.5, label=payment)
    for i, v in enumerate(values):          # write the % inside each segment that is wide enough
        if v >= 4:
            ax.text(left[i] + v / 2, i, f"{v:.0f}%", ha="center", va="center", fontsize=10, color=label_colors[payment])
    left += values
ax.set_xlim(0, 100)
ax.set_xlabel("Share of the channel's trips (%)")
ax.set_ylabel("Booking channel")
ax.set_title("Street Hail is mostly paid in cash (79%); the app wallet is used mainly for App bookings (57%)")
ax.grid(False)
ax.legend(ncol=4, frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.2))
fig.tight_layout()
save_chart(fig, "v12_channel_payment_mix")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> Street Hail 79% tunai. Call Center paling bervariasi (40% tunai, 25% eVoucher, 17,5% MyBluebird App). eVoucher dipakai di ketiga channel (17% App, 25% Call Center, 11% Street Hail), dengan porsi tertinggi di Call Center.
</p>

### <span style="color:#0061AF">3.3.1 Apakah Data Meter yang Buruk Berpengaruh?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Rata-rata jarak dibandingkan dengan dan tanpa flag untuk melihat seberapa jauh jarak terlalu tinggi jika data tidak dibersihkan.

```python
quality_zone = distance_by_zone.loc[["Airport", "CBD", "Mall", "Residential", "Other"]]

fig, ax = plt.subplots(figsize=(10, 5))
x_positions = np.arange(len(quality_zone))
ax.bar(x_positions - 0.2, quality_zone["distance_km"], width=0.38, color=COLOR_NEUTRAL, label="Raw distance_km (with errors)")
ax.bar(x_positions + 0.2, quality_zone["distance_km_clean"], width=0.38, color=COLOR_MAIN, label="distance_km_clean (errors removed)")
ax.set_xticks(x_positions)
ax.set_xticklabels(quality_zone.index)
ax.set_xlabel("Pickup zone")
ax.set_ylabel("Average distance per trip (km)")
ax.set_title(f"Unflagged errors would overstate average distance by about {abs(quality_impact.loc['Average distance, all trips (km)', 'difference_pct']):.0f}%")
ax.legend(frameon=False)
fig.tight_layout()
save_chart(fig, "v10_data_quality_impact")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> tanpa menandai jarak yang tidak andal (lebih dari 500 km atau kurang dari 1 km), rata-rata jarak terlalu tinggi di semua zona, paling besar di CBD (13,4 vs 10,1 km, sekitar 25%) dan paling kecil di Airport (29,5 vs 27,0 km). Analisis jarak, kecepatan, atau tarif per km harus memakai <code>distance_km_clean</code>.

### <span style="color:#0061AF">3.3.2 Zona Penjemputan dan Distribusi Tarif eVoucher</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Grafik kiri membandingkan komposisi zona penjemputan eVoucher dengan pembanding (CBD diberi garis tepi hanya karena dipakai pada skenario 2.6.6; aturan CBD tidak ada di brief). Grafik kanan menampilkan distribusi tarif eVoucher per zona beserta batas tarif tinggi pembanding (Reguler) sebagai garis merah. Skala log dipakai karena tarif sangat miring ke kanan.
</p>

```python
# 3.3.2 - Where do eVoucher trips start, and how are their fares distributed?
fig, (ax_left, ax_right) = plt.subplots(1, 2, figsize=(13, 5))

zone_mix = pd.DataFrame({
    "eVoucher": ev_trips["pickup_zone_type"].value_counts(normalize=True).loc[zones_in_order] * 100,
    "Non-eVoucher trips": other_trips["pickup_zone_type"].value_counts(normalize=True).loc[zones_in_order] * 100,
})
zone_mix.plot(kind="bar", ax=ax_left, color=["#1f77b4", "#bbbbbb"], width=0.75)
ax_left.set_xlabel("Pickup zone")
ax_left.set_ylabel("% of the group's own trips (each group = 100%)")
ax_left.set_title("Pickup zone mix: eVoucher vs other payments\n(CBD outlined: scenario only, not a rule in the brief)")
ax_left.tick_params(axis="x", rotation=0)
for patch in ax_left.patches[zones_in_order.index("CBD")::len(zones_in_order)]:
    patch.set_edgecolor("black")
    patch.set_linewidth(1.5)

fare_by_zone = [ev_trips.loc[ev_trips["pickup_zone_type"] == zone, "total_fare"] / 1000 for zone in zones_in_order]
ax_right.boxplot(fare_by_zone, tick_labels=zones_in_order, showfliers=True, medianprops={"color": "black", "linewidth": 1.5}, flierprops={"marker": ".", "markersize": 2, "alpha": 0.3})
# High-fare fence = Q3 + 1.5 x IQR of the NON-eVoucher trips of the same pickup zone and trip type (see 2.6.7).
# Reguler and Charter trips have very different fares, so each trip type has its own fence.
fence_table = screen.drop_duplicates(["pickup_zone_type", "trip_type"]).set_index(["trip_type", "pickup_zone_type"])["total_fare_fence"] / 1000
zone_positions = range(1, len(zones_in_order) + 1)
ax_right.scatter(zone_positions, fence_table.loc["Reguler"].loc[zones_in_order], marker="_", s=900, color="#d62728",
                 label="High-fare fence, Reguler trips")
ax_right.scatter(zone_positions, fence_table.loc["Charter"].loc[zones_in_order], marker="_", s=900, color="#9467bd",
                 label="High-fare fence, Charter trips")
ax_right.set_yscale("log")
ax_right.set_xlabel("Pickup zone")
ax_right.set_ylabel("Fare per trip (Rp thousand, log scale:\neach gridline is about 10x the one below)")
ax_right.set_title("eVoucher fare distribution by pickup zone\n(box = middle 50% of trips, line in box = median)")
ax_right.legend(loc="upper center", bbox_to_anchor=(0.5, -0.17), ncol=2, fontsize=8, frameon=False)

plt.tight_layout()
plt.savefig(output_dir / "charts" / "v22_evoucher_zone_fare.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Support table for the numbers quoted below (all eVoucher trips; Non-eVoucher = Cash + Credit Card + MyBluebird App).
fare_support = pd.DataFrame({
    "eVoucher: % of trips": ev_trips["pickup_zone_type"].value_counts(normalize=True) * 100,
    "Non-eVoucher: % of trips": other_trips["pickup_zone_type"].value_counts(normalize=True) * 100,
    "eVoucher: % of eVoucher fare": ev_trips.groupby("pickup_zone_type")["total_fare"].sum() / ev_trips["total_fare"].sum() * 100,
    "eVoucher: median fare (Rp)": ev_trips.groupby("pickup_zone_type")["total_fare"].median(),
    "Non-eVoucher: median fare (Rp)": other_trips.groupby("pickup_zone_type")["total_fare"].median(),
    "fence Reguler (Rp)": screen[screen["trip_type"] == "Reguler"].drop_duplicates("pickup_zone_type").set_index("pickup_zone_type")["total_fare_fence"],
    "fence Charter (Rp)": screen[screen["trip_type"] == "Charter"].drop_duplicates("pickup_zone_type").set_index("pickup_zone_type")["total_fare_fence"],
    "eVoucher: high-fare trips": ev_trips.groupby("pickup_zone_type")["flag_high_fare"].sum(),
}).loc[zones_in_order]
display(fare_support.round(1))
print("Non-eVoucher group:", other_trips["payment_method"].value_counts().to_dict())
print("eVoucher trips by trip type:", ev_trips["trip_type"].value_counts().to_dict())

high_fare_ev = ev_trips[ev_trips["flag_high_fare"]]
print(f"High-fare eVoucher trips: {len(high_fare_ev):,} ({len(high_fare_ev) / len(ev_trips) * 100:.1f}% of eVoucher trips)")
print(f"  of which ALSO high fare per km: {int(high_fare_ev['flag_high_fare_per_km'].sum()):,}; "
      f"high fare but NOT high per km (fare follows distance): {int((~high_fare_ev['flag_high_fare_per_km']).sum()):,} "
      f"= {(~high_fare_ev['flag_high_fare_per_km']).mean() * 100:.1f}%")
print(f"  median distance: high-fare {high_fare_ev['distance_km_clean'].median():.1f} km vs other eVoucher trips "
      f"{ev_trips.loc[~ev_trips['flag_high_fare'], 'distance_km_clean'].median():.1f} km")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Cara membaca angka.</strong> <em>Non-eVoucher</em> (di grafik dan tabel sebelumnya disebut pembanding Non-B2B) = semua 254.925 trip yang dibayar Cash, Credit Card, atau MyBluebird App; di data ini trip ber-<code>company_id</code> persis sama dengan trip eVoucher. Persentase zona di grafik kiri dihitung di dalam masing-masing kelompok (tiap kelompok = 100%). Median tarif = nilai tengah tarif trip pada zona itu (separuh trip lebih murah, separuh lebih mahal): eVoucher Airport Rp128.500, sedangkan zona lain Rp52.500 (Residential) sampai Rp63.000 (Other), dengan CBD Rp53.000 dan Mall Rp58.500. Airport menyumbang 36,3% nilai tarif eVoucher (Rp1,83 miliar dari Rp5,04 miliar) dari 22,0% trip eVoucher (12.343 dari 56.082). Garis merah dan ungu adalah <em>batas tarif tinggi</em>: kuartil ketiga + 1,5 × rentang interkuartil (IQR) dari trip non-eVoucher pada zona dan tipe trip yang sama; Reguler (merah) dan Charter (ungu) punya batas sendiri karena tarif Charter jauh lebih tinggi (sekitar Rp2,1 juta). Skala log dipakai karena tarif sangat miring: tiap garis grid bernilai sekitar 10 kali garis di bawahnya.
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> eVoucher lebih banyak dijemput di Airport (22,0% vs 11,4%) dan lebih sedikit di CBD (33,6% vs 38,3%); zona lain mirip. Median tarif Airport hampir sama pada trip non-eVoucher (Rp129.500 vs Rp128.500), jadi tarif tinggi adalah ciri zona Airport dan bukan ciri eVoucher; itu sebabnya Airport menyumbang porsi tarif lebih besar daripada porsi trip. Hanya 3,4% trip eVoucher (1.922) melewati batas tarif tinggi, hampir sama dengan 3,5% pada non-eVoucher. Dari 1.922 trip itu, sekitar 91% tidak tinggi per kilometer (tarif sejalan dengan jarak; median jarak 20,4 km vs 9,1 km pada trip lain). Trip di atas batas adalah <strong>kandidat tinjauan</strong>, bukan bukti pelanggaran; titik Charter di atas garis merah tidak dihitung karena batas Charter adalah garis ungu. <strong>Batasan:</strong> non-eVoucher adalah pembanding yang tidak sempurna (pelanggan berbeda, bukan perjalanan dinas), dan tidak ada zona tujuan, sehingga grafik hanya menggambarkan titik jemput.
</p>


## <span style="color:#0061AF">3.4 Request dan Kinerja Operasional</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">

### <span style="color:#0061AF">3.4.1 Kapan Request Gagal?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Panel atas menunjukkan jumlah request per jam (penyebut), panel bawah menunjukkan proporsi gagal per jam. Tujuannya melihat apakah kegagalan hanya karena jam tersibuk.
</p>

```python
fig, (ax_top, ax_bottom) = plt.subplots(2, 1, figsize=(11, 7), sharex=True, gridspec_kw={"height_ratios": [1, 1.2]})

hours = by_hour.index
is_peak = hours.isin(failure_peak_hours)

ax_top.bar(hours, by_hour["requests"], color=COLOR_MAIN)
ax_top.set_ylabel("Requests (count)")
ax_top.set_title("Demand peaks at midday, but failures peak at 07-08h and 17-18h")
ax_top.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:,.0f}"))

ax_bottom.bar(hours, by_hour["failed_pct"], color=[COLOR_PROBLEM if p else COLOR_NEUTRAL for p in is_peak])
ax_bottom.axhline(overall_failed_pct, color=COLOR_TEXT, linestyle="--", linewidth=1)
ax_bottom.text(23.4, overall_failed_pct + 0.6, f"Overall {overall_failed_pct:.1f}%", ha="right", fontsize=9, color=COLOR_TEXT)
for h in failure_peak_hours:
    ax_bottom.text(h, by_hour.loc[h, "failed_pct"] + 0.5, f"{by_hour.loc[h, 'failed_pct']:.0f}%", ha="center", fontsize=9)
ax_bottom.set_ylabel("Failed requests (% of requests in the hour)")
ax_bottom.set_xlabel("Hour of day the request was made (00 = midnight)")
ax_bottom.set_xticks(range(24))
ax_bottom.set_xticklabels(hour_labels)
ax_bottom.legend(handles=[Patch(color=COLOR_PROBLEM, label="Failure-peak hours (rate > 2x overall)"),
                          Patch(color=COLOR_NEUTRAL, label="Other hours")], loc="upper left", frameon=False)
fig.tight_layout()
save_chart(fig, "v1_failure_by_hour")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> jam tersibuk (12–16) hanya gagal sekitar 4%, sedangkan pukul 07, 08, 17, dan 18 melonjak ke 22%, 22%, 18%, dan 22%. Empat jam ini hanya 14,3% request tetapi 45,6% kegagalan. Kegagalan terkait jam tertentu, bukan sekadar banyaknya request. Pukul 18 hanya punya 1.258 request sehingga angka 22% bertumpu pada basis kecil.
</p>

### <span style="color:#0061AF">3.4.2 Heatmap Zona × Jam</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Heatmap dipilih karena membandingkan dua dimensi sekaligus. Setiap sel adalah request gagal dibagi seluruh request pada zona dan jam tersebut.
</p>

```python
heat = req.pivot_table(index="pickup_zone_type", columns="hour", values="is_failed", aggfunc="mean") * 100
heat = heat.loc[heat.mean(axis=1).sort_values(ascending=False).index]      # worst zone on top

fig, ax = plt.subplots(figsize=(12, 3.8))
image = ax.imshow(heat.values, aspect="auto", cmap="Oranges", vmin=0)
ax.set_xticks(range(24))
ax.set_xticklabels(hour_labels)
ax.set_yticks(range(len(heat.index)))
ax.set_yticklabels(heat.index)
ax.set_xlabel("Hour of day the request was made")
ax.set_ylabel("Pickup zone")
ax.set_title("Failure rate by zone and hour: the CBD at 07-08h and 17-18h stands out")
ax.grid(False)
colorbar = fig.colorbar(image, ax=ax)
colorbar.set_label("Failed requests (% of requests)")
# Write the number only in the very high cells, so the chart stays readable.
for row in range(heat.shape[0]):
    for col in range(heat.shape[1]):
        if heat.iat[row, col] >= 20:
            ax.text(col, row, f"{heat.iat[row, col]:.0f}", ha="center", va="center", fontsize=8, color="white")
fig.tight_layout()
save_chart(fig, "v2_heatmap_zone_hour")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> hanya CBD yang punya pita gelap: 36–42% request gagal pada pukul 07, 08, 17, dan 18. Mall, Residential, dan Other naik hanya ke sekitar 7–8%, dan Airport tetap sekitar 4% atau kurang. Masalahnya ada pada CBD saat jam sibuk, bukan pada Airport. CBD pukul 18 (42%) hanya bertumpu pada 554 request.
</p>

### <span style="color:#0061AF">3.4.3 CBD, Jam, atau Keduanya?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Tingkat kegagalan dibandingkan antar zona dan periode. Jam puncak kegagalan = 07, 08, 17, 18; jam puncak permintaan = 12–16.
</p>

```python
period_order = ["Demand-peak hours", "Failure-peak hours", "Other hours"]
period_colors = {"Demand-peak hours": COLOR_MAIN, "Failure-peak hours": COLOR_PROBLEM, "Other hours": COLOR_NEUTRAL}
zone_order = list(zone_by_period.mean(axis=1).sort_values(ascending=False).index)

fig, ax = plt.subplots(figsize=(11, 5))
bar_width = 0.26
x_positions = np.arange(len(zone_order))
for i, period in enumerate(period_order):
    values = zone_by_period.loc[zone_order, period]
    ax.bar(x_positions + (i - 1) * bar_width, values, width=bar_width - 0.02, color=period_colors[period], label=period)
cbd_x = zone_order.index("CBD") + 0.0 * bar_width
ax.text(cbd_x, zone_by_period.loc["CBD", "Failure-peak hours"] + 0.8,
        f"{zone_by_period.loc['CBD', 'Failure-peak hours']:.1f}%", ha="center", fontsize=10, fontweight="bold")
ax.set_xticks(x_positions)
ax.set_xticklabels(zone_order)
ax.set_xlabel("Pickup zone")
ax.set_ylabel("Failed requests (% of requests)")
ax.set_title("Failures are a CBD + rush-hour problem, not a busy-hour problem")
ax.legend(frameon=False)
fig.tight_layout()
save_chart(fig, "v3_zone_by_period")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> request CBD pada jam puncak kegagalan gagal 36,6%, dibanding 5,1% pada jam puncak permintaan dan 6,1% pada jam lain. Request ini 7,1% dari total tetapi memuat 38,5% kegagalan (odds ratio sekitar 12). Jika manajemen ingin menurunkan kegagalan, CBD pada empat jam tersebut adalah titik terbesar.
</p>

### <span style="color:#0061AF">3.4.4 Ketersediaan Pengemudi per Jam</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Garis jumlah pengemudi bertugas disandingkan dengan jam kegagalan untuk melihat apakah pasokan sejalan dengan kegagalan (rata-rata hari penuh).

```python
fig, (ax_top, ax_bottom) = plt.subplots(2, 1, figsize=(11, 7), sharex=True)

ax_top.plot(supply_demand.index, supply_demand["drivers_on_shift"], color=COLOR_MAIN, linewidth=2, marker="o", markersize=4)
ax_top.set_ylim(top=supply_demand["drivers_on_shift"].max() * 1.12)
ax_top.set_ylabel("Drivers with an active shift\n(average per day)")
ax_top.set_title("Drivers on shift fall sharply after 15h, yet 07-08h also fails although supply is rising")
ax_top.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:,.0f}"))

ax_bottom.plot(supply_demand.index, supply_demand["failed_pct"], color=COLOR_PROBLEM, linewidth=2, marker="o", markersize=4)
ax_bottom.set_ylabel("Failed requests (% of requests\nmade in that hour)")
ax_bottom.set_xlabel("Hour of day (hour start: 17 = 17:00-17:59)")
ax_bottom.set_xticks(range(24))
ax_bottom.set_xticklabels(hour_labels)

# Direct labels only for the hours quoted in the interpretation (07, 08, 17, 18).
for h in failure_peak_hours:
    ax_top.annotate(f"{supply_demand.loc[h, 'drivers_on_shift']:,.0f}", (h, supply_demand.loc[h, "drivers_on_shift"]),
                    textcoords="offset points", xytext=(0, 8), ha="center", fontsize=9)
    ax_bottom.annotate(f"{supply_demand.loc[h, 'failed_pct']:.0f}%", (h, supply_demand.loc[h, "failed_pct"]),
                       textcoords="offset points", xytext=(0, 8), ha="center", fontsize=9)

for axis in (ax_top, ax_bottom):          # grey bands mark the four failure-peak hours on both panels
    for h in failure_peak_hours:
        axis.axvspan(h - 0.5, h + 0.5, color=COLOR_NEUTRAL, alpha=0.35, linewidth=0)
ax_top.legend(handles=[Patch(color=COLOR_NEUTRAL, alpha=0.35, label="Failure-peak hours (07-08h, 17-18h)")],
              loc="upper right", frameon=False)
fig.tight_layout()
save_chart(fig, "v4_supply_vs_failure")
```

```python
# Support table: where the numbers quoted below come from (hours 07, 08, 17, 18).
# Both panels use the 90 full days (2023-10-01 to 2023-12-29). Hour = hour the request was made / hour the shift was active.
key_hours = [7, 8, 17, 18]
requests_by_hour = full_days_req.groupby("hour").size()
failed_by_hour = full_days_req.groupby("hour")["is_failed"].sum()      # failed = Cancelled-No Driver + Cancelled-Timeout

support_table = pd.DataFrame({
    "requests (90 days)": requests_by_hour,
    "failed requests (90 days)": failed_by_hour,
    "failed % = failed / requests": failed_by_hour / requests_by_hour * 100,
    "drivers with an active shift (avg per day)": supply_demand["drivers_on_shift"],
    "drivers, time-weighted (avg per day, 2.5.10)": utilisation_by_hour["drivers_on_shift"],
    "CBD requests per CBD driver (2.5.1)": cbd_supply_demand["cbd_requests_per_driver"],
}).loc[key_hours]
assert np.allclose(support_table["failed % = failed / requests"], supply_demand.loc[key_hours, "failed_pct"])   # same as the chart
display(support_table.round(2))
print("Average drivers with an active shift, 08-12h:", round(supply_demand.loc[8:12, "drivers_on_shift"].mean()))
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Cara membaca angka.</strong> Panel atas: jumlah shift yang aktif (sebagian atau seluruhnya) pada jam itu, dirata-rata atas 90 hari penuh. Panel bawah: request gagal (Cancelled-No Driver + Cancelled-Timeout) dibagi seluruh request yang dibuat pada jam itu, dijumlahkan atas 90 hari. Contoh: pukul 17 ada 1.627 request gagal dari 8.917 request = 18,2%. Pukul 07 dan 08 gagal 22,0% dan 22,4%; pukul 18 gagal 22,1% tetapi hanya dari 1.258 request. Label jam adalah awal jam (17 = 17:00–17:59).
</p>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> jumlah shift aktif turun dari sekitar 2.300 (08–12) menjadi 351 pada 17 dan 163 pada 18, karena sebagian besar shift pagi berakhir antara pukul 13 dan 17 (puncak clock-out 14–16) dan shift sore baru mulai pukul 18–22. Pada 07–08 jumlah shift aktif sedang naik (1.794 ke 2.322) tetapi sekitar 22% request tetap gagal, sehingga alasan "kekurangan pengemudi" tidak cocok untuk pagi hari. Pada 17, CBD punya request per pengemudi tertinggi dari 24 jam: 39,6 request per jam dibagi 109,5 shift berzona dominan CBD yang aktif = 0,36 (rata-rata semua zona 0,28), yang sesuai dengan celah pergantian shift. <strong>Catatan:</strong> angka 351 menghitung shift yang hanya aktif sebagian jam 17; jika dihitung menurut menit tugas, rata-rata pengemudi pada 17 hanya sekitar 183 (grafik 3.4.6). Korelasi antar-jam CBD antara request per pengemudi dan tingkat kegagalan hanya 0,21, dan pada 07–08 CBD gagal 36,6% dengan rasio hanya 0,14. Jadi pasokan paling masuk akal menjelaskan puncak sore, bukan pagi. Ini <strong>hipotesis</strong>: data tidak menunjukkan di mana setiap pengemudi berada saat request masuk, sehingga jumlah agregat tidak membuktikan kekurangan lokal.
</p>

### <span style="color:#0061AF">3.4.6 Pengemudi Bertugas vs Sibuk, dan Utilisasi vs Kegagalan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Area pengemudi bertugas dibandingkan dengan pengemudi sibuk per jam, dengan garis kegagalan, untuk melihat apakah request gagal karena semua pengemudi sibuk.

```python
fig, (ax_top, ax_bottom) = plt.subplots(2, 1, figsize=(11, 8), sharex=True)
hours_axis = utilisation_by_hour.index

ax_top.fill_between(hours_axis, utilisation_by_hour["drivers_on_shift"], color=COLOR_NEUTRAL, label="Drivers on shift")
ax_top.fill_between(hours_axis, utilisation_by_hour["drivers_busy"], color=COLOR_MAIN, label="Drivers busy with a trip")
ax_top.set_ylabel("Average number of drivers\n(time-weighted)")
ax_top.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:,.0f}"))
ax_top.set_title("The blue area (busy) is a thin slice of the grey area (on shift) in every hour")
ax_top.legend(frameon=False, loc="upper right")

ax_bottom.plot(hours_axis, utilisation_by_hour["utilisation_pct"], color=COLOR_MAIN, linewidth=2.2, marker="o", markersize=4,
               label="Driver utilisation (% of on-shift time spent on trips)")
ax_bottom.plot(hours_axis, utilisation_by_hour["failed_pct"], color=COLOR_PROBLEM, linewidth=2.2, marker="o", markersize=4,
               label="Failed requests (% of requests)")
for h in failure_peak_hours:
    ax_bottom.axvspan(h - 0.5, h + 0.5, color=COLOR_NEUTRAL, alpha=0.3, linewidth=0)
ax_bottom.annotate(f"07-08h: drivers are on shift but only about {utilisation_by_hour.loc[7:8, 'utilisation_pct'].mean():.0f}% busy,\nyet {utilisation_by_hour.loc[7:8, 'failed_pct'].mean():.0f}% of requests fail",
                   xy=(8, utilisation_by_hour.loc[8, "failed_pct"]), xytext=(9.3, 30), fontsize=9,
                   arrowprops={"arrowstyle": "->", "color": COLOR_TEXT})
ax_bottom.annotate(f"17h: few drivers left, {utilisation_by_hour.loc[17, 'utilisation_pct']:.0f}% busy,\n{utilisation_by_hour.loc[17, 'failed_pct']:.0f}% of requests fail",
                   xy=(17, utilisation_by_hour.loc[17, "utilisation_pct"]), xytext=(18.3, 38), fontsize=9,
                   arrowprops={"arrowstyle": "->", "color": COLOR_TEXT})
ax_bottom.set_ylabel("Percent")
ax_bottom.set_xlabel("Hour of day (grey bands = failure-peak hours)")
ax_bottom.set_xticks(range(24))
ax_bottom.set_xticklabels(hour_labels)
ax_bottom.legend(frameon=False, loc="upper left")
fig.tight_layout()
save_chart(fig, "v17_capacity_vs_demand")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">

### <span style="color:#0061AF">3.5.1 Ritase per Pengemudi per Hari</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Histogram dipilih untuk distribusi jumlah trip per shift. Unit analisis: satu shift (= satu driver-day, lihat 2.5.4). Target 6 trip adalah asumsi.
</p>

```python
trip_count_share = shifts_eda["trip_count"].value_counts(normalize=True).sort_index() * 100

fig, ax = plt.subplots(figsize=(10, 5))
bar_colors = [COLOR_PROBLEM if count == 0 else (COLOR_MAIN if count < 6 else COLOR_NEUTRAL) for count in trip_count_share.index]
ax.bar(trip_count_share.index, trip_count_share.values, color=bar_colors)
ax.axvline(5.5, color=COLOR_TEXT, linestyle="--", linewidth=1)
ax.text(5.6, trip_count_share.max() * 0.85, f"Target: 6+ trips\n= {shifts_eda['ritase_target_met'].mean()*100:.2f}% of shifts",
        fontsize=10, va="top")
ax.text(0, trip_count_share.loc[0] + 0.8, f"{trip_count_share.loc[0]:.1f}%", ha="center", fontsize=10)
ax.legend(handles=[Patch(color=COLOR_PROBLEM, label="0 trips"),
                   Patch(color=COLOR_MAIN, label="1-5 trips (below target)"),
                   Patch(color=COLOR_NEUTRAL, label="6+ trips (target met)")], frameon=False, loc="upper right")
ax.set_xticks(trip_count_share.index)
ax.set_xlabel("Trips completed in the shift (trip_count)")
ax.set_ylabel("Share of all shifts (%)")
ax.set_title(f"Average ritase is {shifts_eda['trip_count'].mean():.2f} trips per shift; "
             f"{shifts_eda['zero_trip_shift'].mean()*100:.0f}% of shifts have no trip at all")
fig.tight_layout()
save_chart(fig, "v5_trips_per_shift")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">

### <span style="color:#0061AF">3.5.5 Seberapa Besar Variasi yang Dijelaskan Tiap Faktor?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Faktor kategorik memakai eta-squared dan faktor kontinu memakai r-squared. Setiap faktor dihitung <strong>sendiri-sendiri</strong>.
</p>

```python
def eta_squared(column):
    groups = shifts_eda.groupby(column)["trip_count"]
    between = (groups.count() * (groups.mean() - shifts_eda["trip_count"].mean()) ** 2).sum()
    return between / ((shifts_eda["trip_count"] - shifts_eda["trip_count"].mean()) ** 2).sum() * 100

def r_squared(column):
    valid = shifts_eda[column].notna()
    return np.corrcoef(shifts_eda.loc[valid, column], shifts_eda.loc[valid, "trip_count"])[0, 1] ** 2 * 100

explained = pd.Series({
    "Dominant zone (place)": eta_squared("dominant_zone_worked"),
    "Pool (garage)": eta_squared("pool_id"),
    "Shift start (morning / evening)": eta_squared("start_group"),
    "Requests per driver in the zone (crowding)": r_squared("requests_per_driver_zone"),
    "Employment status": eta_squared("employment_status"),
    "Vehicle type": eta_squared("vehicle_type_assigned"),
    "Weekday": eta_squared("weekday"),
    "Driver rating": r_squared("driver_rating"),
    "Tenure (days)": r_squared("tenure_days"),
}).sort_values()
display(explained.round(3).to_frame("variance_explained_pct"))
if have_statsmodels:
    explained = pd.concat([explained, pd.Series({"ALL factors together (OLS R-squared)": model_with_pool.rsquared * 100})])

fig, ax = plt.subplots(figsize=(10, 5.2))
bar_colors = ["#52514e" if name.startswith("ALL") else (COLOR_MAIN if value >= 0.5 else COLOR_NEUTRAL)
              for name, value in explained.items()]
ax.barh(explained.index, explained.values, color=bar_colors)
for position, value in enumerate(explained.values):
    ax.text(value + 0.15, position, f"{value:.2f}%", va="center", fontsize=9)
ax.set_xlabel("Share of the variation in trips per shift explained by the factor alone (%)")
ax.set_title("The place (zone) explains most; crowding mostly mirrors the zone; the rest explains almost nothing")
ax.set_xlim(0, 13)
ax.grid(axis="y", visible=False)
fig.tight_layout()
save_chart(fig, "v19_variance_explained")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> zona menjelaskan 9,7% variasi trip per shift, faktor tunggal terbesar. "Request per pengemudi di zona" menjelaskan 7,4% tetapi sebagian besar hanya mengulang zona. Pool (1,4%) dan waktu mulai (1,0%) kecil, sedangkan status kerja, kendaraan, hari, rating, dan masa kerja sekitar 0%. Jika manajemen ingin mengubah idle time, tuasnya adalah <strong>di mana pengemudi bekerja</strong>, bukan siapa pengemudinya atau pool mana. Seluruh faktor bersama 10,4%, hampir sama dengan batas maksimum sekitar 10,5% untuk data hitung seperti ini (2.7.3); sisa variasi adalah keacakan jumlah trip per shift, bukan faktor yang terlewat.
</p>

```python
fig, (ax_left, ax_right) = plt.subplots(1, 2, figsize=(12.5, 4.8), gridspec_kw={"width_ratios": [1.2, 1]})

zones_in_order = ["Airport", "CBD", "Mall", "Residential", "Other"]
unusual_matrix = zone_rule["unusual_time_pct"].unstack("customer_segment").loc[zones_in_order]
positions = np.arange(len(zones_in_order))
ax_left.bar(positions - 0.2, unusual_matrix["B2B"], width=0.38, color=COLOR_MAIN, label="B2B (eVoucher)")
ax_left.bar(positions + 0.2, unusual_matrix["Non-B2B"], width=0.38, color="#52514e", label="Non-B2B")
for i, zone in enumerate(zones_in_order):
    ax_left.text(i - 0.2, unusual_matrix.loc[zone, "B2B"] + 1, f"{unusual_matrix.loc[zone, 'B2B']:.0f}%", ha="center", fontsize=9)
    ax_left.text(i + 0.2, unusual_matrix.loc[zone, "Non-B2B"] + 1, f"{unusual_matrix.loc[zone, 'Non-B2B']:.0f}%", ha="center", fontsize=9)
ax_left.set_xticks(positions)
ax_left.set_xticklabels(zones_in_order)
ax_left.set_xlabel("Pickup zone")
ax_left.set_ylabel("Trips at night (22-04h) or weekend (% of group's trips)")
ax_left.set_ylim(0, 80)
ax_left.set_title("Unusual-time share by zone")
ax_left.legend(frameon=False, loc="upper right")

day_positions = np.arange(7)
ax_right.bar(day_positions - 0.2, airport_weekday["B2B"], width=0.38, color=COLOR_MAIN, label="B2B (eVoucher)")
ax_right.bar(day_positions + 0.2, airport_weekday["Non-B2B"], width=0.38, color="#52514e", label="Non-B2B")
ax_right.set_xticks(day_positions)
ax_right.set_xticklabels(weekday_names)
ax_right.set_xlabel("Weekday of the trip")
ax_right.set_ylabel("Share of the group's Airport trips (%)")
ax_right.set_title("Airport trips by weekday")
ax_right.legend(frameon=False, loc="upper left")
fig.suptitle("The unusual-time pattern is an Airport pattern: 66% of B2B Airport trips are weekend or late night", fontweight="bold", fontsize=12.5)
fig.tight_layout()
save_chart(fig, "v14_evoucher_airport_unusual_time")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> B2B lebih rendah daripada Non-B2B di empat zona (sekitar 15% vs 38%) tetapi jauh lebih tinggi di Airport (66% vs 23%). Trip B2B Airport 25–26% pada setiap hari akhir pekan vs sekitar 9% untuk Non-B2B. Pertanyaan kontrol eVoucher adalah pertanyaan Airport akhir pekan/malam, dan berlaku bagi hampir semua perusahaan.
</p>

### <span style="color:#0061AF">3.5.11 eVoucher di Luar Hari dan Jam Kerja: Perusahaan atau Tempat?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Pertanyaan: apakah pola eVoucher akhir pekan/malam disebabkan beberapa perusahaan atau tempat (Airport)? Metrik: porsi trip eVoucher yang mulai Sabtu/Minggu atau pukul 22:00–04:59 (asumsi kami, tidak ada aturan kontrak pada data). 3.5.11 membandingkan pola per hari dan sebaran porsi per perusahaan.
</p>

```python
# 3.5.11 - Is "unusual-time eVoucher use" a few-companies problem, or a place problem?
MIN_COMPANY_TRIPS = 20     # same cut-off as 2.6.5, so company percentages are not noise

b2b_trips = ev[ev["customer_segment"] == "B2B"].copy()
b2b_trips["group"] = np.where(b2b_trips["pickup_zone_type"] == "Airport", "Airport pickups", "Other zones")

# Left chart: share of trips per weekday (Airport only), reuse the table from 2.6.5
# Right chart: unusual-time share per company, separately for Airport and other zones
company_rate = b2b_trips.groupby(["group", "company_id"]).agg(
    trips=("trip_id", "size"),
    unusual_pct=("is_unusual_time", lambda s: s.mean() * 100),
).reset_index()
company_rate = company_rate[company_rate["trips"] >= MIN_COMPANY_TRIPS]

airport_rates = company_rate.loc[company_rate["group"] == "Airport pickups", "unusual_pct"]
other_rates = company_rate.loc[company_rate["group"] == "Other zones", "unusual_pct"]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

x_pos = np.arange(7)
ax1.bar(x_pos - 0.2, airport_weekday["B2B"], width=0.4, label="B2B (eVoucher)", color="#1f77b4")
ax1.bar(x_pos + 0.2, airport_weekday["Non-B2B"], width=0.4, label="Non-B2B", color="#bbbbbb")
ax1.set_xticks(x_pos)
ax1.set_xticklabels(airport_weekday.index)
ax1.set_ylabel("% of the group's Airport trips")
ax1.set_title("Airport pickups: eVoucher peaks on Sat/Sun,\nother customers peak on weekdays")
ax1.legend()

bins = np.arange(0, 100, 4)
ax2.hist(other_rates, bins=bins, alpha=0.8, color="#2ca02c", label=f"Other zones ({len(other_rates)} companies)")
ax2.hist(airport_rates, bins=bins, alpha=0.8, color="#d62728", label=f"Airport pickups ({len(airport_rates)} companies)")
ax2.set_xlabel("% of the company's eVoucher trips at weekend / night")
ax2.set_ylabel("Number of companies")
ax2.set_title("Every company shifts together:\nthe Airport group is higher, with no obvious outlier company")
ax2.legend()

plt.tight_layout()
plt.savefig(output_dir / "charts" / "v21_evoucher_unusual_time.png", dpi=150, bbox_inches="tight")
plt.show()

print(f"Airport: median company = {airport_rates.median():.1f}%, range {airport_rates.min():.1f}% - {airport_rates.max():.1f}%")
print(f"Other zones: median company = {other_rates.median():.1f}%, range {other_rates.min():.1f}% - {other_rates.max():.1f}%")
print(f"Companies in the top 10% of Airport unusual share hold {b2b_trips[b2b_trips['company_id'].isin(company_rate.loc[(company_rate['group']=='Airport pickups') & (company_rate['unusual_pct'] >= airport_rates.quantile(0.9)), 'company_id'])].shape[0] / len(b2b_trips) * 100:.1f}% of all B2B trips")
```

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Interpretasi:</strong> pola ini bukan milik beberapa perusahaan. Semua perusahaan dengan cukup trip Airport berada pada rentang sempit (48–85%, median 67%), pola sama di semua industri (65–67%), dan perusahaan yang sama terlihat normal di luar Airport (median 15%). Polanya mengikuti <strong>tempat</strong>, bukan perusahaan. Mencari satu perusahaan yang "melanggar" bernilai rendah, dan "pelanggaran" tidak bisa dibuktikan karena tidak ada aturan kontrak tentang hari/jam yang diizinkan. Penjelasan yang mungkin (hipotesis): perjalanan bisnis mengikuti jadwal penerbangan, pemakaian pribadi, atau voucher tanpa batasan waktu.
</p>

### <span style="color:#0061AF">3.6.2 Mengapa Pengemudi Idle di Sebuah Pool?</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Diurutkan menurut kekuatan bukti (teramati = ditunjukkan data; hipotesis = kemungkinan penyebab, belum terbukti).
</p>

| # | Alasan | Bukti | Uji atau metode | Status |
|---|---|---|---|---|
| 1 | **Kapasitas: jam shift jauh melebihi permintaan tercatat** (idle tinggi adalah gejalanya, bukan penjelasannya) | Utilisasi 9,4%; permintaan hanya butuh 18,9–31,5% jam-pengemudi yang ada pada skenario 50–30% sibuk (2.5.10, 2.5.11, 3.4.8) | Ukuran deskriptif (menit sibuk / menit bertugas), skenario kapasitas | Teramati; tingkat utilisasi target = asumsi |
| 2 | **Tempat: Airport penuh pengemudi** | Zona faktor terkuat (9,7% variasi); shift CBD 2,49 kali trip shift Airport; shift tanpa trip 19% (CBD) vs 51% (Airport); Airport 0,61 request per pengemudi vs 1,38 di CBD (2.7.1, 3.5.5, 3.5.6) | Kruskal-Wallis, chi-square, regresi Poisson dan logistik, indeks kepadatan | Teramati (label zona sebagian hasil pekerjaan) |
| 3 | **Waktu: shift mulai sore** | Trip 1,25 kali lebih sedikit daripada mulai pagi; rank-biserial r = 0,14 (2.7.1) | Mann-Whitney U, regresi Poisson | Teramati; sebab (permintaan sore tipis) = hipotesis |
| 4 | **Pagi: masalah pencocokan, bukan kekurangan** | Pukul 07–08 sekitar 5% pengemudi sibuk tetapi 22% request gagal; di CBD 0,14 request per pengemudi per jam tetapi 36,6% gagal (2.5.1, 3.4.6) | Perbandingan deskriptif per jam, validasi Desember | Teramati; sebab (dispatch atau ketersediaan di sistem) = hipotesis |
| 5 | **Pool BUKAN penyebab mandiri** | Setelah zona dikendalikan, F-test pool p = 0,62; rate ratio 1,00–1,01 (2.7.3) | F-test model bertingkat, regresi Poisson | Teramati |
| 6 | **Bukan penyebab:** status kerja, tipe kendaraan, hari, rating, masa kerja, tren mingguan | Semua diabaikan atau tidak signifikan (2.7.4) | Mann-Whitney U, Kruskal-Wallis, Spearman, uji kemiringan | Teramati |
| 7 | **Sisa variasi adalah keacakan** | R-squared 10,4% dari maksimum sekitar 10,5%; varians = rata-rata di setiap zona (2.7.3) | Regresi OLS, cek dispersi Poisson | Teramati: shift dengan kondisi sama tetap berbeda karena kebetulan, bukan karena faktor yang hilang |

**Apakah karena tempat ramai?** Sebagian ya, dalam arti yang tepat: **Airport punya banyak pengemudi untuk sedikit request** dan pengemudi di sana lebih idle. Di dalam satu zona, tambahan permintaan per pengemudi hampir tidak menambah trip setelah sekitar 1 request per pengemudi, sehingga "ramai" adalah cerita <strong>antar zona</strong>, bukan di dalam zona.

### <span style="color:#0061AF">3.6.3 Daftar Uji Statistik dan Metode yang Digunakan</span>

**Uji hipotesis** (setiap uji punya H0 yang menyatakan "tidak ada efek"):

| Uji | Dipakai di | H0 | Hasil |
|---|---|---|---|
| Kruskal-Wallis H | 2.7.1 (zona, pool, kendaraan, hari) | semua kelompok berdistribusi sama | zona dan pool menolak H0; kendaraan dan hari tidak |
| Mann-Whitney U | 2.7.1 (mulai pagi vs sore, Tetap vs Mitra) | kedua kelompok berdistribusi sama | waktu mulai menolak H0 (efek kecil); status kerja tidak |
| Chi-square independensi (+ Cramer's V) | 2.4.4, 2.7.1 (zona × shift tanpa trip) | kedua variabel independen | keduanya menolak H0 |
| Uji z dua proporsi (+ CI 95%, Cohen's h) | 2.7.1 (CBD jam puncak vs lainnya; B2B Airport vs Non-B2B) | kedua proporsi sama | keduanya menolak H0, efek besar; jam puncak dipilih dari data yang sama, sehingga divalidasi dengan data Desember (2.7.3) |
| Uji Wald gabungan dengan standard error per pengemudi (cluster) | 2.7.3 (pool setelah zona, shift tidak independen) | semua efek pool = 0 | tidak menolak H0 (p = 0,78) |
| Korelasi Spearman | 2.7.1 (kepadatan, rating, masa kerja vs trip) | rho = 0 | kepadatan signifikan tetapi diabaikan; rating dan masa kerja tidak menolak H0 |
| F-test model bertingkat (OLS) | 2.7.3 (pool setelah mengendalikan zona dll.) | semua koefisien pool = 0 | tidak menolak H0 (p = 0,62) |
| Uji t kemiringan regresi + Spearman | 2.5.7 / 2.7.1 (tren mingguan ritase) | kemiringan = 0 | tidak menolak H0 (tidak ada tren) |
| Skor z (penyaringan proporsi) | 2.6.2, 2.6.6, 2.6.9 (per perusahaan dan per pengemudi, batas Bonferroni) | porsi kelompok sama dengan keseluruhan | tidak ada perusahaan yang menonjol; 1 dari 4.043 pengemudi (sinyal lemah) |

**Model dan effect size:** regresi OLS (R-squared, partial eta-squared); regresi Poisson dengan robust standard error dan standard error per pengemudi (rate ratio); cek dispersi Poisson dan batas R-squared maksimum (2.7.3); regresi logistik (odds ratio); epsilon-squared, eta-squared, r-squared, rank-biserial, Cramer's V, Cohen's h, odds ratio; CI 95%; koreksi **Bonferroni** untuk 15 uji.

**Metode analisis:** ambang batas dari celah data (batas jarak); cek kardinalitas join; pembersihan dengan flag; rasio dengan pembilang dan penyebut jelas; group-by dan pivot; standarisasi langsung menurut komposisi zona (2.5.5); penghitungan pengemudi bertugas per jam (2.5.1, 2.5.10); utilisasi (2.5.10); indeks kepadatan (2.7.2); ukuran kontrafaktual kegagalan berlebih (2.4.7); validasi di luar sampel (pilih jam dengan Oktober–November, uji di Desember, 2.7.3); skenario kapasitas per jam (2.5.11); korelasi phi antar sinyal tinjauan eVoucher (2.6.7); perbandingan komposisi trip yang dikeluarkan vs disimpan (1.8.5).

**Variabel yang dikeluarkan dari pengujian:** lihat 2.5.9.

Tabel lain tidak dihapus barisnya; record bermasalah hanya diberi kolom flag. <strong>Cara validasi:</strong> fingerprint tabel mentah saat dimuat (1.2.2) dan tabel bersih di akhir Part 1 (1.9.3) dihitung ulang sebelum export untuk membuktikan tidak ada perubahan. <strong>Keterbatasan:</strong> CSV menyimpan semua nilai sebagai teks, sehingga tipe data (datetime, boolean) harus di-parse ulang saat file dibaca; file mentah adalah hasil export dari database, bukan salinan byte asli dari sumber.

## <span style="color:#0061AF">4.1 Export Data Mentah (RAW)</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Keenam tabel diexport dari DataFrame <code>*_raw</code> yang dimuat di 1.2.1. Fingerprint dicek dulu untuk membuktikan tabel ini belum pernah diubah.
</p>


```python
# 4.1 RAW exports: the six tables exactly as loaded from MySQL, before any cleaning.
raw_exports = {
    "bb_corporate_clients": df_clients_raw,
    "bb_trips": df_trips_raw,
    "bb_driver_shifts": df_shifts_raw,
    "bb_drivers": df_drivers_raw,
    "bb_pools": df_pools_raw,
    "bb_service_requests": df_requests_raw,
}
# Proof that the raw DataFrames are still the untouched source data (fingerprints taken in 1.2.2).
for name, df in raw_exports.items():
    assert int(pd.util.hash_pandas_object(df, index=True).sum()) == raw_fingerprints[name], f"{name}: raw data changed after loading"
print("Raw fingerprints match the values taken right after loading: the raw tables were never modified.")

import math

MAX_FILE_MB = 50                              # every exported file must be at most 50 MB
MAX_FILE_BYTES = MAX_FILE_MB * 1000 * 1000    # 50 MB = 50,000,000 bytes
TARGET_PART_BYTES = 49 * 1000 * 1000          # aim at 49 MB per part to keep a safety margin

export_log = []          # one entry per exported dataset, used for the manifest and the validation below

def export_csv(df, folder, file_name, dataset, stage, note):
    """Write one DataFrame to CSV (UTF-8 with BOM so Excel shows it correctly, no index) and log it.
    If the CSV would be larger than MAX_FILE_MB, the rows are split into numbered parts
    (name_part1.csv, name_part2.csv, ...). Every part has the same header, so the parts
    can be stacked back in order with pd.concat."""
    stem = Path(file_name).stem
    # Delete files from an earlier run (the single file or old parts), so no stale or oversized file stays behind.
    for old_file in [folder / file_name, *folder.glob(f"{stem}_part*.csv")]:
        old_file.unlink(missing_ok=True)

    csv_size = len(df.to_csv(index=False).encode("utf-8-sig"))       # size of the whole table as one CSV
    n_parts = max(1, math.ceil(csv_size / TARGET_PART_BYTES))
    row_groups = np.array_split(np.arange(len(df)), n_parts)          # consecutive rows, roughly equal sizes

    paths = []
    for part_number, rows in enumerate(row_groups, start=1):
        path = folder / (file_name if n_parts == 1 else f"{stem}_part{part_number}.csv")
        df.iloc[rows].to_csv(path, index=False, encoding="utf-8-sig")
        size_mb = path.stat().st_size / 1e6
        assert path.stat().st_size <= MAX_FILE_BYTES, f"{path.name} is {size_mb:.1f} MB, above the {MAX_FILE_MB} MB limit"
        paths.append(path.as_posix())
        print(f"{stage:<9} {path.as_posix():<56} {len(rows):>9,} rows x {df.shape[1]:>2} columns {size_mb:>6.1f} MB")
    export_log.append({"dataset": dataset, "stage": stage, "files": paths, "frame": df, "note": note})

for name, df in raw_exports.items():
    export_csv(df, raw_dir, f"{name}_raw.csv", name, "RAW", "As loaded from MySQL; no cleaning applied")
```

Tabel bersih diambil dari <code>tables_final</code> (1.8.2), yaitu tabel yang dipakai di Part 2–3. Fingerprint dicek untuk membuktikan analisis tidak mengubahnya. File exception sudah ditulis di 1.8.1 dan 2.6.7, dan di sini hanya didaftarkan untuk divalidasi.
</p>


```python
# 4.2 CLEANED exports: the final tables from 1.8.2 (tables_final), the same objects used in PART 2-3.
cleaned_names = {
    "bb_corporate_clients": "clients", "bb_trips": "trips", "bb_driver_shifts": "shifts",
    "bb_drivers": "drivers", "bb_pools": "pools", "bb_service_requests": "requests",
}
cleaned_notes = {
    "bb_corporate_clients": "No rows removed; text trimmed and standardised; no columns added",
    "bb_pools": "No rows removed; text trimmed and standardised; no columns added",
    "bb_drivers": "No rows removed; join_date review flags added (flag_join_after_data_end, flag_first_shift_before_join)",
    "bb_trips": "7,091 trips excluded (negative distance or negative booking-to-meter delay); review flags kept as columns",
    "bb_driver_shifts": "No rows removed; shift_duration_minutes, review flags, trips_in_table and flag_trip_count_mismatch added",
    "bb_service_requests": "No rows removed; wait_time_minutes_clean and review flags added; matched_trip_id kept even if the trip was excluded",
}
# Proof that the analysis did not change the cleaned tables after PART 1 (fingerprints taken in 1.9.3).
for name, key in cleaned_names.items():
    assert int(pd.util.hash_pandas_object(tables_final[key], index=True).sum()) == cleaned_fingerprints[key], f"{name}: cleaned table changed after PART 1"
print("Cleaned fingerprints match the end of PART 1: PART 2-3 did not modify the exported tables.")

for name, key in cleaned_names.items():
    export_csv(tables_final[key], cleaned_dir, f"{name}_cleaned.csv", name, "CLEANED",
               cleaned_notes.get(name, "No rows removed; text labels standardised, types converted, flag columns added"))

# EXCEPTION files were written earlier (1.8.1 and 2.6.7). They are registered here so they are validated too.
for name, frame, file_name, note in [
    ("bb_trips", trip_exceptions, "bb_trips_exceptions.csv", "Every trip with a problem flag; column treatment = excluded or kept"),
    ("bb_driver_shifts", shift_exceptions, "bb_driver_shifts_exceptions.csv", "Every shift with a review flag (overlap, before join_date, duration/idle checks); all kept"),
    ("evoucher", review_candidates[review_columns], "evoucher_review_candidates.csv", "Review priority list (2-3 independent signals); NOT a fraud list"),
    ("evoucher_close_pairs", evoucher_close_pairs, "evoucher_close_pairs.csv", "Same company, same pickup zone, trips starting < 2 minutes apart (2.6.10); review list"),
]:
    export_log.append({"dataset": name, "stage": "EXCEPTION", "files": [(exceptions_dir / file_name).as_posix()], "frame": frame, "note": note})
print("Exception files registered:", ", ".join(entry["files"][0] for entry in export_log if entry["stage"] == "EXCEPTION"))
```

Tabel lain punya jumlah baris yang sama di RAW dan CLEANED; record bermasalah tetap disimpan dengan kolom flag (rinciannya di 1.9.2). <strong>Batas 50 MB:</strong> fungsi <code>export_csv</code> memecah tabel yang CSV-nya lebih dari 50 MB menjadi beberapa bagian dan menghentikan notebook jika satu bagian melebihi 50 MB. Ukuran setiap file tercetak di output 4.1 dan 4.2. # <span style="color:#0061AF">PENUTUP</span>

## <span style="color:#0061AF">Problem Statement SMART</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Metrik inti: tingkat kegagalan request CBD di jam puncak</strong> = jumlah request CBD berstatus Cancelled-No Driver + Cancelled-Timeout ÷ jumlah semua request CBD pada pukul 07.00–08.59 dan 17.00–18.59 × 100% (semakin kecil semakin baik).
</p>

| Aspek definisi | Ketentuan yang digunakan |
|---|---|
| Unit analisis | Satu baris = satu request pelanggan (tabel <code>bb_service_requests</code>) |
| Periode | Request 1 Oktober–30 Desember 2023 |
| Zona | <code>pickup_zone_type</code> = CBD |
| Jam puncak | Jam pembuatan request 07, 08, 17, dan 18 (07 = 07.00–07.59), dipilih dari data karena tingkat gagalnya lebih dari 2 kali rata-rata (2.4.2) dan terkonfirmasi di data Desember (2.7.3) |
| Gagal | Status <code>Cancelled-No Driver</code> atau <code>Cancelled-Timeout</code>; request <code>Fulfilled</code> dihitung berhasil |
| Pembanding | Tingkat kegagalan request CBD di jam lain (zona yang sama, di luar 07, 08, 17, 18) |
| Batasan | Hanya request App dan Call Center; pelanggan Street Hail yang gagal mendapat taksi tidak tercatat |

| Huruf | Arti | Isi untuk kasus Blue Bird |
|---|---|---|
| **S – Specific** | Satu metrik, satu objek, cakupan jelas | Tingkat kegagalan request di zona CBD pada pukul 07.00–08.59 dan 17.00–18.59 |
| **M – Measurable** | Ada rumus, data, dan baseline | Dihitung dari <code>bb_service_requests.status</code>; baseline 36,6% (7.470 gagal dari 20.436 request CBD di jam puncak), dibanding 5,5% di jam lain CBD (2.4.4, 2.4.7) |
| **A – Achievable** | Ada tindakan konkret yang bisa dijalankan | Pemeriksaan proses dispatch dan kesiapan pengemudi di pagi hari, serta uji coba penambahan shift atau penundaan clock-out pada pukul 15.00–19.00 di CBD (Rekomendasi 1 dan 2) |
| **R – Relevant** | Terhubung dengan masalah bisnis | Brief menyebut permintaan yang tidak terpenuhi di waktu dan lokasi tertentu; CBD di jam puncak hanya 7,1% dari semua request tetapi menyumbang 38,5% dari seluruh kegagalan (2.4.4) |
| **T – Time-bound** | Ada batas waktu | Perubahan harus terlihat dalam 1 kuartal setelah uji coba dimulai, dibandingkan dengan pool yang tidak ikut uji coba pada periode yang sama |

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Problem statement SMART:</strong>
</p>

> *Bagaimana manajemen operasional armada Blue Bird dapat menurunkan tingkat kegagalan request di CBD pada pukul 07.00–08.59 dan 17.00–18.59 dari baseline 36,6% (Oktober–Desember 2023) mendekati tingkat kegagalan CBD di jam lain (5,5%) dalam 1 kuartal setelah uji coba dimulai, melalui perbaikan proses dispatch pagi dan penyesuaian jadwal pergantian shift sore?*

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Catatan:</strong> angka target pastinya ditetapkan oleh manajemen. Tingkat kegagalan CBD di jam lain (5,5%) dipakai sebagai acuan karena menunjukkan bahwa zona yang sama bisa melayani dengan baik di luar jam puncak. Batas waktu 1 kuartal adalah usulan kami.
</p>

## <span style="color:#0061AF">Kesimpulan</span>

<p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
Berdasarkan seluruh analisis atas data Oktober–Desember 2023, diperoleh kesimpulan berikut:
</p>

1. **Request gagal terkonsentrasi di CBD pada 07–08 dan 17–18.** CBD pada jam tersebut gagal 36,6% dibanding 5,5% pada jam lain di zona yang sama (2.4.4, 3.4.1–3.4.3). Pola ini juga muncul di data Desember yang tidak dipakai untuk memilih jam (36,7% vs 5,5%; 2.7.3). Kegagalan berlebih di CBD sebanyak 6.346 request (2.4.7).
2. **Dua puncak itu kemungkinan punya penyebab berbeda.** Puncak sore bertepatan dengan pergantian shift: pada skenario kapasitas, pengemudi yang bertugas hanya kurang dari kebutuhan pada pukul 16–17, saat shift pagi berakhir (2.5.11, 3.4.8). Ini masih hipotesis, karena skenario tersebut bergantung pada asumsi tingkat sibuk pengemudi. Kegagalan pagi (07–08) tidak dijelaskan oleh jumlah pengemudi yang bertugas: di CBD ada 564–731 shift aktif dengan hanya 0,14 request per pengemudi per jam. Dugaan terkuat adalah proses dispatch atau kesiapan pengemudi di sistem, tetapi ini juga hipotesis yang perlu log dispatch. Request App lebih sering gagal daripada Call Center di setiap zona dan jam (2.4.8), dan kecepatan perjalanan hampir rata sepanjang hari, sehingga kemacetan tidak terlihat sebagai penjelasan (2.3.2).
3. **Jam shift jauh melebihi permintaan yang tercatat.** Utilisasi 9,4% dan 1,26 trip per shift. Pada skenario kapasitas, permintaan tercatat hanya butuh sekitar 18,9% jam-pengemudi yang ada jika pengemudi sibuk 50% (31,5% jika sibuk 30%) (2.5.11). Angka ini bergantung pada asumsi tingkat sibuk, dan data kemungkinan berupa sampel (2.2.3), sehingga belum bisa disimpulkan bahwa armada pasti berlebih sebelum skala data dikonfirmasi. Jam tambahan tidak menambah trip (shift hingga 7 jam dan 9–10 jam sama-sama 1,26 trip), dan tidak ada pengemudi yang konsisten berbeda (2.5.13). Perbedaan antar pool sejalan dengan komposisi zona yang tetap untuk setiap pool (2.5.14); setelah zona diperhitungkan, pool tidak lagi berpengaruh (F-test p = 0,62; dengan standard error per pengemudi p = 0,78). Dalam pendapatan, Airport tidak seburuk ritasenya (73,9% pendapatan shift CBD) dan Residential sama rendahnya per jam (2.5.12).
4. **Ritase rendah tetapi tidak menurun.** Tidak ada tren dalam 12 minggu (3.4.5). Penurunan pada brief tidak bisa dikonfirmasi dengan tiga bulan data. Sisa variasi ritase antar shift adalah keacakan data hitung (model mencapai 10,4% dari maksimum sekitar 10,5%; 2.7.3).
5. **Satu pola eVoucher yang nyata: Airport di akhir pekan.** 51,4% trip eVoucher Airport terjadi Sabtu–Minggu vs 17,6% trip Airport lain, seragam di 200 perusahaan (2.6.5). Pada akhir pekan, trip yang dimulai di Airport menjadi 54,6% trip dan 71,2% tarif eVoucher, sedangkan median tarif per trip tidak berubah (2.6.11, 3.5.12). Ini masalah aturan dan kontrol, bukan bukti penyalahgunaan; 1.290 trip diprioritaskan untuk tinjauan sampel (2.6.7). Tarif eVoucher juga tidak lebih tinggi, baik per trip maupun per km (2.6.7, 2.3.2), dan tidak ditemukan tanda kolusi pengemudi–perusahaan; pasangan transaksi berdekatan hanya sinyal lemah (337 trip untuk dicek; 2.6.10). Skenario CBD-only tidak berdasar brief dan tidak menunjukkan perbedaan berarti (2.6.6).
6. **Kualitas data meter terkendali.** 7.091 trip (2,2%) dengan jarak atau delay negatif dikeluarkan; komposisinya hampir sama dengan trip yang disimpan (selisih terbesar 1,0 poin) dan dampak ke rata-rata di bawah 0,2% (1.8.4–1.8.5). 0,38% trip dengan jarak tidak andal masih menaikkan rata-rata jarak mentah sekitar 18%, sehingga kolom <code>*_clean</code> wajib dipakai (2.3.1). Pada jarak yang andal, tarif sangat konsisten dengan jarak (R-squared 0,97–0,98 per layanan; 2.3.2), sehingga argometer dapat dipercaya untuk analisis tarif.

**Keterbatasan:** hanya tiga bulan data; semua hasil adalah asosiasi, bukan sebab-akibat; skala dan keteraturan data menunjukkan kemungkinan sampel atau simulasi, sehingga angka absolut perlu dikonfirmasi (2.2.3); request gagal tidak membawa info pengemudi atau perusahaan; tidak ada log dispatch dan zona tujuan; "waktu tidak lazim", batas jarak 500 km / 1 km, dan tingkat utilisasi skenario adalah asumsi kami; angka target ritase tidak disebut di brief (batas 6 trip pada <code>ritase_target_met</code> dibaca dari data).

<strong>Siapa yang menjalankan.</strong> Manajer operasional armada: rekomendasi 1, 2, 3, dan 6 (1 dan 2 untuk jangka pendek, 3 dan 6 untuk jangka menengah). <p style="text-indent: 30px; font-family: 'Roboto', sans-serif; font-size: 16px; line-height: 1.5;">
<strong>Arti jenis rekomendasi.</strong> <em>Segera</em> berarti buktinya sudah cukup dan risikonya kecil. <em>Investigasi</em> berarti masih perlu data tambahan sebelum diputuskan. <em>Uji strategis</em> berarti masih berupa dugaan yang perlu diuji coba dulu dengan kelompok pembanding sebelum diterapkan lebih luas. <strong>Risiko kalau langsung dijalankan tanpa uji coba:</strong> mengurangi shift (rekomendasi 3) sebelum skala data dipastikan bisa membuat kegagalan naik; menambah pengemudi di pagi hari tidak akan menyelesaikan masalah kalau penyebabnya ternyata dispatch; dan menyebut trip eVoucher sebagai kecurangan sebelum ada aturan bisa merusak hubungan dengan klien. Kami tidak mencantumkan angka penghematan karena data biaya per shift tidak tersedia.
